```python

# ============================================================
# BIML 2.0 — CELL 1: ENVIRONMENT AND PROJECT CONFIGURATION
# ============================================================

from pathlib import Path
import sys
import warnings
import platform

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn

warnings.filterwarnings("ignore", category=FutureWarning)

SEED = 42
np.random.seed(SEED)

PROJECT = Path.cwd()
DATA_DIR = PROJECT / "data"
OUTPUT_DIR = PROJECT / "outputs"
FIG_DIR = OUTPUT_DIR / "figures"
TABLE_DIR = OUTPUT_DIR / "tables"

for folder in [DATA_DIR, OUTPUT_DIR, FIG_DIR, TABLE_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

ENV_FILE = DATA_DIR / "Environmental_2013_2024_with_ERA5_T2M.csv"
INC_FILE = DATA_DIR / "Incidence_history_32countries_2012_2024.csv"

print("=" * 65)
print("BIML 2.0 — REPRODUCIBLE RESEARCH ENVIRONMENT")
print("=" * 65)
print("Python:", sys.version.split()[0])
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Scikit-learn:", sklearn.__version__)
print("Platform:", platform.platform())
print("Project directory:", PROJECT)
print("Environment file exists:", ENV_FILE.exists())
print("Incidence file exists:", INC_FILE.exists())
```

```python

# ============================================================
# CELL 2: LOAD ORIGINAL DATA WITHOUT ALTERING SOURCE FILES
# ============================================================

env = pd.read_csv(ENV_FILE)
inc = pd.read_csv(INC_FILE)

print("=" * 65)
print("ORIGINAL DATASET STRUCTURE")
print("=" * 65)

for name, df in [("ENVIRONMENT", env), ("INCIDENCE", inc)]:
    print(f"\n{name}")
    print("Shape:", df.shape)
    print("Countries:", df["Country"].nunique())
    print("Years:", df["Year"].min(), "-", df["Year"].max())
    print("Duplicate rows:", df.duplicated().sum())
    print("\nColumns:")
    print(df.dtypes.to_string())
    print("\nMissing values:")
    print(df.isna().sum().to_string())

required_env = [
    "Country", "Year", "Month", "Date",
    "ERA5_T2M_C", "Rainfall_mm", "RH_pct",
    "Landsat_LST_C", "MODIS_Night_LST_C",
    "NDVI", "NDWI"
]

required_inc = [
    "Country", "Year",
    "Malaria_incidence_per_1000", "Incidence_lag1"
]

assert set(required_env).issubset(env.columns)
assert set(required_inc).issubset(inc.columns)

print("\nRequired columns verified.")
```

```python

from pathlib import Path
import pandas as pd
import numpy as np

# ============================================================
# BIML 2.0 — CELL 13
# WHO INDICATOR AVAILABILITY AND COUNTRY HARMONIZATION
# ============================================================
env = pd.read_csv(ENV_FILE)
# Adjust this path if the downloaded WHO CSVs are elsewhere.
WHO_DIR = DATA_DIR

who_files = {
    "incidence_a": "malaria-incidence-(per-1-000-population-at-risk).csv",
    "incidence_b": "malaria-incidence-(per-1-000-population-at-risk)WHO.csv",
    "itn_use": "Malaria ITN use: population in malaria-endemic areas who slept_under_an_insecticide-treated bed_net (ITN) the previous night (%).csv",
    "itn_access": "Malaria protection_insecticide-treated_bed_net _ITN) _population with_access_modelled.csv",
    "treatment": "Malaria_treatment_cases_treated_with_any _irst _line_tx_courses_(including_artemisinin-based_combination_therapies_(ACTs))_number.csv",
    "confirmed": "number-confirmed-malaria-cases.csv",
    "irs": "number-of-people-protected-from-malaria-by-indoor-residual-spraying-irs.csv",
    "suspected": "number-of-suspected-malaria-cases.csv"
}

# Verify that all source files are present.
missing_files = [
    name for name, filename in who_files.items()
    if not (WHO_DIR / filename).exists()
]

if missing_files:
    raise FileNotFoundError(
        f"Missing WHO files in {WHO_DIR}: {missing_files}. "
        "Copy the downloaded files into the notebook's data/ folder."
    )

# Read and standardize.
who_data = {}
audit_rows = []

for name, filename in who_files.items():

    df = pd.read_csv(WHO_DIR / filename)

    df = df[
        [
            "SpatialDimValueCode",
            "Location",
            "Period",
            "IndicatorCode",
            "FactValueNumeric"
        ]
    ].copy()

    df = df.rename(columns={
        "SpatialDimValueCode": "CountryCode",
        "Location": "Country_WHO",
        "Period": "Year",
        "FactValueNumeric": name
    })

    df["Year"] = pd.to_numeric(
        df["Year"], errors="coerce"
    )

    df[name] = pd.to_numeric(
        df[name], errors="coerce"
    )

    df = df[df["Year"].between(2013, 2024)].copy()
    df["Year"] = df["Year"].astype(int)

    duplicates = df.duplicated(
        ["CountryCode", "Year"]
    ).sum()

    if duplicates:
        raise ValueError(
            f"{name} contains {duplicates} duplicate country-years."
        )

    who_data[name] = df

    audit_rows.append({
        "indicator": name,
        "countries": df["CountryCode"].nunique(),
        "country_years": len(df),
        "nonmissing_values": df[name].notna().sum(),
        "first_year": df["Year"].min(),
        "last_year": df["Year"].max()
    })

audit = pd.DataFrame(audit_rows)

print("=" * 85)
print("WHO INDICATOR COVERAGE — 2013–2024")
print("=" * 85)
print(audit.to_string(index=False))

# ------------------------------------------------------------
# Merge WHO datasets using ISO-style country codes
# ------------------------------------------------------------

who_panel = None

for name, df in who_data.items():

    current = df[
        ["CountryCode", "Year", name]
    ].copy()

    if who_panel is None:
        who_panel = current
    else:
        who_panel = who_panel.merge(
            current,
            on=["CountryCode", "Year"],
            how="outer",
            validate="one_to_one"
        )

# Map original BIML country names to WHO country codes
# using country-name matching, then inspect exceptions.
country_lookup = (
    pd.concat([
        df[["CountryCode", "Country_WHO"]]
        for df in who_data.values()
    ])
    .drop_duplicates()
)

country_lookup["Country_normalized"] = (
    country_lookup["Country_WHO"]
    .str.strip()
    .str.casefold()
)

env_countries = pd.DataFrame({
    "Country": sorted(env["Country"].unique())
})

env_countries["Country_normalized"] = (
    env_countries["Country"]
    .str.strip()
    .str.casefold()
)

country_matches = env_countries.merge(
    country_lookup,
    on="Country_normalized",
    how="left"
)

print("\n" + "=" * 85)
print("ORIGINAL BIML COUNTRIES WITHOUT AN EXACT WHO NAME MATCH")
print("=" * 85)

unmatched = country_matches[
    country_matches["CountryCode"].isna()
]

print(
    unmatched["Country"].drop_duplicates().to_string(index=False)
    if len(unmatched)
    else "None"
)

# ------------------------------------------------------------
# Compare the two incidence files on shared keys
# ------------------------------------------------------------

comparison = who_data["incidence_a"][
    ["CountryCode", "Year", "incidence_a"]
].merge(
    who_data["incidence_b"][
        ["CountryCode", "Year", "incidence_b"]
    ],
    on=["CountryCode", "Year"],
    how="inner"
)

comparison["absolute_difference"] = (
    comparison["incidence_a"]
    - comparison["incidence_b"]
).abs()

print("\n" + "=" * 85)
print("WHO INCIDENCE FILE COMPARISON")
print("=" * 85)

print("Shared country-years:", len(comparison))
print(
    "Maximum absolute difference:",
    comparison["absolute_difference"].max()
)
print(
    "Country-years differing by more than 0.01:",
    (comparison["absolute_difference"] > 0.01).sum()
)

# ------------------------------------------------------------
# Coverage of intervention indicators
# ------------------------------------------------------------

predictors = [
    "itn_use", "itn_access",
    "treatment", "irs"
]

coverage = (
    who_panel.groupby("Year")[predictors]
    .count()
)

print("\n" + "=" * 85)
print("ANNUAL INTERVENTION COVERAGE — ALL WHO COUNTRIES")
print("=" * 85)
print(coverage.to_string())

# Save audit outputs.
audit.to_csv(
    TABLE_DIR / "who_indicator_audit.csv",
    index=False
)

country_matches.to_csv(
    TABLE_DIR / "who_country_name_matches.csv",
    index=False
)

comparison.to_csv(
    TABLE_DIR / "who_incidence_comparison.csv",
    index=False
)

who_panel.to_csv(
    DATA_DIR / "BIML_WHO_raw_merged_audit.csv",
    index=False
)

print("\nCELL 13 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 14
# COUNTRY OVERLAP AND USABLE PANEL ASSESSMENT
# ============================================================

import pandas as pd
import numpy as np

# Explicit mapping for the known naming difference
country_name_fixes = {
    "Côte d'Ivoire": "Cote d'Ivoire"
}

# Build WHO name-to-code lookup
who_lookup = country_lookup[
    ["CountryCode", "Country_WHO"]
].drop_duplicates()

who_lookup["normalized_name"] = (
    who_lookup["Country_WHO"]
    .str.strip()
    .str.casefold()
)

env_lookup = pd.DataFrame({
    "Country": sorted(env["Country"].unique())
})

env_lookup["WHO_name"] = (
    env_lookup["Country"]
    .replace(country_name_fixes)
)

env_lookup["normalized_name"] = (
    env_lookup["WHO_name"]
    .str.strip()
    .str.casefold()
)

env_lookup = env_lookup.merge(
    who_lookup[["CountryCode", "normalized_name"]],
    on="normalized_name",
    how="left",
    validate="many_to_one"
)

# Verify and report mapping
print("=" * 80)
print("ENVIRONMENTAL COUNTRY MAPPING")
print("=" * 80)

print(
    env_lookup[
        ["Country", "CountryCode"]
    ].to_string(index=False)
)

unmapped = env_lookup[
    env_lookup["CountryCode"].isna()
]

if len(unmapped):
    print("\nWARNING: Unmapped countries:")
    print(unmapped["Country"].tolist())

# ------------------------------------------------------------
# Aggregate environmental variables annually
# ------------------------------------------------------------

environmental_variables = [
    "ERA5_T2M_C",
    "Rainfall_mm",
    "RH_pct",
    "NDVI",
    "NDWI",
    "MODIS_Night_LST_C",
    "Landsat_LST_C"
]

annual_env = (
    env.groupby(["Country", "Year"])
    .agg(
        ERA5_T2M_C_mean=("ERA5_T2M_C", "mean"),
        Rainfall_total_mm=("Rainfall_mm", "sum"),
        RH_pct_mean=("RH_pct", "mean"),
        NDVI_mean=("NDVI", "mean"),
        NDWI_mean=("NDWI", "mean"),
        MODIS_Night_LST_C_mean=(
            "MODIS_Night_LST_C", "mean"
        ),
        Landsat_LST_C_mean=(
            "Landsat_LST_C", "mean"
        )
    )
    .reset_index()
)

annual_env = annual_env.merge(
    env_lookup[["Country", "CountryCode"]],
    on="Country",
    how="left",
    validate="many_to_one"
)

# ------------------------------------------------------------
# Merge WHO target and interventions with environment
# ------------------------------------------------------------

core = who_panel[
    [
        "CountryCode", "Year",
        "incidence_b", "itn_use", "itn_access"
    ]
].copy()

merged = annual_env.merge(
    core,
    on=["CountryCode", "Year"],
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# Compare candidate periods
# ------------------------------------------------------------

candidate_periods = [
    (2013, 2024),
    (2015, 2024),
    (2018, 2024)
]

summary = []

for start, end in candidate_periods:

    subset = merged[
        merged["Year"].between(start, end)
    ].copy()

    required = [
        "incidence_b",
        "itn_use",
        "itn_access",
        "ERA5_T2M_C_mean",
        "Rainfall_total_mm",
        "RH_pct_mean"
    ]

    complete = subset.dropna(subset=required)

    summary.append({
        "period": f"{start}-{end}",
        "total_country_years": len(subset),
        "complete_core_rows": len(complete),
        "complete_core_countries": (
            complete["CountryCode"].nunique()
        ),
        "countries_with_all_years": sum(
            complete.groupby("CountryCode")["Year"].nunique()
            == (end - start + 1)
        )
    })

period_summary = pd.DataFrame(summary)

print("\n" + "=" * 80)
print("CANDIDATE STUDY PERIODS")
print("=" * 80)

print(period_summary.to_string(index=False))

# ------------------------------------------------------------
# Per-country coverage: 2015–2024
# ------------------------------------------------------------

focus = merged[
    merged["Year"].between(2015, 2024)
].copy()

coverage = (
    focus.groupby(["Country", "CountryCode"])
    .agg(
        incidence_years=("incidence_b", "count"),
        itn_use_years=("itn_use", "count"),
        itn_access_years=("itn_access", "count"),
        climate_years=("ERA5_T2M_C_mean", "count")
    )
    .reset_index()
)

print("\n" + "=" * 80)
print("COUNTRY COVERAGE — 2015–2024")
print("=" * 80)

print(coverage.to_string(index=False))

# ------------------------------------------------------------
# Save audit results
# ------------------------------------------------------------

period_summary.to_csv(
    TABLE_DIR / "candidate_study_periods.csv",
    index=False
)

coverage.to_csv(
    TABLE_DIR / "country_core_coverage.csv",
    index=False
)

merged.to_csv(
    DATA_DIR / "BIML_WHO_environmental_coverage_audit.csv",
    index=False
)

print("\nCELL 14 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 15 (REVISED)
# WHO-BASED CORE DATASET VALIDATION
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# A. Define the new WHO-based study panel
# ------------------------------------------------------------

required_variables = [
    "incidence_b",
    "itn_use",
    "itn_access",
    "ERA5_T2M_C_mean",
    "Rainfall_total_mm",
    "RH_pct_mean"
]

core_panel = merged.loc[
    merged["Year"].between(2015, 2024)
].dropna(subset=required_variables).copy()

core_panel = core_panel.sort_values(
    ["CountryCode", "Year"]
).reset_index(drop=True)

# ------------------------------------------------------------
# B. Validate panel structure
# ------------------------------------------------------------

assert not core_panel.duplicated(
    ["CountryCode", "Year"]
).any(), "Duplicate country-year observations found."

country_year_counts = (
    core_panel.groupby("CountryCode")["Year"].nunique()
)

print("=" * 85)
print("BIML 2.0 — WHO CORE DATASET VALIDATION")
print("=" * 85)

print("Countries:", core_panel["CountryCode"].nunique())
print("Observations:", len(core_panel))
print(
    "Study period:",
    core_panel["Year"].min(),
    "to",
    core_panel["Year"].max()
)

print(
    "Balanced panel:",
    bool(country_year_counts.nunique() == 1)
)

print(
    "Years per country:",
    sorted(country_year_counts.unique())
)

# ------------------------------------------------------------
# C. Verify all expected country-years exist
# ------------------------------------------------------------

expected_years = set(range(2015, 2025))

incomplete_countries = {}

for country, group in core_panel.groupby("CountryCode"):

    missing_years = sorted(
        expected_years - set(group["Year"])
    )

    if missing_years:
        incomplete_countries[country] = missing_years

print("\nIncomplete countries:", incomplete_countries)

# ------------------------------------------------------------
# D. Inspect missing values
# ------------------------------------------------------------

audit_columns = [
    "incidence_b",
    "itn_use",
    "itn_access",
    "ERA5_T2M_C_mean",
    "Rainfall_total_mm",
    "RH_pct_mean",
    "NDVI_mean",
    "NDWI_mean",
    "MODIS_Night_LST_C_mean",
    "Landsat_LST_C_mean"
]

missing_audit = pd.DataFrame({
    "missing": core_panel[audit_columns].isna().sum(),
    "missing_pct": (
        core_panel[audit_columns].isna().mean() * 100
    )
})

print("\n" + "=" * 85)
print("MISSING VALUE AUDIT")
print("=" * 85)

print(missing_audit.round(2).to_string())

# ------------------------------------------------------------
# E. Validate WHO incidence target
# ------------------------------------------------------------

target = core_panel["incidence_b"]

print("\n" + "=" * 85)
print("WHO INCIDENCE TARGET DISTRIBUTION")
print("=" * 85)

print(target.describe().round(4).to_string())

print(
    "\nNegative incidence values:",
    int((target < 0).sum())
)

print(
    "Zero incidence values:",
    int((target == 0).sum())
)

# ------------------------------------------------------------
# F. Inspect intervention ranges
# ------------------------------------------------------------

print("\n" + "=" * 85)
print("INTERVENTION INDICATOR DISTRIBUTIONS")
print("=" * 85)

print(
    core_panel[
        ["itn_use", "itn_access"]
    ].describe().round(4).to_string()
)

# Percentage indicators should be within 0–100.
for variable in ["itn_use", "itn_access"]:

    invalid = (
        (core_panel[variable] < 0)
        | (core_panel[variable] > 100)
    ).sum()

    print(f"{variable} outside 0–100: {invalid}")

# ------------------------------------------------------------
# G. Save validated WHO-based panel
# ------------------------------------------------------------

output_path = (
    DATA_DIR / "BIML_WHO_core_2015_2024.csv"
)

core_panel.to_csv(
    output_path,
    index=False
)

missing_audit.to_csv(
    TABLE_DIR / "BIML_WHO_core_missingness.csv"
)

print("\nSaved:", output_path)
print("\nCELL 15 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 16
# ITN REDUNDANCY AND WITHIN-COUNTRY DIAGNOSTICS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = core_panel.copy()
df = df.sort_values(["CountryCode", "Year"])

variables = ["incidence_b", "itn_use", "itn_access"]

# ------------------------------------------------------------
# A. Overall correlations
# ------------------------------------------------------------

print("=" * 80)
print("OVERALL PEARSON CORRELATIONS")
print("=" * 80)

print(
    df[variables].corr(method="pearson")
    .round(4).to_string()
)

print("\nOVERALL SPEARMAN CORRELATIONS")

print(
    df[variables].corr(method="spearman")
    .round(4).to_string()
)

# ------------------------------------------------------------
# B. Within-country anomalies
# ------------------------------------------------------------

for variable in variables:

    df[f"{variable}_within"] = (
        df[variable]
        - df.groupby("CountryCode")[variable]
        .transform("mean")
    )

within_variables = [
    f"{variable}_within"
    for variable in variables
]

print("\n" + "=" * 80)
print("WITHIN-COUNTRY PEARSON CORRELATIONS")
print("=" * 80)

print(
    df[within_variables]
    .corr(method="pearson")
    .round(4).to_string()
)

# ------------------------------------------------------------
# C. Year-to-year changes
# ------------------------------------------------------------

for variable in variables:

    df[f"{variable}_change"] = (
        df.groupby("CountryCode")[variable]
        .diff()
    )

change_variables = [
    f"{variable}_change"
    for variable in variables
]

print("\n" + "=" * 80)
print("YEAR-TO-YEAR CHANGE CORRELATIONS")
print("=" * 80)

print(
    df[change_variables]
    .corr(method="pearson")
    .round(4).to_string()
)

# ------------------------------------------------------------
# D. Quantify redundancy
# ------------------------------------------------------------

df["itn_gap"] = (
    df["itn_access"] - df["itn_use"]
)

print("\n" + "=" * 80)
print("ITN ACCESS–USE GAP")
print("=" * 80)

print(
    df["itn_gap"].describe()
    .round(4).to_string()
)

print(
    "\nNegative gaps:",
    int((df["itn_gap"] < 0).sum())
)

# ------------------------------------------------------------
# E. Visualization
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.scatterplot(
    data=df,
    x="itn_access",
    y="itn_use",
    ax=axes[0],
    alpha=0.7
)

axes[0].set_title("ITN Access vs ITN Use")
axes[0].set_xlabel("ITN access (%)")
axes[0].set_ylabel("ITN use (%)")

sns.scatterplot(
    data=df,
    x="itn_use_within",
    y="incidence_b_within",
    ax=axes[1],
    alpha=0.7
)

axes[1].axhline(0, color="gray", linestyle="--")
axes[1].axvline(0, color="gray", linestyle="--")

axes[1].set_title(
    "Within-Country ITN Use vs Incidence"
)
axes[1].set_xlabel(
    "ITN use deviation from country mean"
)
axes[1].set_ylabel(
    "Incidence deviation from country mean"
)

plt.tight_layout()

plt.savefig(
    FIG_DIR / "itn_incidence_diagnostics.png",
    dpi=250,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# F. Save outputs
# ------------------------------------------------------------

df.to_csv(
    DATA_DIR / "BIML_ITN_diagnostic_panel.csv",
    index=False
)

print("\nCELL 16 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 17
# ADDITIONAL WHO INDICATOR COVERAGE AUDIT
# ============================================================

import pandas as pd
import numpy as np

additional_variables = [
    "treatment",
    "confirmed",
    "suspected",
    "irs"
]

# Use the validated 29-country panel as the reference.
keys = core_panel[
    ["Country", "CountryCode", "Year"]
].copy()

# Join additional WHO indicators without removing rows.
additional_panel = keys.merge(
    who_panel[
        ["CountryCode", "Year"] + additional_variables
    ],
    on=["CountryCode", "Year"],
    how="left",
    validate="one_to_one"
)

assert len(additional_panel) == 290
assert not additional_panel.duplicated(
    ["CountryCode", "Year"]
).any()

# ------------------------------------------------------------
# A. Overall availability
# ------------------------------------------------------------

summary = []

for variable in additional_variables:
    s = additional_panel[variable]

    summary.append({
        "indicator": variable,
        "available_rows": int(s.notna().sum()),
        "coverage_pct": round(100*s.notna().mean(), 2),
        "countries_with_data": int(
            additional_panel.loc[
                s.notna(), "CountryCode"
            ].nunique()
        ),
        "min": s.min(),
        "median": s.median(),
        "max": s.max(),
        "negative_values": int((s < 0).sum())
    })

summary = pd.DataFrame(summary)

print("=" * 85)
print("ADDITIONAL WHO INDICATORS — 29-COUNTRY PANEL")
print("=" * 85)
print(summary.round(2).to_string(index=False))

# ------------------------------------------------------------
# B. Availability by year
# ------------------------------------------------------------

yearly_coverage = (
    additional_panel.groupby("Year")[additional_variables]
    .count()
)

print("\n" + "=" * 85)
print("AVAILABLE COUNTRY RECORDS BY YEAR")
print("=" * 85)
print(yearly_coverage.to_string())

# ------------------------------------------------------------
# C. Availability by country
# ------------------------------------------------------------

country_coverage = (
    additional_panel.groupby(
        ["Country", "CountryCode"]
    )[additional_variables]
    .count()
    .reset_index()
    .sort_values("Country")
)

print("\n" + "=" * 85)
print("AVAILABLE YEARS BY COUNTRY")
print("=" * 85)
print(country_coverage.to_string(index=False))

# ------------------------------------------------------------
# D. Assess complete-case sample loss
# ------------------------------------------------------------

print("\n" + "=" * 85)
print("SAMPLE SIZE IF ADDITIONAL INDICATORS ARE REQUIRED")
print("=" * 85)

for variable in additional_variables:
    complete = additional_panel.dropna(subset=[variable])
    print(
        f"{variable:12s}: "
        f"{len(complete):3d} rows, "
        f"{complete.CountryCode.nunique():2d} countries"
    )

complete_all = additional_panel.dropna(
    subset=additional_variables
)

print(
    f"\nAll four: {len(complete_all)} rows, "
    f"{complete_all.CountryCode.nunique()} countries"
)

# ------------------------------------------------------------
# E. Save audit outputs
# ------------------------------------------------------------

summary.to_csv(
    TABLE_DIR / "additional_who_indicator_summary.csv",
    index=False
)

yearly_coverage.to_csv(
    TABLE_DIR / "additional_who_yearly_coverage.csv"
)

country_coverage.to_csv(
    TABLE_DIR / "additional_who_country_coverage.csv",
    index=False
)

additional_panel.to_csv(
    DATA_DIR / "BIML_additional_WHO_audit.csv",
    index=False
)

print("\nCELL 17 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 18
# WHO INCIDENCE TEMPORAL DYNAMICS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = core_panel.copy()

df = df.sort_values(
    ["CountryCode", "Year"]
).reset_index(drop=True)

# ------------------------------------------------------------
# A. Construct previous-year incidence
# ------------------------------------------------------------

df["incidence_lag1"] = (
    df.groupby("CountryCode")["incidence_b"]
    .shift(1)
)

df["incidence_change"] = (
    df["incidence_b"] - df["incidence_lag1"]
)

df["absolute_change"] = df["incidence_change"].abs()

df["relative_change_pct"] = np.where(
    df["incidence_lag1"] > 0,
    100 * df["incidence_change"]
    / df["incidence_lag1"],
    np.nan
)

# ------------------------------------------------------------
# B. Persistence diagnostic
# ------------------------------------------------------------

valid = df.dropna(
    subset=["incidence_lag1"]
).copy()

mae = valid["absolute_change"].mean()

rmse = np.sqrt(
    np.mean(valid["incidence_change"] ** 2)
)

correlation = valid[
    ["incidence_b", "incidence_lag1"]
].corr().iloc[0, 1]

print("=" * 80)
print("WHO INCIDENCE TEMPORAL DYNAMICS")
print("=" * 80)

print("Valid transitions:", len(valid))
print("Countries:", valid["CountryCode"].nunique())
print("Persistence MAE:", round(mae, 4))
print("Persistence RMSE:", round(rmse, 4))
print("Incidence lag-1 correlation:", round(correlation, 4))

print("\nAnnual incidence changes:")

print(
    valid["incidence_change"]
    .describe().round(4).to_string()
)

# ------------------------------------------------------------
# C. Country-level dynamics
# ------------------------------------------------------------

country_summary = (
    valid.groupby(["Country", "CountryCode"])
    .agg(
        mean_incidence=("incidence_b", "mean"),
        mean_absolute_change=("absolute_change", "mean"),
        max_absolute_change=("absolute_change", "max")
    )
    .reset_index()
    .sort_values(
        "mean_absolute_change",
        ascending=False
    )
)

print("\n" + "=" * 80)
print("COUNTRY-LEVEL INCIDENCE DYNAMICS")
print("=" * 80)

print(
    country_summary.round(3).to_string(index=False)
)

# ------------------------------------------------------------
# D. Year-level dynamics
# ------------------------------------------------------------

year_summary = (
    valid.groupby("Year")
    .agg(
        mean_incidence=("incidence_b", "mean"),
        mean_absolute_change=("absolute_change", "mean"),
        median_absolute_change=("absolute_change", "median")
    )
)

print("\n" + "=" * 80)
print("YEAR-LEVEL INCIDENCE DYNAMICS")
print("=" * 80)

print(year_summary.round(3).to_string())

# ------------------------------------------------------------
# E. Plot incidence trajectories
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 6))

for country, group in df.groupby("Country"):
    ax.plot(
        group["Year"],
        group["incidence_b"],
        alpha=0.35,
        linewidth=1
    )

ax.set_title(
    "WHO Malaria Incidence Trajectories — 29 Countries"
)
ax.set_xlabel("Year")
ax.set_ylabel(
    "Incidence per 1,000 population at risk"
)

ax.grid(alpha=0.2)

plt.tight_layout()

plt.savefig(
    FIG_DIR / "who_incidence_trajectories.png",
    dpi=250,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# F. Save results
# ------------------------------------------------------------

df.to_csv(
    DATA_DIR / "BIML_WHO_temporal_panel.csv",
    index=False
)

country_summary.to_csv(
    TABLE_DIR / "who_country_incidence_dynamics.csv",
    index=False
)

year_summary.to_csv(
    TABLE_DIR / "who_year_incidence_dynamics.csv"
)

print("\nCELL 18 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 20
# FORECAST-READY FEATURE ALIGNMENT AUDIT
# ============================================================

import pandas as pd
import numpy as np

df = core_panel.copy()

df = df.sort_values(
    ["CountryCode", "Year"]
).reset_index(drop=True)

selected_predictors = [
    "incidence_b",
    "itn_access",
    "ERA5_T2M_C_mean",
    "Rainfall_total_mm",
    "RH_pct_mean",
    "NDVI_mean",
    "NDWI_mean"
]

# ------------------------------------------------------------
# A. Construct previous-year predictors
# ------------------------------------------------------------

for variable in selected_predictors:

    df[f"{variable}_lag1"] = (
        df.groupby("CountryCode")[variable]
        .shift(1)
    )

# Verify that lagged values correspond to the preceding year.
df["previous_year"] = (
    df.groupby("CountryCode")["Year"].shift(1)
)

valid_previous_year = (
    df["previous_year"] == df["Year"] - 1
)

lagged_columns = [
    f"{variable}_lag1"
    for variable in selected_predictors
]

df.loc[
    ~valid_previous_year, lagged_columns
] = np.nan

# ------------------------------------------------------------
# B. Evaluate usable sample
# ------------------------------------------------------------

forecast_panel = df.dropna(
    subset=lagged_columns + ["incidence_b"]
).copy()

print("=" * 80)
print("ONE-YEAR-AHEAD FORECAST FEATURE AUDIT")
print("=" * 80)

print("Original observations:", len(df))
print("Forecast-ready observations:", len(forecast_panel))
print("Countries:", forecast_panel["CountryCode"].nunique())
print(
    "Target years:",
    forecast_panel["Year"].min(),
    "to",
    forecast_panel["Year"].max()
)

print("\nObservations by target year:")
print(
    forecast_panel.groupby("Year")
    .size().to_string()
)

# ------------------------------------------------------------
# C. Inspect feature completeness
# ------------------------------------------------------------

print("\nMissing lagged predictors:")

print(
    df[lagged_columns]
    .isna().sum().to_string()
)

# ------------------------------------------------------------
# D. Verify target and historical incidence alignment
# ------------------------------------------------------------

print("\nExample forecast records:")

print(
    forecast_panel[
        [
            "Country", "Year",
            "incidence_b_lag1",
            "incidence_b",
            "itn_access_lag1",
            "Rainfall_total_mm_lag1"
        ]
    ].head(12).round(3).to_string(index=False)
)

# ------------------------------------------------------------
# E. Save forecast-ready panel
# ------------------------------------------------------------

forecast_panel.to_csv(
    DATA_DIR / "BIML_WHO_forecast_panel.csv",
    index=False
)

print("\nCELL 20 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 21
# FINAL FORECAST PREDICTOR INTEGRITY AUDIT
# ============================================================

import numpy as np
import pandas as pd

df = forecast_panel.copy()

target = "incidence_b"

features = [
    "incidence_b_lag1",
    "itn_access_lag1",
    "ERA5_T2M_C_mean_lag1",
    "Rainfall_total_mm_lag1",
    "RH_pct_mean_lag1",
    "NDVI_mean_lag1",
    "NDWI_mean_lag1"
]

# ------------------------------------------------------------
# A. Structural validation
# ------------------------------------------------------------

assert not df.duplicated(
    ["CountryCode", "Year"]
).any()

assert df[target].notna().all()
assert (df[target] >= 0).all()

assert df[features].notna().all().all()

assert np.isfinite(
    df[features + [target]].to_numpy(dtype=float)
).all()

print("=" * 80)
print("BIML 2.0 — FINAL PREDICTOR INTEGRITY AUDIT")
print("=" * 80)

print("Observations:", len(df))
print("Countries:", df["CountryCode"].nunique())
print("Years:", df["Year"].min(), "-", df["Year"].max())

# ------------------------------------------------------------
# B. Feature distributions
# ------------------------------------------------------------

print("\nFEATURE DISTRIBUTIONS")

print(
    df[features]
    .describe()
    .T[["mean", "std", "min", "max"]]
    .round(4)
    .to_string()
)

# ------------------------------------------------------------
# C. Near-constant features
# ------------------------------------------------------------

near_constant = []

for feature in features:

    if df[feature].nunique() <= 2:
        near_constant.append(feature)

print("\nNear-constant features:", near_constant)

# ------------------------------------------------------------
# D. Feature redundancy
# ------------------------------------------------------------

correlation = df[features].corr().abs()

upper = correlation.where(
    np.triu(
        np.ones(correlation.shape),
        k=1
    ).astype(bool)
)

redundant_pairs = []

for column in upper.columns:

    for index, value in upper[column].dropna().items():

        if value >= 0.95:

            redundant_pairs.append({
                "feature_1": index,
                "feature_2": column,
                "abs_pearson": value
            })

print("\nHIGHLY CORRELATED FEATURE PAIRS (|r| >= 0.95)")

if redundant_pairs:
    print(
        pd.DataFrame(redundant_pairs)
        .round(4).to_string(index=False)
    )
else:
    print("None")

# ------------------------------------------------------------
# E. Verify target-year distribution
# ------------------------------------------------------------

print("\nTARGET DISTRIBUTION BY YEAR")

print(
    df.groupby("Year")[target]
    .agg(["count", "mean", "std", "min", "max"])
    .round(3).to_string()
)

# ------------------------------------------------------------
# F. Save
# ------------------------------------------------------------

df.to_csv(
    DATA_DIR / "BIML_WHO_forecast_validated.csv",
    index=False
)

print("\nCELL 21 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 22
# PRIMARY TEMPORAL EVALUATION PARTITIONS
# ============================================================

import pandas as pd
import numpy as np

df = forecast_panel.copy()

# ------------------------------------------------------------
# A. Define chronological partitions
# ------------------------------------------------------------

train_years = list(range(2016, 2021))
val_years = [2021, 2022]
test_years = [2023, 2024]

df["partition"] = np.select(
    [
        df["Year"].isin(train_years),
        df["Year"].isin(val_years),
        df["Year"].isin(test_years)
    ],
    ["train", "validation", "test"],
    default="unassigned"
)

assert not (df["partition"] == "unassigned").any()

# ------------------------------------------------------------
# B. Verify partition integrity
# ------------------------------------------------------------

assert set(train_years).isdisjoint(val_years)
assert set(train_years).isdisjoint(test_years)
assert set(val_years).isdisjoint(test_years)

assert max(train_years) < min(val_years)
assert max(val_years) < min(test_years)

assert not df.duplicated(
    ["CountryCode", "Year"]
).any()

summary = (
    df.groupby("partition")
    .agg(
        observations=("Year", "size"),
        countries=("CountryCode", "nunique"),
        first_year=("Year", "min"),
        last_year=("Year", "max")
    )
    .reindex(["train", "validation", "test"])
)

summary["percentage"] = (
    100 * summary["observations"] / len(df)
)

print("=" * 80)
print("BIML 2.0 — PRIMARY TEMPORAL SPLIT")
print("=" * 80)

print(summary.round(2).to_string())

# ------------------------------------------------------------
# C. Verify balanced country coverage
# ------------------------------------------------------------

country_partition_counts = (
    df.groupby(["CountryCode", "partition"])
    .size()
    .unstack(fill_value=0)
)

assert (country_partition_counts["train"] == 5).all()
assert (country_partition_counts["validation"] == 2).all()
assert (country_partition_counts["test"] == 2).all()

print("\nAll 29 countries have:")
print("  5 training observations")
print("  2 validation observations")
print("  2 testing observations")

# ------------------------------------------------------------
# D. Preserve split assignments
# ------------------------------------------------------------

split_path = (
    DATA_DIR / "BIML_WHO_temporal_split.csv"
)

df.to_csv(split_path, index=False)

print("\nSaved:", split_path)
print("\nCELL 22 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 23
# PERSISTENCE BASELINE EVALUATION
# ============================================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    DATA_DIR / "BIML_WHO_temporal_split.csv"
)

assert not df.duplicated(
    ["CountryCode", "Year"]
).any()

assert df[
    ["incidence_b", "incidence_b_lag1", "partition"]
].notna().all().all()

# ------------------------------------------------------------
# A. Persistence prediction
# ------------------------------------------------------------

df["persistence_prediction"] = df["incidence_b_lag1"]

df["persistence_error"] = (
    df["incidence_b"] - df["persistence_prediction"]
)

# ------------------------------------------------------------
# B. Metrics by partition
# ------------------------------------------------------------

results = []

for partition in ["train", "validation", "test"]:

    subset = df[df["partition"] == partition]

    y_true = subset["incidence_b"].to_numpy()
    y_pred = subset["persistence_prediction"].to_numpy()

    results.append({
        "partition": partition,
        "n": len(subset),
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, y_pred)
        ),
        "R2": r2_score(y_true, y_pred),
        "Bias": np.mean(y_pred - y_true)
    })

results = pd.DataFrame(results)

print("=" * 80)
print("BIML 2.0 — PERSISTENCE BASELINE")
print("=" * 80)

print(results.round(4).to_string(index=False))

# ------------------------------------------------------------
# C. Country-level test performance
# ------------------------------------------------------------

test = df[df["partition"] == "test"].copy()

country_results = (
    test.groupby(["Country", "CountryCode"])
    .agg(
        MAE=("persistence_error", lambda x: x.abs().mean()),
        RMSE=("persistence_error", lambda x: np.sqrt(
            np.mean(x**2)
        )),
        Bias=("persistence_error", lambda x: -x.mean())
    )
    .reset_index()
    .sort_values("MAE", ascending=False)
)

print("\n" + "=" * 80)
print("COUNTRY-LEVEL TEST PERFORMANCE")
print("=" * 80)

print(
    country_results.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# D. Save outputs
# ------------------------------------------------------------

results.to_csv(
    TABLE_DIR / "persistence_baseline_metrics.csv",
    index=False
)

country_results.to_csv(
    TABLE_DIR / "persistence_country_test_metrics.csv",
    index=False
)

df.to_csv(
    DATA_DIR / "BIML_WHO_with_persistence.csv",
    index=False
)

print("\nCELL 23 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 24
# RIDGE REGRESSION — INCIDENCE HISTORY ONLY
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    DATA_DIR / "BIML_WHO_temporal_split.csv"
)

feature = ["incidence_b_lag1"]
target = "incidence_b"

train = df[df["partition"] == "train"].copy()
validation = df[df["partition"] == "validation"].copy()

X_train = train[feature]
y_train = train[target]

X_val = validation[feature]
y_val = validation[target]

# ------------------------------------------------------------
# A. Train candidate Ridge models
# ------------------------------------------------------------

alphas = [0.01, 0.1, 1, 10, 100, 1000]

results = []
models = {}

for alpha in alphas:

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=alpha))
    ])

    model.fit(X_train, y_train)

    predictions = model.predict(X_val)

    results.append({
        "alpha": alpha,
        "MAE": mean_absolute_error(
            y_val, predictions
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, predictions)
        ),
        "R2": r2_score(y_val, predictions)
    })

    models[alpha] = model

results = pd.DataFrame(results).sort_values("MAE")

best_alpha = results.iloc[0]["alpha"]
best_model = models[best_alpha]

# ------------------------------------------------------------
# B. Compare with persistence on validation only
# ------------------------------------------------------------

ridge_val = best_model.predict(X_val)

persistence_val = validation[
    "incidence_b_lag1"
].to_numpy()

print("=" * 80)
print("RIDGE REGRESSION — VALIDATION RESULTS")
print("=" * 80)

print(results.round(4).to_string(index=False))

print("\nSelected alpha:", best_alpha)

comparison = pd.DataFrame([
    {
        "model": "Persistence",
        "MAE": mean_absolute_error(
            y_val, persistence_val
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, persistence_val)
        )
    },
    {
        "model": "Ridge_history",
        "MAE": mean_absolute_error(
            y_val, ridge_val
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, ridge_val)
        )
    }
])

print("\nVALIDATION COMPARISON")
print(comparison.round(4).to_string(index=False))

# ------------------------------------------------------------
# C. Inspect fitted coefficients
# ------------------------------------------------------------

scaler = best_model.named_steps["scaler"]
ridge = best_model.named_steps["ridge"]

original_slope = (
    ridge.coef_[0] / scaler.scale_[0]
)

original_intercept = (
    ridge.intercept_
    - original_slope * scaler.mean_[0]
)

print("\nFITTED EQUATION")

print(
    f"Predicted incidence = "
    f"{original_intercept:.4f} + "
    f"{original_slope:.4f} × previous incidence"
)

# ------------------------------------------------------------
# D. Save validation results
# ------------------------------------------------------------

results.to_csv(
    TABLE_DIR / "ridge_history_validation.csv",
    index=False
)

comparison.to_csv(
    TABLE_DIR / "ridge_vs_persistence_validation.csv",
    index=False
)

print("\nCELL 24 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 25
# E1: ENVIRONMENTAL-ONLY RIDGE REGRESSION
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    DATA_DIR / "BIML_WHO_temporal_split.csv"
)

# ------------------------------------------------------------
# A. Define environmental predictors
# ------------------------------------------------------------

environment_features = [
    "ERA5_T2M_C_mean_lag1",
    "Rainfall_total_mm_lag1",
    "RH_pct_mean_lag1",
    "NDVI_mean_lag1",
    "NDWI_mean_lag1"
]

target = "incidence_b"

train = df[df["partition"] == "train"].copy()
validation = df[df["partition"] == "validation"].copy()

X_train = train[environment_features]
y_train = train[target]

X_val = validation[environment_features]
y_val = validation[target]

# ------------------------------------------------------------
# B. Train and evaluate candidate models
# ------------------------------------------------------------

alphas = [0.01, 0.1, 1, 10, 100, 1000]

results = []
models = {}

for alpha in alphas:

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=alpha))
    ])

    model.fit(X_train, y_train)

    predictions = model.predict(X_val)

    results.append({
        "alpha": alpha,
        "MAE": mean_absolute_error(
            y_val, predictions
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, predictions)
        ),
        "R2": r2_score(y_val, predictions)
    })

    models[alpha] = model

results = pd.DataFrame(results).sort_values("MAE")

best_alpha = float(results.iloc[0]["alpha"])
best_model = models[best_alpha]

# ------------------------------------------------------------
# C. Compare against established baselines
# ------------------------------------------------------------

best_predictions = best_model.predict(X_val)

persistence_predictions = (
    validation["incidence_b_lag1"].to_numpy()
)

comparison = pd.DataFrame([
    {
        "Model": "Persistence",
        "MAE": mean_absolute_error(
            y_val, persistence_predictions
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, persistence_predictions)
        ),
        "R2": r2_score(
            y_val, persistence_predictions
        )
    },
    {
        "Model": "Ridge_history",
        "MAE": 12.4767,
        "RMSE": 22.0949,
        "R2": 0.9503
    },
    {
        "Model": "Ridge_environment",
        "MAE": mean_absolute_error(
            y_val, best_predictions
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, best_predictions)
        ),
        "R2": r2_score(
            y_val, best_predictions
        )
    }
])

# ------------------------------------------------------------
# D. Inspect standardized coefficients
# ------------------------------------------------------------

coefficients = pd.DataFrame({
    "Feature": environment_features,
    "Standardized_coefficient":
        best_model.named_steps["ridge"].coef_
}).sort_values(
    "Standardized_coefficient",
    key=abs,
    ascending=False
)

print("=" * 80)
print("E1 — ENVIRONMENTAL-ONLY RIDGE")
print("=" * 80)

print("\nHYPERPARAMETER RESULTS")
print(results.round(4).to_string(index=False))

print("\nSelected alpha:", best_alpha)

print("\nVALIDATION COMPARISON")
print(comparison.round(4).to_string(index=False))

print("\nSTANDARDIZED COEFFICIENTS")
print(coefficients.round(4).to_string(index=False))

# ------------------------------------------------------------
# E. Save results
# ------------------------------------------------------------

results.to_csv(
    TABLE_DIR / "E1_environment_ridge_validation.csv",
    index=False
)

comparison.to_csv(
    TABLE_DIR / "E1_environment_comparison.csv",
    index=False
)

coefficients.to_csv(
    TABLE_DIR / "E1_environment_coefficients.csv",
    index=False
)

print("\nCELL 25 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# BIML 2.0 — CELL 27
# E3: HISTORICAL INCIDENCE + ITN ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

df = pd.read_csv(
    DATA_DIR / "BIML_WHO_temporal_split.csv"
)

features = [
    "incidence_b_lag1",
    "itn_access_lag1"
]

target = "incidence_b"

train = df.loc[df["partition"] == "train"].copy()
validation = df.loc[df["partition"] == "validation"].copy()

X_train = train[features]
y_train = train[target]

X_val = validation[features]
y_val = validation[target]

# ------------------------------------------------------------
# A. Fit candidate Ridge models
# ------------------------------------------------------------

alphas = [0.01, 0.1, 1, 10, 100, 1000]

results = []
models = {}

for alpha in alphas:

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=alpha))
    ])

    model.fit(X_train, y_train)

    predictions = model.predict(X_val)

    results.append({
        "alpha": alpha,
        "MAE": mean_absolute_error(y_val, predictions),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, predictions)
        ),
        "R2": r2_score(y_val, predictions)
    })

    models[alpha] = model

results = (
    pd.DataFrame(results)
    .sort_values(["MAE", "alpha"])
    .reset_index(drop=True)
)

best_alpha = float(results.loc[0, "alpha"])
best_model = models[best_alpha]

# ------------------------------------------------------------
# B. Validation comparison
# ------------------------------------------------------------

e3_predictions = best_model.predict(X_val)
persistence_predictions = X_val[
    "incidence_b_lag1"
].to_numpy()

def calculate_metrics(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, y_pred)
        ),
        "R2": r2_score(y_true, y_pred)
    }

comparison = pd.DataFrame([
    {
        "Model": "Persistence",
        **calculate_metrics(
            y_val, persistence_predictions
        )
    },
    {
        "Model": "E3_history_ITN",
        **calculate_metrics(
            y_val, e3_predictions
        )
    }
])

# ------------------------------------------------------------
# C. Incremental value relative to persistence
# ------------------------------------------------------------

persistence_mae = comparison.loc[
    comparison["Model"] == "Persistence", "MAE"
].iloc[0]

e3_mae = comparison.loc[
    comparison["Model"] == "E3_history_ITN", "MAE"
].iloc[0]

improvement_pct = (
    100 * (persistence_mae - e3_mae) / persistence_mae
)

# ------------------------------------------------------------
# D. Inspect standardized coefficients
# ------------------------------------------------------------

coefficients = pd.DataFrame({
    "Feature": features,
    "Standardized_coefficient":
        best_model.named_steps["ridge"].coef_
})

print("=" * 80)
print("E3 — HISTORICAL INCIDENCE + ITN ACCESS")
print("=" * 80)

print("\nHYPERPARAMETER RESULTS")
print(results.round(4).to_string(index=False))

print("\nSelected alpha:", best_alpha)

print("\nVALIDATION COMPARISON")
print(comparison.round(4).to_string(index=False))

print(
    "\nMAE improvement relative to persistence:",
    f"{improvement_pct:.2f}%"
)

print("\nSTANDARDIZED COEFFICIENTS")
print(coefficients.round(4).to_string(index=False))

# ------------------------------------------------------------
# E. Save results
# ------------------------------------------------------------

results.to_csv(
    TABLE_DIR / "E3_history_ITN_validation.csv",
    index=False
)

comparison.to_csv(
    TABLE_DIR / "E3_history_ITN_comparison.csv",
    index=False
)

coefficients.to_csv(
    TABLE_DIR / "E3_history_ITN_coefficients.csv",
    index=False
)

print("\nCELL 27 COMPLETED SUCCESSFULLY")

```

```python

# ============================================================
# BIML 2.0 — CELL 29
# E4 VS PERSISTENCE: COUNTRY-LEVEL VALIDATION DIAGNOSTICS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error

df = pd.read_csv(
    DATA_DIR / "BIML_WHO_temporal_split.csv"
)

features = [
    "incidence_b_lag1",
    "ERA5_T2M_C_mean_lag1",
    "Rainfall_total_mm_lag1",
    "RH_pct_mean_lag1",
    "NDVI_mean_lag1",
    "NDWI_mean_lag1",
    "itn_access_lag1"
]

train = df[df["partition"] == "train"].copy()
validation = df[df["partition"] == "validation"].copy()

# ------------------------------------------------------------
# A. Refit E4 using selected validation hyperparameter
# ------------------------------------------------------------

model = Pipeline([
    ("scaler", StandardScaler()),
    ("ridge", Ridge(alpha=1.0))
])

model.fit(
    train[features],
    train["incidence_b"]
)

validation["E4_prediction"] = model.predict(
    validation[features]
)

validation["Persistence_prediction"] = (
    validation["incidence_b_lag1"]
)

# ------------------------------------------------------------
# B. Calculate paired absolute errors
# ------------------------------------------------------------

validation["E4_absolute_error"] = (
    validation["incidence_b"]
    - validation["E4_prediction"]
).abs()

validation["Persistence_absolute_error"] = (
    validation["incidence_b"]
    - validation["Persistence_prediction"]
).abs()

validation["error_difference"] = (
    validation["E4_absolute_error"]
    - validation["Persistence_absolute_error"]
)

# Positive difference means E4 performs worse.

# ------------------------------------------------------------
# C. Country-level comparison
# ------------------------------------------------------------

country_results = (
    validation.groupby(["Country", "CountryCode"])
    .agg(
        E4_MAE=("E4_absolute_error", "mean"),
        Persistence_MAE=("Persistence_absolute_error", "mean"),
        Difference=("error_difference", "mean")
    )
    .reset_index()
    .sort_values("Difference", ascending=False)
)

print("=" * 80)
print("E4 VS PERSISTENCE — COUNTRY-LEVEL VALIDATION")
print("=" * 80)

print(
    country_results.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# D. Summary of country-level changes
# ------------------------------------------------------------

improved = (country_results["Difference"] < 0).sum()
worsened = (country_results["Difference"] > 0).sum()
unchanged = (country_results["Difference"] == 0).sum()

print("\nCOUNTRY-LEVEL SUMMARY")
print("Countries improved:", improved)
print("Countries worsened:", worsened)
print("Countries unchanged:", unchanged)

# ------------------------------------------------------------
# E. Year-specific performance
# ------------------------------------------------------------

year_results = (
    validation.groupby("Year")
    .agg(
        E4_MAE=("E4_absolute_error", "mean"),
        Persistence_MAE=("Persistence_absolute_error", "mean"),
        Difference=("error_difference", "mean")
    )
    .reset_index()
)

print("\nYEAR-SPECIFIC VALIDATION PERFORMANCE")

print(
    year_results.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# F. Verify aggregate results
# ------------------------------------------------------------

print("\nAGGREGATE VERIFICATION")

print(
    "E4 MAE:",
    round(
        mean_absolute_error(
            validation["incidence_b"],
            validation["E4_prediction"]
        ), 4
    )
)

print(
    "Persistence MAE:",
    round(
        mean_absolute_error(
            validation["incidence_b"],
            validation["Persistence_prediction"]
        ), 4
    )
)

# ------------------------------------------------------------
# G. Save diagnostics
# ------------------------------------------------------------

country_results.to_csv(
    TABLE_DIR / "E4_country_validation_diagnostics.csv",
    index=False
)

year_results.to_csv(
    TABLE_DIR / "E4_year_validation_diagnostics.csv",
    index=False
)

validation.to_csv(
    DATA_DIR / "BIML_E4_validation_predictions.csv",
    index=False
)

print("\nCELL 29 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# CELL 29 — RECONSTRUCT ENVIRONMENTAL VC COMPONENTS
# BIML 2.0
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

DATA_DIR = Path.cwd() / "data"

env_path = DATA_DIR / "Environmental_2013_2024_with_ERA5_T2M.csv"

assert env_path.exists(), f"Dataset not found: {env_path}"

env_bio = pd.read_csv(env_path)

required = ["Country", "Year", "Month",
            "ERA5_T2M_C", "Rainfall_mm"]

missing = [c for c in required if c not in env_bio.columns]
assert not missing, f"Missing columns: {missing}"

T = pd.to_numeric(env_bio["ERA5_T2M_C"], errors="coerce")
R = pd.to_numeric(env_bio["Rainfall_mm"], errors="coerce")

# Previously retained parameterization
# Valid mathematical domain: T > 9.9 °C

valid = T.gt(9.9) & T.notna() & R.ge(0) & R.notna()

env_bio["G_T"] = np.nan
env_bio["a_T"] = np.nan
env_bio["p_T"] = np.nan
env_bio["m_R"] = np.nan

env_bio.loc[valid, "G_T"] = (
    0.5 + 36.5 / (T.loc[valid] - 9.9)
)

env_bio.loc[valid, "a_T"] = (
    0.7 / env_bio.loc[valid, "G_T"]
)

env_bio.loc[valid, "p_T"] = (
    0.5 ** (1 / env_bio.loc[valid, "G_T"])
)

env_bio.loc[valid, "m_R"] = 10 * R.loc[valid]

print("=" * 65)
print("CELL 29 — ENVIRONMENTAL VC COMPONENT AUDIT")
print("=" * 65)

print("Total monthly observations:", len(env_bio))
print("Valid observations:", int(valid.sum()))
print("Invalid observations:", int((~valid).sum()))

print("\nTemperature range:")
print(T.agg(["min", "max"]))

print("\nCalculated biological components:")
print(
    env_bio[["G_T", "a_T", "p_T", "m_R"]]
    .describe()
    .round(5)
)

print("\nFirst five observations:")
print(
    env_bio[
        ["Country", "Year", "Month",
         "ERA5_T2M_C", "Rainfall_mm",
         "G_T", "a_T", "p_T", "m_R"]
    ].head().to_string(index=False)
)

assert env_bio.loc[valid, "p_T"].between(
    0, 1, inclusive="neither"
).all(), "Invalid survival probabilities"

print("\nComponent reconstruction completed.")
print("Final VC and I(T) have NOT yet been calculated.")
```

```python

# ============================================================
# CELL 29 — AUDIT EXISTING BIOLOGICAL FEATURE FILES
# ============================================================

from pathlib import Path
import pandas as pd

ROOT = Path.cwd()

search_roots = [
    ROOT / "data",
    ROOT / "outputs",
    ROOT / "results",
    ROOT / "BIML_final_results",
]

keywords = (
    "vc", "eip", "biolog", "suh", "villena",
    "msos", "detinova", "b2", "infectious"
)

matches = []

for folder in search_roots:
    if not folder.exists():
        continue

    for path in folder.rglob("*"):
        if not path.is_file():
            continue

        if path.suffix.lower() not in (".csv", ".parquet"):
            continue

        if any(word in path.name.lower() for word in keywords):
            matches.append(path)

print("=" * 70)
print("CELL 29 — EXISTING BIOLOGICAL DATA AUDIT")
print("=" * 70)
print("Working directory:", ROOT)
print("Candidate files:", len(matches))

for path in sorted(set(matches)):
    print("\nFILE:", path.relative_to(ROOT))

    try:
        if path.suffix.lower() == ".csv":
            df = pd.read_csv(path, nrows=5)
        else:
            df = pd.read_parquet(path).head(5)

        print("Columns:", df.columns.tolist())
        print("Preview:")
        print(df.head(2).to_string(index=False))

    except Exception as exc:
        print("Read error:", exc)

if not matches:
    print("\nNo matching biological files found.")
    print("We will locate the original implementation next.")
```

```python

# ============================================================
# CELL 31 — BIOLOGICAL TEMPERATURE SUPPORT AUDIT
# BIML 2.0 | ML vs BIML
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path.cwd()
DATA_DIR = ROOT / "data"

env = pd.read_csv(
    DATA_DIR / "Environmental_2013_2024_with_ERA5_T2M.csv"
)

core = pd.read_csv(
    DATA_DIR / "BIML_WHO_core_2015_2024.csv"
)

# ------------------------------------------------------------
# 1. Identify the 29 WHO study countries
# ------------------------------------------------------------

study_countries = set(core["Country"].unique())

env_study = env.loc[
    env["Country"].isin(study_countries)
    & env["Year"].between(2015, 2024)
].copy()

print("=" * 70)
print("CELL 31 — BIOLOGICAL TEMPERATURE SUPPORT")
print("=" * 70)

print("\nWHO countries:", len(study_countries))
print("Matched environmental countries:",
      env_study["Country"].nunique())
print("Monthly observations:", len(env_study))

unmatched = sorted(
    study_countries - set(env_study["Country"].unique())
)

print("Unmatched countries:", unmatched)

# ------------------------------------------------------------
# 2. Temperature support classification
# ------------------------------------------------------------

T_MIN = 17.0
T_MAX = 30.0

T = pd.to_numeric(
    env_study["ERA5_T2M_C"],
    errors="coerce"
)

env_study["temperature_support"] = np.select(
    [
        T < T_MIN,
        T.between(T_MIN, T_MAX),
        T > T_MAX
    ],
    [
        "below",
        "inside",
        "above"
    ],
    default="missing"
)

print("\nTemperature support counts:")
print(
    env_study["temperature_support"]
    .value_counts()
    .to_string()
)

print("\nTemperature support percentages:")
print(
    (
        env_study["temperature_support"]
        .value_counts(normalize=True) * 100
    ).round(2).to_string()
)

# ------------------------------------------------------------
# 3. Country-level diagnostics
# ------------------------------------------------------------

country_support = (
    env_study
    .groupby(["Country", "temperature_support"])
    .size()
    .unstack(fill_value=0)
)

for col in ["below", "inside", "above", "missing"]:
    if col not in country_support.columns:
        country_support[col] = 0

country_support["total"] = country_support[
    ["below", "inside", "above", "missing"]
].sum(axis=1)

country_support["outside"] = (
    country_support["below"]
    + country_support["above"]
)

country_support["outside_pct"] = (
    100 * country_support["outside"]
    / country_support["total"]
)

print("\nCountries with most unsupported months:")
print(
    country_support
    .sort_values("outside", ascending=False)
    .head(15)
    .round(2)
    .to_string()
)

# ------------------------------------------------------------
# 4. Country-year completeness
# ------------------------------------------------------------

annual_support = (
    env_study
    .groupby(["Country", "Year"])
    .agg(
        months=("Month", "count"),
        supported_months=(
            "temperature_support",
            lambda x: (x == "inside").sum()
        ),
        below_months=(
            "temperature_support",
            lambda x: (x == "below").sum()
        ),
        above_months=(
            "temperature_support",
            lambda x: (x == "above").sum()
        )
    )
    .reset_index()
)

annual_support["all_months_supported"] = (
    (annual_support["months"] == 12)
    & (annual_support["supported_months"] == 12)
)

print("\nCountry-years:", len(annual_support))
print("Country-years with all 12 months supported:",
      annual_support["all_months_supported"].sum())

print("Country-years with at least one unsupported month:",
      (~annual_support["all_months_supported"]).sum())

print("\nSupported-month distribution:")
print(
    annual_support["supported_months"]
    .value_counts()
    .sort_index()
    .to_string()
)

# ------------------------------------------------------------
# 5. Save diagnostics only
# ------------------------------------------------------------

OUT = ROOT / "outputs" / "tables"
OUT.mkdir(parents=True, exist_ok=True)

country_support.to_csv(
    OUT / "BIML_temperature_support_by_country.csv"
)

annual_support.to_csv(
    OUT / "BIML_temperature_support_by_country_year.csv",
    index=False
)

print("\nAudit tables saved.")
print("No VC or I(T) values calculated.")
```

```python

# ============================================================
# CELL 32 — PREPARE VALIDATED BIOLOGICAL EIP LOOKUP
# BIML 2.0 | ML vs BIML
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path.cwd()
EIP_DIR = ROOT / "data" / "eip"
OUT_DIR = ROOT / "outputs" / "tables"
OUT_DIR.mkdir(parents=True, exist_ok=True)

grid = pd.read_csv(
    EIP_DIR / "eip_pooled_temperature_grid.csv"
)

moments = pd.read_csv(
    EIP_DIR / "task11_5_eip_moments_pooled.csv"
)

# ------------------------------------------------------------
# 1. Validate source structure
# ------------------------------------------------------------

assert set(grid["EIP"]) == {
    "EIP_10", "EIP_50", "EIP_90"
}

assert grid["model"].eq("pooled").all()
assert moments["model"].eq("pooled").all()

assert not grid.duplicated(["temp", "EIP"]).any()
assert not moments.duplicated(["temp", "quantity"]).any()

assert grid["mean"].notna().all()
assert moments["mean"].notna().all()

# ------------------------------------------------------------
# 2. Reshape EIP quantiles
# ------------------------------------------------------------

eip_lookup = (
    grid.pivot(
        index="temp",
        columns="EIP",
        values="mean"
    )
    .reset_index()
    .sort_values("temp")
)

eip_lookup.columns.name = None

assert (
    (eip_lookup["EIP_10"] <= eip_lookup["EIP_50"])
    & (eip_lookup["EIP_50"] <= eip_lookup["EIP_90"])
).all(), "EIP quantile ordering failed"

# ------------------------------------------------------------
# 3. Extract mean EIP at supported temperatures
# ------------------------------------------------------------

mean_eip = (
    moments.loc[
        moments["quantity"].eq("mean"),
        ["temp", "mean", "lower", "upper"]
    ]
    .rename(columns={
        "mean": "mean_EIP",
        "lower": "mean_EIP_lower",
        "upper": "mean_EIP_upper"
    })
    .sort_values("temp")
)

# Mean EIP is available at 11 temperatures, not 131.
# Keep it separate rather than silently interpolating.

# ------------------------------------------------------------
# 4. Inspect lookup structure
# ------------------------------------------------------------

print("=" * 70)
print("CELL 32 — BIOLOGICAL EIP LOOKUP")
print("=" * 70)

print("\nEIP quantile grid:")
print("Rows:", len(eip_lookup))
print("Temperature range:",
      eip_lookup["temp"].min(),
      "to",
      eip_lookup["temp"].max())

print("\nFirst five rows:")
print(eip_lookup.head().round(4).to_string(index=False))

print("\nMean EIP lookup:")
print("Rows:", len(mean_eip))
print("Temperatures:",
      mean_eip["temp"].tolist())

print("\nMean EIP values:")
print(mean_eip.round(4).to_string(index=False))

# ------------------------------------------------------------
# 5. Save validated lookup tables
# ------------------------------------------------------------

eip_lookup.to_csv(
    OUT_DIR / "BIML_EIP_quantile_lookup.csv",
    index=False
)

mean_eip.to_csv(
    OUT_DIR / "BIML_mean_EIP_lookup.csv",
    index=False
)

print("\nChecks passed.")
print("Lookup tables saved.")
print("No interpolation or extrapolation performed.")
print("No VC or I(T) values calculated.")
```

```python

# ============================================================
# CELL 33 — INTERPOLATE EIP TO MONTHLY ENVIRONMENTAL DATA
# BIML 2.0 | Biological Feature Reconstruction
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()
DATA_DIR = ROOT / "data"
TABLE_DIR = ROOT / "outputs" / "tables"

# ------------------------------------------------------------
# 1. Load environmental observations and EIP lookup tables
# ------------------------------------------------------------

env = pd.read_csv(
    DATA_DIR / "Environmental_2013_2024_with_ERA5_T2M.csv"
)

eip_grid = pd.read_csv(
    TABLE_DIR / "BIML_EIP_quantile_lookup.csv"
)

mean_eip = pd.read_csv(
    TABLE_DIR / "BIML_mean_EIP_lookup.csv"
)

# ------------------------------------------------------------
# 2. Restrict to our established WHO study countries and years
# ------------------------------------------------------------

core = pd.read_csv(
    DATA_DIR / "BIML_WHO_core_2015_2024.csv"
)

countries = set(core["Country"].unique())

bio_monthly = env.loc[
    env["Country"].isin(countries)
    & env["Year"].between(2015, 2024)
].copy()

bio_monthly = bio_monthly.sort_values(
    ["Country", "Year", "Month"]
).reset_index(drop=True)

assert len(bio_monthly) == 3480
assert bio_monthly["Country"].nunique() == 29

# ------------------------------------------------------------
# 3. Identify supported temperatures
# ------------------------------------------------------------

T = pd.to_numeric(
    bio_monthly["ERA5_T2M_C"],
    errors="coerce"
)

supported = T.between(17.0, 30.0)

bio_monthly["bio_temp_supported"] = supported

# ------------------------------------------------------------
# 4. Interpolation helper
# ------------------------------------------------------------

def interpolate_supported(temperature, lookup, value_column):
    """
    Linear interpolation within the validated temperature
    range only. Unsupported temperatures remain NaN.
    """

    result = np.full(len(temperature), np.nan)

    x = lookup["temp"].to_numpy(dtype=float)
    y = lookup[value_column].to_numpy(dtype=float)

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    valid = (
        temperature.notna()
        & temperature.between(x.min(), x.max())
    )

    result[valid.to_numpy()] = np.interp(
        temperature.loc[valid].to_numpy(dtype=float),
        x,
        y
    )

    return result

# ------------------------------------------------------------
# 5. Interpolate EIP quantities
# ------------------------------------------------------------

for quantity in ["EIP_10", "EIP_50", "EIP_90"]:
    bio_monthly[quantity] = interpolate_supported(
        T, eip_grid, quantity
    )

for quantity in [
    "mean_EIP",
    "mean_EIP_lower",
    "mean_EIP_upper"
]:
    bio_monthly[quantity] = interpolate_supported(
        T, mean_eip, quantity
    )

# ------------------------------------------------------------
# 6. Integrity checks
# ------------------------------------------------------------

eip_columns = [
    "EIP_10",
    "EIP_50",
    "EIP_90",
    "mean_EIP",
    "mean_EIP_lower",
    "mean_EIP_upper"
]

assert bio_monthly.loc[
    supported, eip_columns
].notna().all().all()

assert bio_monthly.loc[
    ~supported, eip_columns
].isna().all().all()

assert (
    bio_monthly.loc[supported, "EIP_10"]
    <= bio_monthly.loc[supported, "EIP_50"]
).all()

assert (
    bio_monthly.loc[supported, "EIP_50"]
    <= bio_monthly.loc[supported, "EIP_90"]
).all()

# ------------------------------------------------------------
# 7. Display results
# ------------------------------------------------------------

print("=" * 70)
print("CELL 33 — MONTHLY EIP FEATURE RECONSTRUCTION")
print("=" * 70)

print("\nMonthly observations:", len(bio_monthly))
print("Countries:", bio_monthly["Country"].nunique())

print("\nSupported observations:", supported.sum())
print("Unsupported observations:", (~supported).sum())

print("\nEIP summary:")
print(
    bio_monthly[eip_columns]
    .describe()
    .round(4)
    .to_string()
)

print("\nFirst five observations:")
print(
    bio_monthly[
        ["Country", "Year", "Month",
         "ERA5_T2M_C", "bio_temp_supported",
         "mean_EIP", "EIP_50"]
    ]
    .head()
    .round(4)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 8. Save monthly biological EIP dataset
# ------------------------------------------------------------

output_path = DATA_DIR / "BIML_monthly_EIP_features.csv"

bio_monthly.to_csv(output_path, index=False)

print("\nSaved:", output_path)
print("No VC or I(T) calculated yet.")
```

```python

# ============================================================
# CELL 35 — BIOLOGICAL PARAMETER CONSISTENCY
# BIML 2.0 | VC and I(T)
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()

vc = pd.read_csv(
    ROOT / "data" / "eip" / "fast_vc_336_scenarios.csv"
)

# ------------------------------------------------------------
# 1. Extract unique temperature-dependent parameter sets
# ------------------------------------------------------------

parameter_cols = [
    "temp",
    "alpha",
    "shape_total_S",
    "rate_total_S",
    "mu",
    "k",
    "EIP_DD"
]

parameters = (
    vc[parameter_cols]
    .drop_duplicates()
    .sort_values("temp")
    .reset_index(drop=True)
)

assert parameters["temp"].is_unique
assert len(parameters) == 8

# ------------------------------------------------------------
# 2. Inspect survival distribution moments
# ------------------------------------------------------------

# Diagnostic only:
# If shape_total_S and rate_total_S are Gamma(shape, rate)
# parameters, the following are implied distribution moments.

parameters["gamma_implied_mean"] = (
    parameters["shape_total_S"]
    / parameters["rate_total_S"]
)

parameters["gamma_implied_sd"] = (
    np.sqrt(parameters["shape_total_S"])
    / parameters["rate_total_S"]
)

# ------------------------------------------------------------
# 3. Compare EIP representations
# ------------------------------------------------------------

eip = pd.read_csv(
    ROOT / "outputs" / "tables"
    / "BIML_mean_EIP_lookup.csv"
)

parameters["pooled_mean_EIP_interp"] = np.interp(
    parameters["temp"].to_numpy(),
    eip["temp"].to_numpy(),
    eip["mean_EIP"].to_numpy()
)

parameters["EIP_DD_minus_pooled_mean"] = (
    parameters["EIP_DD"]
    - parameters["pooled_mean_EIP_interp"]
)

# ------------------------------------------------------------
# 4. Inspect feeding-specific parameters
# ------------------------------------------------------------

feeding = (
    vc[["temp", "feed", "g1", "g2"]]
    .drop_duplicates()
    .sort_values(["temp", "feed"])
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 5. Display audit
# ------------------------------------------------------------

print("=" * 70)
print("CELL 35 — BIOLOGICAL PARAMETER CONSISTENCY")
print("=" * 70)

print("\nA. TEMPERATURE-DEPENDENT PARAMETERS")
print(parameters.round(5).to_string(index=False))

print("\nB. FEEDING-SPECIFIC PARAMETERS")
print(feeding.round(6).to_string(index=False))

print("\nC. PARAMETER CORRELATIONS")
print(
    parameters[
        [
            "temp",
            "alpha",
            "shape_total_S",
            "rate_total_S",
            "mu",
            "EIP_DD"
        ]
    ]
    .corr()
    .round(3)
    .to_string()
)

print("\nD. VALIDATION CHECKS")
print("Unique temperatures:", len(parameters))
print("Feeding configurations:", feeding["feed"].nunique())
print("Gamma-implied means finite:",
      np.isfinite(parameters["gamma_implied_mean"]).all())
print("Gamma-implied SDs positive:",
      (parameters["gamma_implied_sd"] > 0).all())

print("\nNo VC or I(T) calculated.")
print("No files modified.")
```

```python

# ============================================================
# CELL 36 — LOCATE ORIGINAL BIOLOGICAL EQUATIONS
# BIML 2.0 | Source-Code Verification
# ============================================================

from pathlib import Path
import re

ROOTS = [
    Path.cwd(),
    Path.home() / "Desktop" / "Suh_et_al_mSOS_gambiae_only"
]

EXTENSIONS = {".R", ".r", ".py", ".stan", ".Rmd"}

PATTERNS = {
    "alpha": r"\balpha\b",
    "survival": r"shape_total_S|rate_total_S|survival",
    "EIP": r"EIP_DD|EIP_50|incubation",
    "vector_competence": r"competence|sporozoite",
    "biting": r"biting|feeding|feed_1|feed_2",
    "vectorial_capacity": r"vectorial.capacity|vectorial_capacity|\bVC\b"
}

MAX_MATCHES_PER_CATEGORY = 20

print("=" * 70)
print("CELL 36 — ORIGINAL BIOLOGICAL EQUATION SEARCH")
print("=" * 70)

source_files = []

for root in ROOTS:
    if not root.exists():
        print("Directory not found:", root)
        continue

    for path in root.rglob("*"):
        if (
            path.is_file()
            and path.suffix in EXTENSIONS
            and not any(
                part in {
                    ".git", ".venv", "venv",
                    "node_modules", "__pycache__"
                }
                for part in path.parts
            )
        ):
            source_files.append(path)

source_files = sorted(set(source_files))

print("\nSource files found:", len(source_files))

matches = {name: [] for name in PATTERNS}

for path in source_files:
    try:
        lines = path.read_text(
            encoding="utf-8",
            errors="replace"
        ).splitlines()
    except OSError:
        continue

    for line_number, line in enumerate(lines, start=1):
        for name, pattern in PATTERNS.items():
            if re.search(pattern, line, flags=re.IGNORECASE):
                matches[name].append(
                    (path, line_number, line.strip())
                )

for name, results in matches.items():

    print("\n" + "-" * 70)
    print(name.upper())
    print("Total matching lines:", len(results))

    for path, number, line in results[:MAX_MATCHES_PER_CATEGORY]:
        print(f"\n{path}")
        print(f"Line {number}: {line[:250]}")

    if len(results) > MAX_MATCHES_PER_CATEGORY:
        print(
            f"\n... {len(results) - MAX_MATCHES_PER_CATEGORY}"
            " additional matches not displayed."
        )

print("\n" + "=" * 70)
print("Source inspection completed.")
print("No source files modified.")
print("No VC or I(T) calculated.")
print("=" * 70)
```

```python

# ============================================================
# CELL 40 — FITTED BIOLOGICAL PARAMETER LOOKUP
# BIML 2.0 | Biting Rate + Vector Competence
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()

SOURCE = (
    Path.home()
    / "Desktop"
    / "Suh_et_al_mSOS_gambiae_only"
)

OUT = ROOT / "outputs" / "tables"
OUT.mkdir(parents=True, exist_ok=True)

biting = pd.read_csv(
    SOURCE / "biting_rate_estimates.csv"
)

competence = pd.read_csv(
    SOURCE
    / "validation_results"
    / "task11"
    / "task11_5_vector_competence_pooled.csv"
)

# ------------------------------------------------------------
# 1. Validate original inputs
# ------------------------------------------------------------

biting = biting.sort_values("temp").reset_index(drop=True)
competence = competence.sort_values("temp").reset_index(drop=True)

assert biting["temp"].is_unique
assert competence["temp"].is_unique

assert biting[["temp", "med", "lower", "upper"]].notna().all().all()
assert competence[
    ["temp", "median", "lower", "upper"]
].notna().all().all()

assert (
    (biting["lower"] <= biting["med"])
    & (biting["med"] <= biting["upper"])
).all()

assert (
    (competence["lower"] <= competence["median"])
    & (competence["median"] <= competence["upper"])
).all()

assert competence["lower"].ge(0).all()
assert competence["upper"].le(1).all()

# ------------------------------------------------------------
# 2. Common validated temperature grid
# ------------------------------------------------------------

grid = pd.DataFrame({
    "temp": np.round(np.arange(17, 30.01, 0.1), 1)
})

# ------------------------------------------------------------
# 3. Interpolate biting-rate posterior summaries
# ------------------------------------------------------------

for source_col, output_col in [
    ("med", "alpha"),
    ("lower", "alpha_lower"),
    ("upper", "alpha_upper")
]:
    grid[output_col] = np.interp(
        grid["temp"],
        biting["temp"],
        biting[source_col]
    )

# ------------------------------------------------------------
# 4. Interpolate vector-competence summaries
# ------------------------------------------------------------

for source_col, output_col in [
    ("median", "competence"),
    ("lower", "competence_lower"),
    ("upper", "competence_upper")
]:
    grid[output_col] = np.interp(
        grid["temp"],
        competence["temp"],
        competence[source_col]
    )

# ------------------------------------------------------------
# 5. Validation
# ------------------------------------------------------------

assert len(grid) == 131
assert grid.notna().all().all()

assert (
    (grid["alpha_lower"] <= grid["alpha"])
    & (grid["alpha"] <= grid["alpha_upper"])
).all()

assert (
    (grid["competence_lower"] <= grid["competence"])
    & (grid["competence"] <= grid["competence_upper"])
).all()

assert grid["alpha_lower"].ge(0).all()

assert grid[
    ["competence_lower", "competence", "competence_upper"]
].ge(0).all().all()

assert grid[
    ["competence_lower", "competence", "competence_upper"]
].le(1).all().all()

# Check exact agreement at original competence temperatures
original = grid.merge(
    competence[["temp", "median"]],
    on="temp",
    how="inner"
)

assert len(original) == len(competence)

assert np.allclose(
    original["competence"],
    original["median"],
    atol=1e-12
)

# ------------------------------------------------------------
# 6. Report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 40 — FITTED BIOLOGICAL PARAMETER LOOKUP")
print("=" * 70)

print("\nTemperature grid:", len(grid))
print("Temperature range:",
      grid["temp"].min(), "to", grid["temp"].max())

print("\nParameter summary:")
print(
    grid.drop(columns="temp")
    .describe()
    .round(6)
    .to_string()
)

print("\nSelected temperatures:")
print(
    grid.loc[
        grid["temp"].isin([17, 20, 23, 25, 27, 30])
    ]
    .round(6)
    .to_string(index=False)
)

print("\nMissing values:")
print(grid.isna().sum().to_string())

# ------------------------------------------------------------
# 7. Save validated lookup
# ------------------------------------------------------------

output_path = OUT / "BIML_biting_competence_lookup.csv"

grid.to_csv(output_path, index=False)

print("\nSaved:", output_path)
print("No VC or I(T) calculated.")
```

```python

# ============================================================
# CELL 41 — GOMPERTZ SURVIVAL CONFIGURATION AUDIT
# BIML 2.0 | Biological Feature Reconstruction
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()

vc = pd.read_csv(
    ROOT / "data" / "eip" / "fast_vc_336_scenarios.csv"
)

# ------------------------------------------------------------
# 1. Unique mortality configurations
# ------------------------------------------------------------

mortality = (
    vc[["temp", "feed", "g1", "g2"]]
    .drop_duplicates()
    .sort_values(["temp", "feed"])
    .reset_index(drop=True)
)

assert len(mortality) == 16
assert mortality[["g1", "g2"]].gt(0).all().all()

# ------------------------------------------------------------
# 2. Original Gompertz survival equation
# ------------------------------------------------------------

def gompertz_survival(duration, age_at_infection, g1, g2):
    """
    Survival over 'duration' days conditional on
    being alive at age_at_infection.

    Implements the original surv_gomp() equation.
    """
    duration = np.asarray(duration, dtype=float)

    cumulative_hazard = (
        (g1 / g2)
        * np.exp(age_at_infection * g2)
        * np.expm1(duration * g2)
    )

    return np.exp(-cumulative_hazard)

# ------------------------------------------------------------
# 3. Sensitivity scenarios
# ------------------------------------------------------------

ages = [0, 5, 10]

records = []

for row in mortality.itertuples(index=False):

    for age in ages:

        for duration in [10, 15, 20]:

            survival = gompertz_survival(
                duration=duration,
                age_at_infection=age,
                g1=row.g1,
                g2=row.g2
            )

            records.append({
                "temp": row.temp,
                "feed": row.feed,
                "age_at_infection": age,
                "duration_days": duration,
                "survival_probability": float(survival)
            })

audit = pd.DataFrame(records)

# ------------------------------------------------------------
# 4. Numerical validation
# ------------------------------------------------------------

assert audit["survival_probability"].between(0, 1).all()

# Survival must decrease as duration increases
duration_check = (
    audit.sort_values("duration_days")
    .groupby(["temp", "feed", "age_at_infection"])[
        "survival_probability"
    ]
    .apply(lambda x: x.is_monotonic_decreasing)
)

assert duration_check.all()

# Survival must not increase with age at infection
age_check = (
    audit.sort_values("age_at_infection")
    .groupby(["temp", "feed", "duration_days"])[
        "survival_probability"
    ]
    .apply(lambda x: x.is_monotonic_decreasing)
)

assert age_check.all()

# ------------------------------------------------------------
# 5. Results
# ------------------------------------------------------------

print("=" * 70)
print("CELL 41 — GOMPERTZ SURVIVAL AUDIT")
print("=" * 70)

print("\nMortality configurations:", len(mortality))
print("Age scenarios:", ages)

print("\nA. SURVIVAL AT 15 DAYS")

selected = audit[
    audit["duration_days"].eq(15)
].copy()

print(
    selected.pivot_table(
        index=["temp", "feed"],
        columns="age_at_infection",
        values="survival_probability"
    )
    .round(6)
    .to_string()
)

print("\nB. OVERALL SURVIVAL SUMMARY")

print(
    audit.groupby(["feed", "age_at_infection"])[
        "survival_probability"
    ]
    .agg(["min", "median", "max"])
    .round(6)
    .to_string()
)

print("\nC. VALIDATION")

print("All probabilities within [0,1]: PASSED")
print("Survival decreases with duration: PASSED")
print("Survival decreases with infection age: PASSED")

print("\nNo VC or I(T) calculated.")
print("No files modified.")
```

```python

# ============================================================
# CELL 42 — MORTALITY PARAMETER PROVENANCE
# BIML 2.0 | Original Source Audit
# ============================================================

from pathlib import Path
import re

SOURCE = (
    Path.home()
    / "Desktop"
    / "Suh_et_al_mSOS_gambiae_only"
)

files = [
    SOURCE / "vectorial_capacity.R",
    SOURCE / "utils" / "VC_functions.R",
    SOURCE / "vectorial_capacity_cluster.R"
]

patterns = [
    r"gomp",
    r"mortality",
    r"survival",
    r"feed",
    r"age",
    r"alpha",
    r"parameter",
    r"Shapiro",
    r"Iacovidou",
    r"g1",
    r"g2"
]

print("=" * 70)
print("CELL 42 — MORTALITY PARAMETER PROVENANCE")
print("=" * 70)

for path in files:

    print("\n" + "=" * 70)
    print("FILE:", path.name)
    print("=" * 70)

    if not path.exists():
        print("FILE NOT FOUND")
        continue

    lines = path.read_text(
        encoding="utf-8",
        errors="replace"
    ).splitlines()

    matched = set()

    for i, line in enumerate(lines):
        if any(
            re.search(p, line, flags=re.IGNORECASE)
            for p in patterns
        ):
            matched.add(i)

    # Include two lines of context around each match
    selected = set()

    for i in matched:
        selected.update(
            range(
                max(0, i - 2),
                min(len(lines), i + 3)
            )
        )

    # Avoid excessive output
    selected = sorted(selected)

    print("Total lines:", len(lines))
    print("Matching lines:", len(matched))

    for i in selected[:250]:
        print(f"{i + 1:04d}: {lines[i]}")

    if len(selected) > 250:
        print(
            f"\n... {len(selected) - 250}"
            " additional context lines omitted."
        )

print("\n" + "=" * 70)
print("Source audit completed.")
print("No files modified.")
print("No VC or I(T) calculated.")
```

```python

# ============================================================
# CELL 44 — GOMPERTZ PARAMETER DOMAIN VALIDATION
# BIML 2.0 | Age-Dependent Mortality
# ============================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Temperature grid
# ------------------------------------------------------------

temp = np.round(np.arange(17, 30.01, 0.1), 1)

# ------------------------------------------------------------
# 2. Original mortality parameterizations
# ------------------------------------------------------------

def mortality_parameters(T, feed):

    if feed == "feed_1":
        g1 = 0.000233*T**2 - 0.0101*T + 0.118
        g2 = 0.00809*T - 0.111

    elif feed == "feed_2":
        g1 = 0.00005632*T**2 - 0.00244*T + 0.028
        g2 = 0.00851*T - 0.0916

    else:
        raise ValueError("Unknown feeding configuration")

    return g1, g2

records = []

for feed in ["feed_1", "feed_2"]:

    g1, g2 = mortality_parameters(temp, feed)

    for T, p1, p2 in zip(temp, g1, g2):

        records.append({
            "temp": T,
            "feed": feed,
            "g1": p1,
            "g2": p2
        })

mortality = pd.DataFrame(records)

# ------------------------------------------------------------
# 3. Check hazards and survival
# ------------------------------------------------------------

ages = np.arange(0, 31)

results = []

for row in mortality.itertuples(index=False):

    hazards = row.g1 * np.exp(row.g2 * ages)

    # Conditional survival over 15 days from age 0
    if abs(row.g2) < 1e-12:
        cumulative_hazard = row.g1 * 15
    else:
        cumulative_hazard = (
            row.g1 / row.g2
            * np.expm1(row.g2 * 15)
        )

    survival = np.exp(-cumulative_hazard)

    results.append({
        "temp": row.temp,
        "feed": row.feed,
        "min_hazard": hazards.min(),
        "max_hazard": hazards.max(),
        "survival_15d": survival
    })

check = pd.DataFrame(results)

# ------------------------------------------------------------
# 4. Validation
# ------------------------------------------------------------

assert np.isfinite(
    mortality[["g1", "g2"]].to_numpy()
).all()

assert np.isfinite(
    check[["min_hazard", "max_hazard", "survival_15d"]].to_numpy()
).all()

assert (check["min_hazard"] > 0).all()

assert check["survival_15d"].between(0, 1).all()

# ------------------------------------------------------------
# 5. Report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 44 — GOMPERTZ PARAMETER DOMAIN VALIDATION")
print("=" * 70)

print("\nParameter ranges:")
print(
    mortality.groupby("feed")[["g1", "g2"]]
    .agg(["min", "max"])
    .round(8)
    .to_string()
)

print("\nHazard and survival ranges:")
print(
    check.groupby("feed")[
        ["min_hazard", "max_hazard", "survival_15d"]
    ]
    .agg(["min", "max"])
    .round(6)
    .to_string()
)

print("\nSelected temperatures:")
print(
    check[
        check["temp"].isin([17, 20, 23, 25, 27, 30])
    ]
    .round(6)
    .to_string(index=False)
)

print("\nNumerical validation: PASSED")
print("Temperature domain: 17–30°C")
print("No VC or I(T) calculated.")
print("No files modified.")
```

```python

# ============================================================
# CELL 46 — HARMONIZED BIOLOGICAL INPUT LOOKUP
# BIML 2.0 | Fitted Biological Components
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path.home() / "Desktop" / "BIML_"
TABLES = ROOT / "outputs" / "tables"

bio = pd.read_csv(
    TABLES / "BIML_biting_competence_lookup.csv"
)

mean_eip = pd.read_csv(
    TABLES / "BIML_mean_EIP_lookup.csv"
)

eip_q = pd.read_csv(
    TABLES / "BIML_EIP_quantile_lookup.csv"
)

# ------------------------------------------------------------
# 1. Standardize temperature keys
# ------------------------------------------------------------

for df in [bio, mean_eip, eip_q]:
    df["temp"] = pd.to_numeric(
        df["temp"], errors="raise"
    ).round(1)

    assert df["temp"].is_unique
    assert df["temp"].notna().all()

expected_grid = np.round(
    np.arange(17.0, 30.01, 0.1), 1
)

assert np.array_equal(
    np.sort(bio["temp"].to_numpy()),
    expected_grid
)

assert np.array_equal(
    np.sort(eip_q["temp"].to_numpy()),
    expected_grid
)

assert mean_eip["temp"].between(17, 30).all()

# ------------------------------------------------------------
# 2. Interpolate mean EIP summaries
# ------------------------------------------------------------

mean_eip = mean_eip.sort_values("temp")

lookup = bio.merge(
    eip_q,
    on="temp",
    how="inner",
    validate="one_to_one"
)

for col in [
    "mean_EIP",
    "mean_EIP_lower",
    "mean_EIP_upper"
]:
    lookup[col] = np.interp(
        lookup["temp"],
        mean_eip["temp"],
        mean_eip[col]
    )

lookup = lookup.sort_values("temp").reset_index(drop=True)

# ------------------------------------------------------------
# 3. Validate biological constraints
# ------------------------------------------------------------

assert len(lookup) == 131
assert lookup.notna().all().all()

assert (
    (lookup["EIP_10"] <= lookup["EIP_50"])
    & (lookup["EIP_50"] <= lookup["EIP_90"])
).all()

assert (
    (lookup["mean_EIP_lower"] <= lookup["mean_EIP"])
    & (lookup["mean_EIP"] <= lookup["mean_EIP_upper"])
).all()

assert lookup[
    ["EIP_10", "EIP_50", "EIP_90",
     "mean_EIP_lower", "mean_EIP"]
].gt(0).all().all()

assert lookup["alpha"].ge(0).all()
assert lookup["competence"].between(0, 1).all()

# ------------------------------------------------------------
# 4. Check agreement at original EIP temperatures
# ------------------------------------------------------------

original_points = lookup.merge(
    mean_eip[["temp", "mean_EIP"]],
    on="temp",
    how="inner",
    suffixes=("_interpolated", "_original"),
    validate="one_to_one"
)

assert len(original_points) == len(mean_eip)

max_error = np.max(
    np.abs(
        original_points["mean_EIP_interpolated"]
        - original_points["mean_EIP_original"]
    )
)

assert max_error < 1e-10

# ------------------------------------------------------------
# 5. Save lookup
# ------------------------------------------------------------

output = TABLES / "BIML_harmonized_biological_lookup.csv"

lookup.to_csv(output, index=False)

# ------------------------------------------------------------
# 6. Report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 46 — HARMONIZED BIOLOGICAL INPUT LOOKUP")
print("=" * 70)

print("\nShape:", lookup.shape)
print("Temperature range:",
      lookup["temp"].min(),
      "to",
      lookup["temp"].max())

print("\nColumns:")
print(lookup.columns.tolist())

print("\nSelected temperatures:")
print(
    lookup.loc[
        lookup["temp"].isin([17, 20, 23, 25, 27, 30]),
        [
            "temp",
            "alpha",
            "competence",
            "EIP_10",
            "EIP_50",
            "EIP_90",
            "mean_EIP"
        ]
    ].round(5).to_string(index=False)
)

print("\nMissing values:",
      int(lookup.isna().sum().sum()))

print("Maximum original-point interpolation error:",
      max_error)

print("\nSaved:", output)
print("\nInterpretation:")
print("- Alpha and competence: interpolated fitted summaries")
print("- EIP quantiles: original reconstructed quantile grid")
print("- Mean EIP: interpolated posterior mean summaries")
print("- Bounds are marginal summaries, not joint uncertainty")
print("- No VC or I(T) calculated")
```

```python

# ============================================================
# CELL 47 — AGE-DEPENDENT INFECTIOUS-BITE COMPONENTS
# BIML 2.0 | Suh–Stopard Biological Integration
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd
from scipy.integrate import quad

ROOT = Path.home() / "Desktop" / "BIML_"

lookup = pd.read_csv(
    ROOT / "outputs/tables/BIML_harmonized_biological_lookup.csv"
)

original = pd.read_csv(
    ROOT / "data/eip/fast_vc_336_scenarios.csv"
)

# ------------------------------------------------------------
# 1. Original temperature-dependent mortality parameters
# ------------------------------------------------------------

def gompertz_parameters(T, feed):

    if feed == "feed_1":
        g1 = 0.000233*T**2 - 0.0101*T + 0.118
        g2 = 0.00809*T - 0.111

    elif feed == "feed_2":
        g1 = 0.00005632*T**2 - 0.00244*T + 0.028
        g2 = 0.00851*T - 0.0916

    else:
        raise ValueError(f"Unknown feeding scenario: {feed}")

    return float(g1), float(g2)


# ------------------------------------------------------------
# 2. Original conditional Gompertz survival
# ------------------------------------------------------------

def gompertz_survival(duration, age, g1, g2):

    if duration < 0 or age < 0:
        raise ValueError("Age and duration must be nonnegative")

    cumulative_hazard = (
        g1 * np.exp(g2 * age)
        * np.expm1(g2 * duration) / g2
    )

    return float(np.exp(-cumulative_hazard))


# ------------------------------------------------------------
# 3. Expected remaining lifetime at a given age
# ------------------------------------------------------------

def expected_remaining_lifetime(age, g1, g2):

    result, error = quad(
        lambda duration: gompertz_survival(
            duration, age, g1, g2
        ),
        0,
        np.inf,
        epsabs=1e-9,
        epsrel=1e-8,
        limit=200
    )

    return result


# ------------------------------------------------------------
# 4. Validate against original scenario parameters
# ------------------------------------------------------------

scenarios = (
    original[["temp", "feed", "g1", "g2"]]
    .drop_duplicates()
    .sort_values(["temp", "feed"])
    .reset_index(drop=True)
)

assert len(scenarios) == 16

errors = []

for row in scenarios.itertuples(index=False):

    g1, g2 = gompertz_parameters(row.temp, row.feed)

    errors.append({
        "temp": row.temp,
        "feed": row.feed,
        "g1_error": abs(g1 - row.g1),
        "g2_error": abs(g2 - row.g2)
    })

errors = pd.DataFrame(errors)

max_g1_error = errors["g1_error"].max()
max_g2_error = errors["g2_error"].max()

assert max_g1_error < 1e-8
assert max_g2_error < 1e-8


# ------------------------------------------------------------
# 5. Expected post-EIP bites at selected ages
# ------------------------------------------------------------

records = []

for T in [17.0, 20.0, 23.0, 25.0, 27.0, 30.0]:

    alpha = float(
        lookup.loc[
            np.isclose(lookup["temp"], T),
            "alpha"
        ].iloc[0]
    )

    for feed in ["feed_1", "feed_2"]:

        g1, g2 = gompertz_parameters(T, feed)

        for age in [0, 5, 10, 15]:

            remaining_days = expected_remaining_lifetime(
                age, g1, g2
            )

            expected_bites = alpha * remaining_days

            records.append({
                "temp": T,
                "feed": feed,
                "age": age,
                "alpha": alpha,
                "remaining_days": remaining_days,
                "expected_bites_after_EIP": expected_bites
            })

results = pd.DataFrame(records)


# ------------------------------------------------------------
# 6. Numerical checks
# ------------------------------------------------------------

assert np.isfinite(
    results[
        ["remaining_days", "expected_bites_after_EIP"]
    ].to_numpy()
).all()

assert (results["remaining_days"] > 0).all()

assert (
    results["expected_bites_after_EIP"] >= 0
).all()

for (T, feed), group in results.groupby(["temp", "feed"]):

    group = group.sort_values("age")

    assert np.all(
        np.diff(group["remaining_days"]) <= 1e-10
    )


# ------------------------------------------------------------
# 7. Report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 47 — AGE-DEPENDENT INFECTIOUS-BITE COMPONENTS")
print("=" * 70)

print("\nOriginal scenario parameter agreement:")
print("Maximum g1 error:", max_g1_error)
print("Maximum g2 error:", max_g2_error)

print("\nExpected remaining lifetime and biting:")

print(
    results.round(5).to_string(index=False)
)

print("\nValidation: PASSED")
print("Original Gompertz parameterization: MATCHED")
print("Age-dependent remaining lifetime: VALIDATED")

print("\nInterpretation:")
print("- Expected bites are conditional on reaching an infectious age")
print("- EIP survival has not yet been integrated")
print("- Vector competence has not been multiplied in")
print("- Population-level VC has not been calculated")
print("- No files modified")
```

```python

# ============================================================
# CELL 49 — ORIGINAL mSOS EIP DENSITY VALIDATION
# BIML 2.0 | Faithful R-to-Python Reconstruction
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd

from scipy.integrate import quad
from scipy.special import gammaincc, gammaln

ROOT = Path.home() / "Desktop" / "BIML_"

df = pd.read_csv(
    ROOT / "data/eip/fast_vc_336_scenarios.csv"
)

parameters = (
    df.groupby("temp", as_index=False)[
        ["shape_total_S", "rate_total_S", "mu", "k"]
    ]
    .first()
    .sort_values("temp")
)

# ------------------------------------------------------------
# 1. Original mSOS EIP density
# ------------------------------------------------------------

def msos_eip_pdf(t, shape, rate, mu, k):

    if t <= 0:
        return 0.0

    if min(shape, rate, mu, k) <= 0:
        raise ValueError("EIP parameters must be positive")

    # Upper regularized incomplete gamma Q(shape, rate*t)
    Q = gammaincc(shape, rate * t)

    # Equivalent to R's Rgamma(shape, rate*t),
    # subject to verification of the original function.
    fraction = (
        k + mu * Q
    ) / (k + mu)

    log_gamma_pdf = (
        shape * np.log(rate)
        + (shape - 1) * np.log(t)
        - rate * t
        - gammaln(shape)
    )

    normalization = (
        1 - (k / (k + mu))**k
    )

    adjustment = (
        mu
        * (k / (k + mu))**(1 + k)
        * fraction**(-1 - k)
        / normalization
    )

    return float(
        np.exp(log_gamma_pdf) * adjustment
    )


# ------------------------------------------------------------
# 2. Integrate density over positive EIP duration
# ------------------------------------------------------------

records = []

for row in parameters.itertuples(index=False):

    integral, error = quad(
        msos_eip_pdf,
        0,
        np.inf,
        args=(
            row.shape_total_S,
            row.rate_total_S,
            row.mu,
            row.k
        ),
        epsabs=1e-9,
        epsrel=1e-8,
        limit=300
    )

    records.append({
        "temp": row.temp,
        "integral": integral,
        "integration_error": error,
        "absolute_error_from_1": abs(integral - 1)
    })

results = pd.DataFrame(records)

# ------------------------------------------------------------
# 3. Report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 49 — ORIGINAL mSOS EIP DENSITY VALIDATION")
print("=" * 70)

print(
    results.round(10).to_string(index=False)
)

print("\nMaximum normalization error:",
      results["absolute_error_from_1"].max())

assert np.isfinite(results["integral"]).all()

assert (
    results["absolute_error_from_1"].max() < 1e-5
), "EIP density normalization failed"

print("\nEIP density normalization: PASSED")
print("No VC or I(T) calculated.")
print("No files modified.")
```

```python

# ============================================================
# CELL 50B — STABLE mSOS INFECTIOUS-BITE BENCHMARK
# BIML 2.0 | Correct Numerical Overflow
# ============================================================

import numpy as np
import pandas as pd

from scipy.integrate import quad
from scipy.special import exp1

# Reuse:
# df and msos_eip_pdf from Cell 50

# ------------------------------------------------------------
# 1. Stable scaled exponential integral exp(x) * E1(x)
# ------------------------------------------------------------

def scaled_exp1(x):

    if x <= 0:
        raise ValueError("x must be positive")

    if x < 50:
        return float(np.exp(x) * exp1(x))

    # Asymptotic expansion for large x:
    # exp(x) E1(x) = 1/x - 1/x² + 2!/x³ - ...
    # Evaluate with a controlled number of terms.

    inv = 1.0 / x
    term = inv
    total = term

    for n in range(1, 15):
        term *= -n * inv

        # Stop before asymptotic terms grow
        if abs(term) > abs(total) and n > 1:
            break

        total += term

        if abs(term) < 1e-14 * abs(total):
            break

    return max(float(total), 0.0)


# ------------------------------------------------------------
# 2. Stable Gompertz survival
# ------------------------------------------------------------

def survival_stable(duration, age, g1, g2):

    if duration < 0:
        return 0.0

    exponent = g2 * age

    if exponent > 700:
        return 0.0

    x = (g1 / g2) * np.exp(exponent)

    y = g2 * duration

    if y > 700:
        return 0.0

    hazard = x * np.expm1(y)

    if hazard > 700:
        return 0.0

    return float(np.exp(-hazard))


# ------------------------------------------------------------
# 3. Stable remaining lifetime
# ------------------------------------------------------------

def remaining_lifetime_stable(age, g1, g2):

    log_x = np.log(g1 / g2) + g2 * age

    if log_x > 700:
        return 0.0

    x = np.exp(log_x)

    return scaled_exp1(x) / g2


# ------------------------------------------------------------
# 4. Expected infectious bites
# ------------------------------------------------------------

def expected_infectious_bites_stable(row):

    shape = row["shape_total_S"]
    rate = row["rate_total_S"]
    mu = row["mu"]
    k = row["k"]

    alpha = row["alpha"]
    g1 = row["g1"]
    g2 = row["g2"]
    a0 = row["a0"]

    def integrand(t):

        pdf = msos_eip_pdf(
            t, shape, rate, mu, k
        )

        surv = survival_stable(
            t, a0, g1, g2
        )

        if surv == 0:
            return 0.0

        lifetime = remaining_lifetime_stable(
            a0 + t, g1, g2
        )

        return pdf * surv * lifetime

    integral, error = quad(
        integrand,
        0,
        250,
        epsabs=1e-9,
        epsrel=1e-8,
        limit=300
    )

    return alpha * integral


# ------------------------------------------------------------
# 5. Recalculate all 336 scenarios
# ------------------------------------------------------------

results = df.copy()

results["python_result"] = results.apply(
    expected_infectious_bites_stable,
    axis=1
)

results["absolute_error"] = (
    results["python_result"] - results["E_j_a0"]
).abs()

results["relative_error_pct"] = (
    100 * results["absolute_error"]
    / results["E_j_a0"].abs().clip(lower=1e-12)
)


# ------------------------------------------------------------
# 6. Numerical validation
# ------------------------------------------------------------

assert np.isfinite(
    results["python_result"]
).all(), "Non-finite predictions detected"

assert (
    results["python_result"] >= 0
).all(), "Negative expected bites detected"


# ------------------------------------------------------------
# 7. Benchmark report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 50B — STABLE mSOS INFECTIOUS-BITE BENCHMARK")
print("=" * 70)

print("\nScenarios evaluated:", len(results))

print("\nFirst 20 comparisons:")

print(
    results[
        [
            "temp",
            "feed",
            "a0",
            "E_j_a0",
            "python_result",
            "absolute_error",
            "relative_error_pct"
        ]
    ].head(20).round(8).to_string(index=False)
)

print("\nError summary:")

print(
    results[
        ["absolute_error", "relative_error_pct"]
    ].describe().round(8).to_string()
)

print("\nMaximum absolute error:",
      results["absolute_error"].max())

print("\nMaximum relative error (%):",
      results["relative_error_pct"].max())

print("\nMean absolute error:",
      results["absolute_error"].mean())

print("\nNumerical finiteness: PASSED")
print("Benchmark equivalence: NOT YET DETERMINED")
print("No files modified.")
print("No VC or I(T) calculated.")
```

```python

# ============================================================
# CELL 52 — CONTINUOUS VS DISCRETE POISSON EXPECTATION
# BIML 2.0 | Diagnose Original R Integration
# ============================================================

import numpy as np
import pandas as pd

from scipy.integrate import quad
from scipy.special import gammaln

# ------------------------------------------------------------
# 1. Continuous extension of Poisson probability
# ------------------------------------------------------------

def continuous_poisson_weight(j, lam):

    if j < 0:
        return 0.0

    if lam <= 0:
        return 0.0

    log_weight = (
        j * np.log(lam)
        - lam
        - gammaln(j + 1)
    )

    return float(np.exp(log_weight))


# ------------------------------------------------------------
# 2. Compare continuous integral with discrete expectation
# ------------------------------------------------------------

lambda_values = [
    0.1, 0.25, 0.5, 1.0, 2.0,
    3.0, 5.0, 8.0, 12.0
]

records = []

for lam in lambda_values:

    continuous, integration_error = quad(
        lambda j: j * continuous_poisson_weight(j, lam),
        0,
        np.inf,
        epsabs=1e-10,
        epsrel=1e-9,
        limit=300
    )

    discrete = lam

    records.append({
        "lambda": lam,
        "discrete_expectation": discrete,
        "continuous_integral": continuous,
        "ratio_continuous_discrete": continuous / discrete,
        "relative_difference_pct": (
            100 * (continuous - discrete) / discrete
        )
    })

comparison = pd.DataFrame(records)

# ------------------------------------------------------------
# 3. Report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 52 — CONTINUOUS VS DISCRETE POISSON EXPECTATION")
print("=" * 70)

print(
    comparison.round(8).to_string(index=False)
)

print("\nMaximum absolute relative difference (%):")
print(
    comparison["relative_difference_pct"].abs().max()
)

print("\nInterpretation:")
print("- Discrete Poisson expectation equals lambda exactly.")
print("- Continuous integration is not a discrete probability sum.")
print("- This is a diagnostic, not yet a reproduction of R cubature.")
print("- No files modified.")
print("- No VC or I(T) calculated.")
```

```python

# ============================================================
# CELL 58 — CONTINUOUS POISSON + GOMPERTZ TEST
# BIML 2.0 | One Post-EIP Age
# ============================================================

import numpy as np
from scipy.integrate import quad
from scipy.special import gammaln
import time

# Use the first benchmark scenario
x = df[
    (df["temp"] == 17.5)
    & (df["feed"] == "feed_1")
    & (df["a0"] == 0)
].iloc[0]

alpha = float(x["alpha"])
g1 = float(x["g1"])
g2 = float(x["g2"])

# Fixed mosquito age at EIP completion
a1 = 30.0

# ------------------------------------------------------------
# 1. Gompertz death-time density
# ------------------------------------------------------------

def death_density(u, age, g1, g2):

    if u < 0:
        return 0.0

    log_hazard = np.log(g1) + g2 * (age + u)

    if log_hazard > 700:
        return 0.0

    hazard = np.exp(log_hazard)

    cumulative = (
        (g1 / g2)
        * np.exp(g2 * age)
        * np.expm1(g2 * u)
    )

    if cumulative > 700:
        return 0.0

    return float(hazard * np.exp(-cumulative))


# ------------------------------------------------------------
# 2. Continuous Poisson expectation
# ------------------------------------------------------------

def continuous_poisson_mean(lam):

    if lam <= 0:
        return 0.0

    def integrand(j):

        log_p = (
            j * np.log(lam)
            - lam
            - gammaln(j + 1)
        )

        return j * np.exp(log_p)

    result, _ = quad(
        integrand,
        0,
        np.inf,
        epsabs=1e-9,
        epsrel=1e-8,
        limit=200
    )

    return result


# ------------------------------------------------------------
# 3. Integrate over mosquito remaining lifespan
# ------------------------------------------------------------

start = time.perf_counter()

def continuous_integrand(u):

    lam = alpha * u

    return (
        continuous_poisson_mean(lam)
        * death_density(u, a1, g1, g2)
    )

continuous_result, continuous_error = quad(
    continuous_integrand,
    0,
    150,
    epsabs=1e-7,
    epsrel=1e-6,
    limit=150
)

continuous_runtime = time.perf_counter() - start

# ------------------------------------------------------------
# 4. Reduced discrete expectation
# ------------------------------------------------------------

discrete_result, _ = quad(
    lambda u: alpha * u
    * death_density(u, a1, g1, g2),
    0,
    150,
    epsabs=1e-9,
    epsrel=1e-8
)

# ------------------------------------------------------------
# 5. Report
# ------------------------------------------------------------

print("=" * 70)
print("CELL 58 — CONTINUOUS POISSON + GOMPERTZ TEST")
print("=" * 70)

print("\nTemperature:", x["temp"])
print("Feeding:", x["feed"])
print("Post-EIP age:", a1)

print("\nBiting rate:", alpha)
print("Gompertz g1:", g1)
print("Gompertz g2:", g2)

print("\nContinuous-j expectation:", continuous_result)
print("Discrete expectation:", discrete_result)

print(
    "Absolute difference:",
    abs(continuous_result - discrete_result)
)

print(
    "Relative difference (%):",
    100 * abs(continuous_result - discrete_result)
    / max(abs(discrete_result), 1e-12)
)

print(
    "\nContinuous calculation runtime (seconds):",
    round(continuous_runtime, 3)
)

print("Integration error estimate:", continuous_error)

print("\nNo files modified.")
print("No population-level VC or I(T) calculated.")
```

```python

# ============================================================
# CELL 59 — FULL EIP-WEIGHTED CONTINUOUS-J BENCHMARK
# BIML 2.0 | One Original Scenario
# ============================================================

import numpy as np
import pandas as pd
from scipy.integrate import quad
from scipy.special import gammaln, gammaincc, gammaln as loggamma
from scipy.interpolate import PchipInterpolator
import time

# ------------------------------------------------------------
# 1. Original benchmark scenario
# ------------------------------------------------------------

x = df[
    (df["temp"] == 17.5)
    & (df["feed"] == "feed_1")
    & (df["a0"] == 0)
].iloc[0]

a0 = float(x["a0"])
alpha = float(x["alpha"])
g1 = float(x["g1"])
g2 = float(x["g2"])
shape = float(x["shape_total_S"])
rate = float(x["rate_total_S"])
mu = float(x["mu"])
k = float(x["k"])

reference = float(x["E_j_a0"])

# ------------------------------------------------------------
# 2. Original EIP density (direct mathematical translation)
# ------------------------------------------------------------

def original_eip_pdf(t):

    if t <= 0:
        return 0.0

    q = gammaincc(shape, rate * t)

    fraction = (k + mu * q) / (k + mu)

    log_gamma_density = (
        shape * np.log(rate)
        + (shape - 1) * np.log(t)
        - rate * t
        - loggamma(shape)
    )

    adjustment = (
        mu
        * (k / (k + mu)) ** (1 + k)
        * fraction ** (-1 - k)
        / (1 - (k / (k + mu)) ** k)
    )

    return float(np.exp(log_gamma_density) * adjustment)


# ------------------------------------------------------------
# 3. Gompertz survival and death density
# ------------------------------------------------------------

def gompertz_survival(u, age):

    cumulative = (
        (g1 / g2)
        * np.exp(g2 * age)
        * np.expm1(g2 * u)
    )

    if cumulative > 700:
        return 0.0

    return float(np.exp(-cumulative))


def gompertz_death_density(u, age):

    return (
        g1 * np.exp(g2 * (age + u))
        * gompertz_survival(u, age)
    )


# ------------------------------------------------------------
# 4. Continuous Poisson expectation
# ------------------------------------------------------------

def continuous_poisson_mean(lam):

    if lam <= 0:
        return 0.0

    def integrand(j):

        log_p = (
            j * np.log(lam)
            - lam
            - gammaln(j + 1)
        )

        return j * np.exp(log_p)

    result, _ = quad(
        integrand,
        0,
        np.inf,
        epsabs=1e-8,
        epsrel=1e-7,
        limit=150
    )

    return result


# ------------------------------------------------------------
# 5. Precompute post-EIP continuous bite expectations
# ------------------------------------------------------------

start = time.perf_counter()

age_grid = np.linspace(a0, a0 + 160, 161)

bite_values = []

for age in age_grid:

    value, _ = quad(
        lambda u:
            continuous_poisson_mean(alpha * u)
            * gompertz_death_density(u, age),
        0,
        160,
        epsabs=1e-6,
        epsrel=1e-5,
        limit=100
    )

    bite_values.append(value)

bite_interp = PchipInterpolator(
    age_grid,
    bite_values,
    extrapolate=False
)

# ------------------------------------------------------------
# 6. Full EIP integration
# ------------------------------------------------------------

def full_integrand(t):

    if t < 0 or t > 160:
        return 0.0

    return (
        original_eip_pdf(t)
        * gompertz_survival(t, a0)
        * float(bite_interp(a0 + t))
    )

result, error = quad(
    full_integrand,
    0,
    160,
    epsabs=1e-7,
    epsrel=1e-6,
    limit=250
)

elapsed = time.perf_counter() - start

# ------------------------------------------------------------
# 7. Benchmark comparison
# ------------------------------------------------------------

absolute_error = abs(result - reference)

relative_error_pct = (
    100 * absolute_error
    / max(abs(reference), 1e-12)
)

print("=" * 70)
print("CELL 59 — FULL CONTINUOUS-J BENCHMARK")
print("=" * 70)

print("\nScenario:")
print("Temperature:", x["temp"])
print("Feed:", x["feed"])
print("Age at infection:", a0)

print("\nOriginal R benchmark:", reference)
print("Python continuous-j result:", result)

print("\nAbsolute error:", absolute_error)
print("Relative error (%):", relative_error_pct)

print("\nNumerical integration error:", error)
print("Runtime (seconds):", round(elapsed, 3))

if relative_error_pct < 0.1:
    print("\nBenchmark agreement: EXCELLENT (<0.1%)")
elif relative_error_pct < 1:
    print("\nBenchmark agreement: GOOD (<1%)")
else:
    print("\nBenchmark agreement: REQUIRES INVESTIGATION")

print("\nNo files modified.")
print("No population-level VC or I(T) calculated.")
```

```python

# ============================================================
# CELL 61 — RECOVER SAVED SUH & VILLENA BIOLOGICAL RESULTS
# BIML 2.0 | Read-Only Inventory
# ============================================================

from pathlib import Path
import pandas as pd

ROOTS = [
    Path.home() / "Desktop" / "BIML_Research",
    Path.home() / "Desktop" / "BIML_",
    Path.home() / "Desktop" / "Suh_et_al_mSOS_gambiae_only",
]

EXTENSIONS = {
    ".csv", ".rds", ".rdata", ".rda",
    ".parquet", ".feather", ".pkl",
    ".npz", ".npy"
}

KEYWORDS = [
    "suh", "villena", "biolog",
    "vector", "competence", "capacity",
    "biting", "mortality", "survival",
    "eip", "infectious", "trait",
    "posterior", "temperature", "lookup",
    "validation", "stage10"
]

SKIP_DIRS = {
    ".git", ".venv", "venv", "__pycache__",
    "node_modules", ".ipynb_checkpoints"
}

print("=" * 78)
print("CELL 61 — SAVED SUH & VILLENA RESULT INVENTORY")
print("=" * 78)

records = []

for root in ROOTS:

    print(f"\nROOT: {root}")
    print("Exists:", root.exists())

    if not root.exists():
        continue

    for path in root.rglob("*"):

        if not path.is_file():
            continue

        if any(part in SKIP_DIRS for part in path.parts):
            continue

        if path.suffix.lower() not in EXTENSIONS:
            continue

        relative = str(path.relative_to(root))
        searchable = relative.lower()

        matched = [
            keyword for keyword in KEYWORDS
            if keyword in searchable
        ]

        if not matched:
            continue

        record = {
            "root": str(root),
            "relative_path": relative,
            "extension": path.suffix.lower(),
            "size_mb": round(
                path.stat().st_size / (1024 ** 2), 3
            ),
            "matched_keywords": ", ".join(matched)
        }

        records.append(record)

inventory = pd.DataFrame(records)

if inventory.empty:
    print("\nNo matching numerical files found.")

else:
    inventory = inventory.sort_values(
        ["root", "relative_path"]
    ).reset_index(drop=True)

    print("\nTotal matching numerical files:", len(inventory))

    print("\nFiles by extension:")
    print(inventory["extension"].value_counts().to_string())

    print("\nFILES FOUND:")
    for i, row in inventory.iterrows():
        print(
            f"\n[{i + 1}] {row['root']}/"
            f"{row['relative_path']}"
        )
        print(
            f"    Type: {row['extension']} | "
            f"Size: {row['size_mb']} MB"
        )

    # --------------------------------------------------------
    # Inspect promising CSV files without loading full datasets
    # --------------------------------------------------------

    priority_terms = [
        "vector_competence",
        "vectorial_capacity",
        "biting_rate",
        "mortality",
        "eip",
        "infectious",
        "biological",
        "trait",
        "lookup",
        "fast_vc"
    ]

    candidates = inventory[
        inventory["extension"].eq(".csv")
        & inventory["relative_path"].str.lower().apply(
            lambda p: any(t in p for t in priority_terms)
        )
    ]

    print("\n" + "=" * 78)
    print("PRIORITY CSV SCHEMA INSPECTION")
    print("=" * 78)

    for _, row in candidates.iterrows():

        path = Path(row["root"]) / row["relative_path"]

        print("\nFILE:", path)

        try:
            sample = pd.read_csv(path, nrows=5)

            print("Columns:", list(sample.columns))
            print("Sample rows:", len(sample))
            print(sample.head(2).to_string(index=False))

        except Exception as exc:
            print("Inspection failed:", str(exc))

print("\n" + "=" * 78)
print("INVENTORY COMPLETED")
print("=" * 78)
print("No files modified.")
print("No biological calculations performed.")
```

```python

# ============================================================
# CELL 64 — TRACE ORIGINAL BIOLOGICAL FEATURE CONSTRUCTION
# BIML 2.0 | READ-ONLY SOURCE AUDIT
# ============================================================

from pathlib import Path

ROOT = Path("/home/gato/Desktop/BIML_Research")

SEARCH_DIRS = [
    ROOT / "biml_pipeline",
    ROOT / "BIML_Stage10_B2_Semantic_Correction",
    ROOT / "biological_models" / "villena",
]

TERMS = [
    "B2_deterministic_competence_biting_survival_index",
    "B2_transmission_index",
    "B2_bites_survival_index",
    "B2_Suh_vector_competence_pooled_median",
    "B1_posterior_full_median",
    "B2_feed1_survival_at_mean_EIP",
]

EXTENSIONS = {".py", ".r", ".R", ".ipynb"}

MAX_FILES = 20
CONTEXT_LINES = 5

matches = []

print("=" * 78)
print("CELL 64 — BIOLOGICAL FEATURE SOURCE TRACE")
print("=" * 78)

for directory in SEARCH_DIRS:

    print(f"\nSearching: {directory}")
    print("Exists:", directory.exists())

    if not directory.exists():
        continue

    for path in directory.rglob("*"):

        if not path.is_file():
            continue

        if path.suffix not in EXTENSIONS:
            continue

        if any(
            part in {".git", ".venv", "__pycache__"}
            for part in path.parts
        ):
            continue

        try:
            lines = path.read_text(
                encoding="utf-8",
                errors="replace"
            ).splitlines()
        except Exception:
            continue

        hits = [
            i for i, line in enumerate(lines)
            if any(term in line for term in TERMS)
        ]

        if hits:
            matches.append((path, lines, hits))

print("\nMATCHING SOURCE FILES:", len(matches))

for path, lines, hits in matches[:MAX_FILES]:

    print("\n" + "-" * 78)
    print("FILE:", path)
    print("MATCHING LINES:", len(hits))

    displayed = set()

    for hit in hits[:12]:

        start = max(0, hit - CONTEXT_LINES)
        end = min(len(lines), hit + CONTEXT_LINES + 1)

        print(f"\n--- Source lines {start + 1}–{end} ---")

        for index in range(start, end):

            if index in displayed:
                continue

            marker = ">>" if index == hit else "  "

            print(
                f"{marker} {index + 1:04d}: {lines[index]}"
            )

            displayed.add(index)

if len(matches) > MAX_FILES:
    print(
        "\nAdditional matching files omitted:",
        len(matches) - MAX_FILES
    )

print("\n" + "=" * 78)
print("SOURCE TRACE COMPLETED")
print("No files modified.")
print("No biological calculations repeated.")
print("=" * 78)
```

```python

# ============================================================
# CELL 65 — LOCATE EXISTING VC FORMULAS AND IMPLEMENTATIONS
# Read-only source inspection
# ============================================================

from pathlib import Path
import re

ROOT = Path("/home/gato/Desktop/BIML_Research")

SEARCH_DIRS = [
    ROOT / "biml_pipeline",
    ROOT / "biological_models" / "suh",
    ROOT / "biological_models" / "villena",
    ROOT / "BIML_Stage10_B2_Semantic_Correction",
]

PATTERNS = [
    r"\bvectorial.capacity\b",
    r"\bvectorial_capacity\b",
    r"\bfast_E_j_a0\b",
    r"\bE_j_bites_a0\b",
    r"\bE_j_a0\b",
    r"\bVC\s*=",
    r"\bvc\s*=",
    r"\bI_T\b",
    r"\bI\(T\)",
]

SOURCE_EXTENSIONS = {".py", ".R", ".r"}

matches = []

for directory in SEARCH_DIRS:
    if not directory.exists():
        continue

    for path in directory.rglob("*"):
        if not path.is_file():
            continue
        if path.suffix not in SOURCE_EXTENSIONS:
            continue
        if any(
            part in {".git", ".venv", "__pycache__"}
            for part in path.parts
        ):
            continue

        try:
            lines = path.read_text(
                encoding="utf-8",
                errors="replace"
            ).splitlines()
        except OSError:
            continue

        hits = [
            i for i, line in enumerate(lines)
            if any(
                re.search(p, line, flags=re.IGNORECASE)
                for p in PATTERNS
            )
        ]

        if hits:
            matches.append((path, lines, hits))

print("=" * 76)
print("CELL 65 — EXISTING VC SOURCE INVENTORY")
print("=" * 76)

print("\nMatching source files:", len(matches))

for path, lines, hits in matches:

    print("\n" + "-" * 76)
    print("FILE:", path)
    print("Matches:", len(hits))

    for i in hits[:8]:
        start = max(0, i - 2)
        end = min(len(lines), i + 4)

        print(f"\nLines {start + 1}–{end}:")
        for j in range(start, end):
            print(f"{j + 1:04d}: {lines[j]}")

print("\n" + "=" * 76)
print("COMPLETED — READ ONLY")
print("=" * 76)
```

```python

from pathlib import Path
import subprocess
import shutil

ROOT = Path(
    "/home/gato/Desktop/BIML_Research/biological_models/suh"
)

targets = [
    ROOT / "params_df_E_j_bites_a0.rds",
    ROOT / "fits/params_df.rds",
]

print("=" * 76)
print("CELL 69 — RECOVERED SUH RDS CONTENT AUDIT")
print("=" * 76)

if shutil.which("Rscript") is None:
    print("ERROR: Rscript is not available on PATH.")
else:
    r_code = r'''
args <- commandArgs(trailingOnly=TRUE)

for (path in args) {
  cat("\n", strrep("-", 76), "\n", sep="")
  cat("FILE:", path, "\n")

  if (!file.exists(path)) {
    cat("MISSING\n")
    next
  }

  obj <- tryCatch(readRDS(path), error=function(e) e)

  if (inherits(obj, "error")) {
    cat("READ ERROR:", conditionMessage(obj), "\n")
    next
  }

  cat("CLASS:", paste(class(obj), collapse=", "), "\n")
  cat("DIMENSIONS:", paste(dim(obj), collapse=" x "), "\n")

  if (!is.data.frame(obj)) {
    cat("OBJECT STRUCTURE:\n")
    str(obj, max.level=1, list.len=15)
    next
  }

  cat("ROWS:", nrow(obj), "\n")
  cat("COLUMNS:", ncol(obj), "\n")
  cat("COLUMN NAMES:\n")
  print(names(obj))

  relevant <- intersect(
    c("location", "temp_source", "temp", "feed", "a0",
      "E_j_a0", "E_j_a0_conv", "E_j_dd",
      "alpha", "g1", "g2", "shape", "rate", "mu_PL", "k"),
    names(obj)
  )

  cat("\nFIRST THREE ROWS (RELEVANT COLUMNS):\n")
  print(utils::head(obj[, relevant, drop=FALSE], 3))

  for (col in c("temp", "feed", "a0", "E_j_a0", "E_j_a0_conv")) {
    if (!col %in% names(obj)) next

    x <- obj[[col]]
    cat("\n", col, ":\n", sep="")
    cat("  Missing:", sum(is.na(x)), "\n")
    cat("  Unique:", length(unique(x)), "\n")

    if (is.numeric(x)) {
      finite <- x[is.finite(x)]
      if (length(finite)) {
        cat("  Range:", min(finite), "to", max(finite), "\n")
      }
    } else {
      print(utils::head(unique(x), 15))
    }
  }

  if (all(c("temp", "feed", "a0") %in% names(obj))) {
    cat("\nDISTINCT TEMP-FEED-AGE COMBINATIONS:",
        nrow(unique(obj[c("temp", "feed", "a0")])), "\n")
  }

  if ("E_j_a0" %in% names(obj)) {
    cat("\nINFECTIOUS-BITE VALUES PRESENT:",
        sum(is.finite(obj$E_j_a0)), "\n")
  }
}
'''

    result = subprocess.run(
        ["Rscript", "-e", r_code, *map(str, targets)],
        capture_output=True,
        text=True,
        timeout=120
    )

    print(result.stdout)

    if result.stderr.strip():
        print("\nR STDERR:")
        print(result.stderr[-3000:])

    print("R EXIT CODE:", result.returncode)

print("\nAUDIT COMPLETED — NO FILES MODIFIED")
```

```python

from pathlib import Path

roots = [
    Path("/home/gato/Desktop/BIML_Research"),
    Path("/home/gato/Desktop/BIML_"),
    Path("/home/gato/Desktop/Suh_et_al_mSOS_gambiae_only"),
]

extensions = {".rds", ".rda", ".rdata", ".csv", ".parquet", ".pkl"}

keywords = (
    "e_j",
    "infectious",
    "bites",
    "vectorial",
    "vc_result",
    "params_df_result",
    "pars_sum",
)

exclude_names = {
    "fit_msos_pooled_full.rds",
    "fit_msos_independent_full.rds",
    "biting_rate_fit.rds",
}

print("=" * 76)
print("CELL 70 — FINAL COMPLETED SUH OUTPUT DISCOVERY")
print("=" * 76)

found = {}

for root in roots:
    print(f"\nSEARCH ROOT: {root}")
    print("Exists:", root.exists())

    if not root.exists():
        continue

    for path in root.rglob("*"):
        if not path.is_file():
            continue

        if path.suffix.lower() not in extensions:
            continue

        if path.name.lower() in exclude_names:
            continue

        if not any(k in path.name.lower() for k in keywords):
            continue

        try:
            resolved = path.resolve()
            found[str(resolved)] = (
                path.stat().st_size / (1024 ** 2)
            )
        except OSError:
            continue

print("\n" + "=" * 76)
print("DISCOVERED FILES")
print("=" * 76)

for path, size in sorted(found.items()):
    print(f"{size:10.3f} MB | {path}")

print("\nTOTAL UNIQUE FILES:", len(found))

print("\n" + "=" * 76)
print("SEARCH COMPLETE — NO FILES MODIFIED")
print("=" * 76)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path("/home/gato/Desktop/BIML_Research")

paths = {
    "Villena_B1": (
        ROOT / "biml_pipeline/outputs/stage05/"
        "b1_monthly_posterior_v1.csv"
    ),
    "Suh_B2": (
        ROOT / "biml_pipeline/outputs/stage10/"
        "stage10_b2_corrected_monthly.csv"
    ),
}

frames = {}

print("=" * 76)
print("CELL 71 — BIML BIOLOGICAL PANEL READINESS AUDIT")
print("=" * 76)

for name, path in paths.items():

    print(f"\n{name}: {path}")

    if not path.exists():
        print("MISSING FILE")
        continue

    df = pd.read_csv(path)

    df = df[
        df["Year"].between(2015, 2024)
    ].copy()

    frames[name] = df

    print("Rows:", len(df))
    print("Countries:", df["Country"].nunique())
    print("Years:", sorted(df["Year"].unique()))
    print("Unique country-year-month keys:",
          df[["Country", "Year", "Month"]]
          .drop_duplicates().shape[0])
    print("Duplicate keys:",
          df.duplicated(
              ["Country", "Year", "Month"]
          ).sum())

    numeric = df.select_dtypes(
        include=np.number
    ).columns

    print("\nNumeric feature missingness:")

    missing = (
        df[numeric]
        .isna()
        .sum()
        .sort_values(ascending=False)
    )

    print(missing.to_string())

if len(frames) == 2:

    b1 = frames["Villena_B1"]
    b2 = frames["Suh_B2"]

    keys = ["Country", "Year", "Month"]

    merged = b1.merge(
        b2,
        on=keys,
        how="outer",
        indicator=True,
        suffixes=("_B1", "_B2"),
        validate="one_to_one"
    )

    print("\n" + "=" * 76)
    print("B1–B2 ALIGNMENT")
    print("=" * 76)

    print(merged["_merge"].value_counts().to_string())

    print("\nTemperature-supported observations:")

    if "B2_temperature_supported" in merged:
        print(
            merged["B2_temperature_supported"]
            .value_counts(dropna=False)
            .to_string()
        )

    print("\nCOUNTRY COVERAGE:")

    coverage = (
        merged.groupby("Country")
        .agg(
            months=("Year", "size"),
            years=("Year", "nunique")
        )
        .sort_index()
    )

    print(coverage.to_string())

    print("\nB1 and B2 temperature agreement:")

    if {
        "ERA5_T2M_C_B1",
        "ERA5_T2M_C_B2"
    }.issubset(merged.columns):

        diff = (
            merged["ERA5_T2M_C_B1"]
            - merged["ERA5_T2M_C_B2"]
        ).abs()

        print("Maximum absolute difference:", diff.max())
        print("Mismatches > 1e-6:", (diff > 1e-6).sum())

print("\n" + "=" * 76)
print("AUDIT COMPLETED — READ ONLY")
print("=" * 76)
```

```python

# ============================================================
# CELL 72 — WHO OUTCOME AND BIOLOGICAL PANEL ALIGNMENT
# BIML 2.0 | READ-ONLY AUDIT
# ============================================================

import pandas as pd
from pathlib import Path

WHO_PATH = Path(
    "/home/gato/Desktop/BIML_/data/"
    "BIML_WHO_core_2015_2024.csv"
)

B1_PATH = Path(
    "/home/gato/Desktop/BIML_Research/biml_pipeline/"
    "outputs/stage05/b1_monthly_posterior_v1.csv"
)

B2_PATH = Path(
    "/home/gato/Desktop/BIML_Research/biml_pipeline/"
    "outputs/stage10/stage10_b2_corrected_monthly.csv"
)

print("=" * 78)
print("CELL 72 — WHO–BIOLOGICAL PANEL ALIGNMENT")
print("=" * 78)

who = pd.read_csv(WHO_PATH)
b1 = pd.read_csv(B1_PATH)
b2 = pd.read_csv(B2_PATH)

print("\n1. WHO DATASET STRUCTURE")
print("Shape:", who.shape)
print("Columns:", list(who.columns))
print("\nFirst three rows:")
print(who.head(3).to_string(index=False))

print("\n2. WHO STUDY POPULATION")

who = who[who["Year"].between(2015, 2024)].copy()

print("Countries:", who["Country"].nunique())
print("Years:", who["Year"].nunique())
print("Records:", len(who))
print("Duplicate country-year keys:",
      who.duplicated(["Country", "Year"]).sum())

print("\n3. COUNTRY ALIGNMENT")

who_countries = set(who["Country"])
b1_countries = set(b1["Country"])
b2_countries = set(b2["Country"])

print("\nWHO countries absent from B1:")
print(sorted(who_countries - b1_countries))

print("\nWHO countries absent from B2:")
print(sorted(who_countries - b2_countries))

print("\nBiological countries outside WHO population:")
print(sorted((b1_countries | b2_countries) - who_countries))

print("\n4. EXPECTED MONTHLY PANEL COVERAGE")

expected = (
    who[["Country", "Year"]]
    .drop_duplicates()
    .merge(
        pd.DataFrame({"Month": range(1, 13)}),
        how="cross"
    )
)

b1_keys = b1[["Country", "Year", "Month"]].drop_duplicates()
b2_keys = b2[["Country", "Year", "Month"]].drop_duplicates()

check = (
    expected
    .merge(
        b1_keys.assign(B1_present=True),
        on=["Country", "Year", "Month"],
        how="left",
        validate="one_to_one"
    )
    .merge(
        b2_keys.assign(B2_present=True),
        on=["Country", "Year", "Month"],
        how="left",
        validate="one_to_one"
    )
)

print("Expected monthly rows:", len(expected))
print("B1 matched:", check["B1_present"].notna().sum())
print("B2 matched:", check["B2_present"].notna().sum())
print("B1 missing:", check["B1_present"].isna().sum())
print("B2 missing:", check["B2_present"].isna().sum())

print("\n5. SUH TEMPERATURE SUPPORT IN WHO COUNTRIES")

b2_study = b2.merge(
    who[["Country", "Year"]].drop_duplicates(),
    on=["Country", "Year"],
    how="inner",
    validate="many_to_one"
)

support = (
    b2_study.groupby("Country")
    .agg(
        total_months=("B2_temperature_supported", "size"),
        supported_months=("B2_temperature_supported", "sum")
    )
)

support["unsupported_months"] = (
    support["total_months"] - support["supported_months"]
)

support["supported_pct"] = (
    100 * support["supported_months"]
    / support["total_months"]
)

print(
    support.sort_values("supported_pct")
    .round(2)
    .to_string()
)

print("\n" + "=" * 78)
print("AUDIT COMPLETED — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 73 — ANNUAL BIOLOGICAL COVERAGE AND AGGREGATION AUDIT
# BIML 2.0 | READ ONLY
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path("/home/gato/Desktop/BIML_Research")

who = pd.read_csv(
    "/home/gato/Desktop/BIML_/data/BIML_WHO_core_2015_2024.csv"
)

b1 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage05/b1_monthly_posterior_v1.csv"
)

b2 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage10/stage10_b2_corrected_monthly.csv"
)

keys = ["Country", "Year", "Month"]

study_keys = who[["Country", "Year"]].drop_duplicates()

b1 = b1.merge(
    study_keys, on=["Country", "Year"],
    how="inner", validate="many_to_one"
)

b2 = b2.merge(
    study_keys, on=["Country", "Year"],
    how="inner", validate="many_to_one"
)

monthly = b1.merge(
    b2.drop(columns=["ERA5_T2M_C"]),
    on=keys,
    validate="one_to_one"
)

monthly["B2_temperature_supported"] = (
    monthly["B2_temperature_supported"].astype(bool)
)

annual = (
    monthly.groupby(["Country", "Year"])
    .agg(
        total_months=("Month", "size"),
        supported_months=("B2_temperature_supported", "sum"),
        B1_full_mean=("B1_posterior_full_median", "mean"),
        B1_noEFD_mean=(
            "B1_posterior_noEFD_sensitivity_median", "mean"
        ),
        B2_competence_mean=(
            "B2_Suh_vector_competence_pooled_median", "mean"
        ),
        B2_EIP_mean_days=(
            "B2_Suh_EIP_mean_days_pooled_median", "mean"
        ),
        B2_deterministic_index_mean=(
            "B2_deterministic_competence_biting_survival_index",
            "mean"
        )
    )
    .reset_index()
)

annual["unsupported_months"] = (
    annual["total_months"] - annual["supported_months"]
)

annual["coverage_pct"] = (
    100 * annual["supported_months"] / annual["total_months"]
)

print("=" * 78)
print("CELL 73 — ANNUAL BIOLOGICAL COVERAGE AUDIT")
print("=" * 78)

print("\n1. PANEL STRUCTURE")
print("Monthly records:", len(monthly))
print("Annual records:", len(annual))
print("Countries:", annual["Country"].nunique())
print("Years:", annual["Year"].nunique())

print("\n2. ANNUAL SUH SUPPORT DISTRIBUTION")
print(
    annual["supported_months"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\n3. COVERAGE BY YEAR")
yearly = (
    annual.groupby("Year")
    .agg(
        country_years=("Country", "size"),
        supported_months=("supported_months", "sum"),
        unsupported_months=("unsupported_months", "sum"),
        complete_country_years=(
            "supported_months", lambda x: (x == 12).sum()
        )
    )
)

print(yearly.to_string())

print("\n4. COUNTRY-YEARS WITH FEWER THAN 9 SUPPORTED MONTHS")
low = annual[annual["supported_months"] < 9]

print("Count:", len(low))

if not low.empty:
    print(
        low[[
            "Country", "Year", "supported_months",
            "coverage_pct"
        ]]
        .sort_values(["Country", "Year"])
        .to_string(index=False)
    )

print("\n5. BIOLOGICAL ANNUAL FEATURE MISSINGNESS")
features = [
    "B1_full_mean",
    "B1_noEFD_mean",
    "B2_competence_mean",
    "B2_EIP_mean_days",
    "B2_deterministic_index_mean"
]

print(annual[features].isna().sum().to_string())

print("\n6. COVERAGE COUNTS")
print("Full 12-month support:",
      int((annual["supported_months"] == 12).sum()))
print("Partial support (1–11):",
      int(annual["supported_months"].between(1, 11).sum()))
print("No supported months:",
      int((annual["supported_months"] == 0).sum()))

print("\n7. ANNUAL TARGET ALIGNMENT")
check = who.merge(
    annual[["Country", "Year", "supported_months"]],
    on=["Country", "Year"],
    how="left",
    validate="one_to_one"
)

print("WHO records:", len(check))
print("Unmatched annual biological records:",
      int(check["supported_months"].isna().sum()))

print("\n" + "=" * 78)
print("AUDIT COMPLETE — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 74 — SUH TEMPERATURE SUPPORT BIAS AUDIT
# BIML 2.0 | READ ONLY
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

path = Path(
    "/home/gato/Desktop/BIML_Research/"
    "biml_pipeline/outputs/stage10/"
    "stage10_b2_corrected_monthly.csv"
)

who_path = Path(
    "/home/gato/Desktop/BIML_/data/"
    "BIML_WHO_core_2015_2024.csv"
)

b2 = pd.read_csv(path)
who = pd.read_csv(who_path)

study = who[["Country", "Year"]].drop_duplicates()

df = b2.merge(
    study,
    on=["Country", "Year"],
    how="inner",
    validate="many_to_one"
)

support = df["B2_temperature_supported"].astype(bool)
temp = df["ERA5_T2M_C"]

df["temperature_status"] = np.select(
    [
        temp < 17,
        temp > 30,
        support
    ],
    [
        "Below 17°C",
        "Above 30°C",
        "Supported"
    ],
    default="Other unsupported"
)

print("=" * 78)
print("CELL 74 — SUH TEMPERATURE SUPPORT BIAS AUDIT")
print("=" * 78)

print("\n1. TEMPERATURE STATUS")
print(df["temperature_status"].value_counts().to_string())

print("\n2. TEMPERATURE RANGES")
print(
    df.groupby("temperature_status")["ERA5_T2M_C"]
    .agg(["count", "min", "median", "max"])
    .round(3)
    .to_string()
)

print("\n3. UNSUPPORTED MONTHS BY COUNTRY AND CATEGORY")

unsupported = df[~support]

print(
    pd.crosstab(
        unsupported["Country"],
        unsupported["temperature_status"]
    ).to_string()
)

print("\n4. UNSUPPORTED MONTHS BY CALENDAR MONTH")

print(
    pd.crosstab(
        unsupported["Month"],
        unsupported["temperature_status"]
    ).to_string()
)

print("\n5. FULL-YEAR VS SUPPORTED-MONTH TEMPERATURE")

annual_all = (
    df.groupby(["Country", "Year"])["ERA5_T2M_C"]
    .mean()
    .rename("full_year_temperature")
)

annual_supported = (
    df[support]
    .groupby(["Country", "Year"])["ERA5_T2M_C"]
    .mean()
    .rename("supported_month_temperature")
)

comparison = pd.concat(
    [annual_all, annual_supported],
    axis=1
).reset_index()

comparison["difference_C"] = (
    comparison["supported_month_temperature"]
    - comparison["full_year_temperature"]
)

print(
    comparison["difference_C"]
    .describe()
    .round(3)
    .to_string()
)

print("\n6. LARGEST ABSOLUTE TEMPERATURE DIFFERENCES")

print(
    comparison.assign(
        absolute_difference=lambda x: x["difference_C"].abs()
    )
    .sort_values("absolute_difference", ascending=False)
    .head(15)
    .round(3)
    .to_string(index=False)
)

print("\n" + "=" * 78)
print("AUDIT COMPLETE — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 75 — SUH COVERAGE SELECTION-BIAS AUDIT
# BIML 2.0 | READ ONLY
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path("/home/gato/Desktop/BIML_Research")

who = pd.read_csv(
    "/home/gato/Desktop/BIML_/data/BIML_WHO_core_2015_2024.csv"
)

b2 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage10/"
    "stage10_b2_corrected_monthly.csv"
)

study = who[["Country", "Year"]].drop_duplicates()

b2 = b2.merge(
    study,
    on=["Country", "Year"],
    how="inner",
    validate="many_to_one"
)

annual_support = (
    b2.groupby(["Country", "Year"])
    .agg(
        supported_months=("B2_temperature_supported", "sum"),
        total_months=("Month", "size"),
        annual_temperature=("ERA5_T2M_C", "mean"),
        B2_index_supported_mean=(
            "B2_deterministic_competence_biting_survival_index",
            "mean"
        )
    )
    .reset_index()
)

annual_support["coverage_group"] = np.select(
    [
        annual_support["supported_months"] == 12,
        annual_support["supported_months"].between(9, 11),
        annual_support["supported_months"] < 9
    ],
    [
        "Complete (12)",
        "High partial (9-11)",
        "Low partial (0-8)"
    ],
    default="Other"
)

panel = who.merge(
    annual_support,
    on=["Country", "Year"],
    how="left",
    validate="one_to_one"
)

print("=" * 78)
print("CELL 75 — BIOLOGICAL COVERAGE SELECTION-BIAS AUDIT")
print("=" * 78)

print("\n1. ANNUAL COVERAGE GROUPS")
print(
    panel["coverage_group"]
    .value_counts()
    .to_string()
)

print("\n2. TEMPERATURE AND MALARIA INCIDENCE BY COVERAGE")

summary = (
    panel.groupby("coverage_group")
    .agg(
        observations=("incidence_b", "size"),
        countries=("Country", "nunique"),
        mean_temperature=("annual_temperature", "mean"),
        median_temperature=("annual_temperature", "median"),
        mean_incidence=("incidence_b", "mean"),
        median_incidence=("incidence_b", "median"),
        mean_supported_months=("supported_months", "mean")
    )
)

print(summary.round(3).to_string())

print("\n3. ASSOCIATIONS WITH COVERAGE")

for col in [
    "annual_temperature",
    "incidence_b",
    "Rainfall_total_mm",
    "RH_pct_mean",
    "itn_use"
]:
    valid = panel[["supported_months", col]].dropna()

    pearson = valid["supported_months"].corr(
        valid[col], method="pearson"
    )

    spearman = valid["supported_months"].corr(
        valid[col], method="spearman"
    )

    print(
        f"{col:25s} "
        f"Pearson={pearson: .4f} "
        f"Spearman={spearman: .4f}"
    )

print("\n4. COUNTRIES AFFECTED BY COMPLETE-SUPPORT FILTERING")

country_summary = (
    panel.groupby("Country")
    .agg(
        annual_records=("Year", "size"),
        complete_years=(
            "supported_months",
            lambda x: int((x == 12).sum())
        ),
        low_coverage_years=(
            "supported_months",
            lambda x: int((x < 9).sum())
        )
    )
)

country_summary["years_lost"] = (
    country_summary["annual_records"]
    - country_summary["complete_years"]
)

print(
    country_summary[
        country_summary["years_lost"] > 0
    ]
    .sort_values("years_lost", ascending=False)
    .to_string()
)

print("\n5. STUDY POPULATION RETENTION")

complete = panel[panel["supported_months"] == 12]

print("Full panel country-years:", len(panel))
print("Complete-support country-years:", len(complete))
print("Complete-support countries:", complete["Country"].nunique())
print(
    "Country-years removed:",
    len(panel) - len(complete)
)

print("\n" + "=" * 78)
print("AUDIT COMPLETE — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 76 — BIOLOGICAL FEATURE DISTRIBUTION AUDIT
# BIML 2.0 | READ ONLY
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path("/home/gato/Desktop/BIML_Research")

who = pd.read_csv(
    "/home/gato/Desktop/BIML_/data/BIML_WHO_core_2015_2024.csv"
)

b1 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage05/"
    "b1_monthly_posterior_v1.csv"
)

b2 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage10/"
    "stage10_b2_corrected_monthly.csv"
)

study_keys = who[["Country", "Year"]].drop_duplicates()

b1 = b1.merge(
    study_keys,
    on=["Country", "Year"],
    how="inner",
    validate="many_to_one"
)

b2 = b2.merge(
    study_keys,
    on=["Country", "Year"],
    how="inner",
    validate="many_to_one"
)

b1_annual = (
    b1.groupby(["Country", "Year"])
    .agg(
        Villena_full=(
            "B1_posterior_full_median", "mean"
        ),
        Villena_noEFD=(
            "B1_posterior_noEFD_sensitivity_median", "mean"
        )
    )
    .reset_index()
)

b2_annual = (
    b2.groupby(["Country", "Year"])
    .agg(
        Suh_competence=(
            "B2_Suh_vector_competence_pooled_median", "mean"
        ),
        Suh_EIP_days=(
            "B2_Suh_EIP_mean_days_pooled_median", "mean"
        ),
        Suh_biting=(
            "B2_An_stephensi_biting_median", "mean"
        ),
        Suh_deterministic_index=(
            "B2_deterministic_competence_biting_survival_index",
            "mean"
        ),
        supported_months=(
            "B2_temperature_supported", "sum"
        )
    )
    .reset_index()
)

panel = (
    who.merge(
        b1_annual,
        on=["Country", "Year"],
        validate="one_to_one"
    )
    .merge(
        b2_annual,
        on=["Country", "Year"],
        validate="one_to_one"
    )
)

features = [
    "Villena_full",
    "Villena_noEFD",
    "Suh_competence",
    "Suh_EIP_days",
    "Suh_biting",
    "Suh_deterministic_index"
]

print("=" * 78)
print("CELL 76 — BIOLOGICAL FEATURE DISTRIBUTIONS")
print("=" * 78)

print("\n1. BIOLOGICAL FEATURE SUMMARY")

print(
    panel[features]
    .describe()
    .T[["min", "25%", "50%", "75%", "max", "std"]]
    .round(6)
    .to_string()
)

print("\n2. CORRELATIONS WITH MALARIA INCIDENCE")

for feature in features:
    valid = panel[[feature, "incidence_b"]].dropna()

    print(
        f"{feature:28s} "
        f"Pearson={valid[feature].corr(valid['incidence_b']): .4f} "
        f"Spearman={valid[feature].corr(valid['incidence_b'], method='spearman'): .4f}"
    )

print("\n3. CORRELATIONS AMONG BIOLOGICAL FEATURES")

print(
    panel[features]
    .corr()
    .round(3)
    .to_string()
)

print("\n4. FULL VS PARTIAL SUH SUPPORT")

panel["support_group"] = np.where(
    panel["supported_months"] == 12,
    "Full support",
    "Partial support"
)

print(
    panel.groupby("support_group")[
        features + ["incidence_b"]
    ]
    .median()
    .round(5)
    .to_string()
)

print("\n5. VILLENA FULL VS NO-EFD SENSITIVITY")

difference = (
    panel["Villena_full"]
    - panel["Villena_noEFD"]
)

print("Mean difference:", difference.mean())
print("Median difference:", difference.median())
print("Maximum absolute difference:", difference.abs().max())
print(
    "Pearson correlation:",
    panel["Villena_full"].corr(panel["Villena_noEFD"])
)

print("\n" + "=" * 78)
print("AUDIT COMPLETE — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 77 — TEMPERATURE EXPLANATORY REDUNDANCY AUDIT
# BIML 2.0 | READ ONLY
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

ROOT = Path("/home/gato/Desktop/BIML_Research")

who = pd.read_csv(
    "/home/gato/Desktop/BIML_/data/BIML_WHO_core_2015_2024.csv"
)

b1 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage05/"
    "b1_monthly_posterior_v1.csv"
)

b2 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage10/"
    "stage10_b2_corrected_monthly.csv"
)

study = who[["Country", "Year"]].drop_duplicates()

b1 = b1.merge(
    study, on=["Country", "Year"],
    validate="many_to_one"
)

b2 = b2.merge(
    study, on=["Country", "Year"],
    validate="many_to_one"
)

b1_annual = (
    b1.groupby(["Country", "Year"])
    .agg(
        Villena_full=("B1_posterior_full_median", "mean"),
        Villena_noEFD=(
            "B1_posterior_noEFD_sensitivity_median", "mean"
        )
    )
    .reset_index()
)

b2_annual = (
    b2.groupby(["Country", "Year"])
    .agg(
        Suh_competence=(
            "B2_Suh_vector_competence_pooled_median", "mean"
        ),
        Suh_EIP_days=(
            "B2_Suh_EIP_mean_days_pooled_median", "mean"
        ),
        Suh_biting=(
            "B2_An_stephensi_biting_median", "mean"
        ),
        Suh_deterministic_index=(
            "B2_deterministic_competence_biting_survival_index",
            "mean"
        ),
        supported_months=("B2_temperature_supported", "sum")
    )
    .reset_index()
)

panel = (
    who.merge(
        b1_annual,
        on=["Country", "Year"],
        validate="one_to_one"
    )
    .merge(
        b2_annual,
        on=["Country", "Year"],
        validate="one_to_one"
    )
)

features = [
    "Villena_full",
    "Villena_noEFD",
    "Suh_competence",
    "Suh_EIP_days",
    "Suh_biting",
    "Suh_deterministic_index"
]

print("=" * 78)
print("CELL 77 — TEMPERATURE EXPLANATORY REDUNDANCY")
print("=" * 78)

results = []

for feature in features:

    subset = panel[
        ["ERA5_T2M_C_mean", feature]
    ].dropna()

    X = subset[["ERA5_T2M_C_mean"]].values
    y = subset[feature].values

    row = {"Feature": feature, "N": len(subset)}

    for degree in [1, 2, 3, 4]:

        poly = PolynomialFeatures(
            degree=degree,
            include_bias=False
        )

        X_poly = poly.fit_transform(X)

        model = LinearRegression()
        model.fit(X_poly, y)

        predictions = model.predict(X_poly)

        row[f"R2_degree_{degree}"] = r2_score(
            y, predictions
        )

    results.append(row)

result_df = pd.DataFrame(results)

print("\n1. IN-SAMPLE R²: ANNUAL TEMPERATURE EXPLAINING BIOLOGY")
print(
    result_df.round(4).to_string(index=False)
)

print("\n2. ANNUAL TEMPERATURE RANGE")
print(
    panel["ERA5_T2M_C_mean"]
    .describe()
    .round(3)
    .to_string()
)

print("\n3. SUPPORT COVERAGE VS ANNUAL TEMPERATURE")
print(
    "Pearson:",
    round(
        panel["supported_months"].corr(
            panel["ERA5_T2M_C_mean"]
        ),
        4
    )
)

print("\n" + "=" * 78)
print("AUDIT COMPLETE — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 78 — EXISTING BIML EXPERIMENT RECOVERY
# BIML 2.0 | READ ONLY
# ============================================================

from pathlib import Path

ROOT = Path("/home/gato/Desktop/BIML_Research")

search_roots = [
    ROOT / "biml_pipeline",
    ROOT / "BIML_Stage10_B2_Semantic_Correction",
]

script_extensions = {".py", ".ipynb", ".r", ".R"}
result_extensions = {".csv", ".json", ".parquet"}

keywords = (
    "experiment", "baseline", "forecast",
    "evaluation", "ablation", "model",
    "stage11", "stage12", "stage13",
    "temporal", "split", "ridge"
)

print("=" * 78)
print("CELL 78 — EXISTING BIML MODELLING INVENTORY")
print("=" * 78)

for root in search_roots:

    print(f"\nDIRECTORY: {root}")
    print("EXISTS:", root.exists())

    if not root.exists():
        continue

    scripts = []
    results = []

    for path in root.rglob("*"):

        if not path.is_file():
            continue

        name = path.name.lower()

        if not any(word in name for word in keywords):
            continue

        if path.suffix in script_extensions:
            scripts.append(path)

        elif path.suffix.lower() in result_extensions:
            results.append(path)

    print("\nMATCHING SCRIPTS:", len(scripts))

    for path in sorted(scripts)[:60]:
        print(" SCRIPT:", path.relative_to(ROOT))

    print("\nMATCHING RESULT FILES:", len(results))

    for path in sorted(results)[:80]:
        size_kb = path.stat().st_size / 1024
        print(
            f" RESULT: {path.relative_to(ROOT)} "
            f"({size_kb:.1f} KB)"
        )

print("\n" + "=" * 78)
print("INVENTORY COMPLETE — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 81 — WHO AND HISTORICAL INCIDENCE COMPATIBILITY
# BIML 2.0 | READ ONLY
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

WHO_PATH = Path(
    "/home/gato/Desktop/BIML_/data/"
    "BIML_WHO_core_2015_2024.csv"
)

OLD_PATH = Path(
    "/home/gato/Desktop/BIML_Research/"
    "biml_pipeline/outputs/stage11/"
    "stage11_supervised_full384.csv"
)

who = pd.read_csv(WHO_PATH)
old = pd.read_csv(OLD_PATH)

who["Country"] = who["Country"].astype(str).str.strip()
old["Country"] = old["Country"].astype(str).str.strip()

who["Year"] = who["Year"].astype(int)
old["Year"] = old["Year"].astype(int)

print("=" * 76)
print("CELL 81 — INCIDENCE COMPATIBILITY AUDIT")
print("=" * 76)

who_countries = set(who["Country"])
old_countries = set(old["Country"])

print("\nCOUNTRY COVERAGE")
print("WHO countries:", len(who_countries))
print("Original countries:", len(old_countries))
print("Common countries:", len(who_countries & old_countries))
print("WHO only:", sorted(who_countries - old_countries))
print("Original only:", sorted(old_countries - who_countries))

merged = who[
    ["Country", "Year", "incidence_b"]
].merge(
    old[
        [
            "Country",
            "Year",
            "Malaria_incidence_per_1000",
            "Incidence_lag1"
        ]
    ],
    on=["Country", "Year"],
    how="inner",
    validate="one_to_one"
)

print("\nOVERLAPPING OBSERVATIONS:", len(merged))

if not merged.empty:

    merged["difference"] = (
        merged["incidence_b"]
        - merged["Malaria_incidence_per_1000"]
    )

    print("\nTARGET COMPARISON")
    print("Maximum absolute difference:",
          merged["difference"].abs().max())
    print("Mean absolute difference:",
          merged["difference"].abs().mean())
    print("Exact matches:",
          (merged["difference"] == 0).sum())
    print("Matches within 1e-6:",
          (merged["difference"].abs() <= 1e-6).sum())

    print("\nLARGEST DIFFERENCES")
    print(
        merged.assign(
            abs_difference=merged["difference"].abs()
        )
        .sort_values("abs_difference", ascending=False)
        [
            [
                "Country", "Year",
                "incidence_b",
                "Malaria_incidence_per_1000",
                "difference"
            ]
        ]
        .head(15)
        .to_string(index=False)
    )

    print("\nHISTORICAL LAG AVAILABILITY")
    print(
        merged["Incidence_lag1"]
        .describe()
        .to_string()
    )

    print("Missing lag values:",
          merged["Incidence_lag1"].isna().sum())

    previous = who[
        ["Country", "Year", "incidence_b"]
    ].copy()

    previous["Year"] += 1

    previous = previous.rename(
        columns={"incidence_b": "WHO_Previous_Incidence"}
    )

    check = merged.merge(
        previous,
        on=["Country", "Year"],
        how="left",
        validate="one_to_one"
    )

    comparable = check.dropna(
        subset=["WHO_Previous_Incidence", "Incidence_lag1"]
    ).copy()

    comparable["lag_difference"] = (
        comparable["Incidence_lag1"]
        - comparable["WHO_Previous_Incidence"]
    )

    print("\nHISTORICAL LAG VS WHO PREVIOUS YEAR")
    print("Comparable rows:", len(comparable))

    if not comparable.empty:
        print("Maximum absolute difference:",
              comparable["lag_difference"].abs().max())
        print("Mean absolute difference:",
              comparable["lag_difference"].abs().mean())
        print("Matches within 1e-6:",
              (comparable["lag_difference"].abs() <= 1e-6).sum())

print("\n" + "=" * 76)
print("AUDIT COMPLETE — NO FILES MODIFIED")
print("=" * 76)
```

```python

# ============================================================
# CELL 82 — PROSPECTIVE TARGET AND TEMPORAL COHORT AUDIT
# BIML 2.0 | READ ONLY
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

WHO_PATH = Path(
    "/home/gato/Desktop/BIML_/data/"
    "BIML_WHO_core_2015_2024.csv"
)

df = pd.read_csv(WHO_PATH)

df["Country"] = df["Country"].astype(str).str.strip()
df["Year"] = df["Year"].astype(int)

df = df.sort_values(["Country", "Year"]).reset_index(drop=True)

assert df.duplicated(["Country", "Year"]).sum() == 0
assert df["incidence_b"].notna().all()

# Predictor year t; target year t+1
future = df[
    ["Country", "Year", "incidence_b"]
].copy()

future["Year"] = future["Year"] - 1

future = future.rename(
    columns={"incidence_b": "Incidence_t_plus_1"}
)

panel = df.merge(
    future,
    on=["Country", "Year"],
    how="inner",
    validate="one_to_one"
)

panel["Target_Year"] = panel["Year"] + 1

panel["Partition"] = np.select(
    [
        panel["Target_Year"].between(2016, 2020),
        panel["Target_Year"].between(2021, 2022),
        panel["Target_Year"].between(2023, 2024)
    ],
    ["Train", "Validation", "Test"],
    default="Excluded"
)

print("=" * 76)
print("CELL 82 — PROSPECTIVE TARGET ALIGNMENT")
print("=" * 76)

print("\nPanel shape:", panel.shape)
print("Countries:", panel["Country"].nunique())
print("Predictor years:", panel["Year"].min(), "-", panel["Year"].max())
print("Target years:", panel["Target_Year"].min(), "-", panel["Target_Year"].max())

print("\nPARTITION COUNTS")
print(
    panel.groupby("Partition")
    .agg(
        observations=("Incidence_t_plus_1", "size"),
        countries=("Country", "nunique"),
        target_min=("Target_Year", "min"),
        target_max=("Target_Year", "max"),
        target_mean=("Incidence_t_plus_1", "mean"),
        target_std=("Incidence_t_plus_1", "std")
    )
    .to_string()
)

print("\nYEAR-BY-YEAR TARGET SUMMARY")
print(
    panel.groupby("Target_Year")["Incidence_t_plus_1"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .round(3)
    .to_string()
)

# Persistence baseline: predict next year using current incidence
panel["Persistence_Error"] = (
    panel["Incidence_t_plus_1"] - panel["incidence_b"]
)

print("\nPERSISTENCE BASELINE BY PARTITION")

for part in ["Train", "Validation", "Test"]:
    subset = panel.loc[panel["Partition"] == part]

    mae = subset["Persistence_Error"].abs().mean()
    rmse = np.sqrt(
        np.mean(subset["Persistence_Error"] ** 2)
    )

    print(
        f"{part:10s} | N={len(subset):3d} "
        f"| MAE={mae:.4f} | RMSE={rmse:.4f}"
    )

print("\nALIGNMENT CHECKS")
print("Unique predictor keys:",
      not panel.duplicated(["Country", "Year"]).any())
print("Unique target keys:",
      not panel.duplicated(["Country", "Target_Year"]).any())
print("All target years = predictor year + 1:",
      (panel["Target_Year"] == panel["Year"] + 1).all())
print("All targets finite:",
      np.isfinite(panel["Incidence_t_plus_1"]).all())

assert len(panel) == 261
assert (panel["Partition"] != "Excluded").all()

print("\nAUDIT COMPLETE — NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 82 — PROSPECTIVE SPLIT AND PERSISTENCE BASELINE
# BIML 2.0 | NEW WHO INCIDENCE ONLY | READ ONLY
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd

WHO_PATH = Path(
    "/home/gato/Desktop/BIML_/data/"
    "BIML_WHO_core_2015_2024.csv"
)

df = pd.read_csv(WHO_PATH)

df["Country"] = df["Country"].astype(str).str.strip()
df["Year"] = df["Year"].astype(int)

df = df.sort_values(
    ["Country", "Year"]
).reset_index(drop=True)

assert len(df) == 290
assert df["Country"].nunique() == 29
assert not df.duplicated(["Country", "Year"]).any()
assert np.isfinite(df["incidence_b"]).all()

# Predictors are measured in year t.
# Target is WHO malaria incidence in year t+1.
future = df[
    ["Country", "Year", "incidence_b"]
].copy()

future["Year"] -= 1
future = future.rename(
    columns={"incidence_b": "Incidence_t_plus_1"}
)

panel = df.merge(
    future,
    on=["Country", "Year"],
    how="inner",
    validate="one_to_one"
)

panel["Target_Year"] = panel["Year"] + 1

# Proposed chronological partitions
panel["Partition"] = np.select(
    [
        panel["Target_Year"].between(2016, 2020),
        panel["Target_Year"].between(2021, 2022),
        panel["Target_Year"].between(2023, 2024)
    ],
    ["Train", "Validation", "Test"],
    default="Excluded"
)

# Persistence: predict incidence(t+1) = incidence(t)
panel["Persistence_Prediction"] = panel["incidence_b"]
panel["Persistence_Error"] = (
    panel["Incidence_t_plus_1"]
    - panel["Persistence_Prediction"]
)

print("=" * 76)
print("CELL 82 — PROSPECTIVE TEMPORAL EVALUATION")
print("=" * 76)

print("\nDATASET")
print("Country-year pairs:", len(panel))
print("Countries:", panel["Country"].nunique())
print("Predictor years:", panel["Year"].min(), "-", panel["Year"].max())
print("Target years:",
      panel["Target_Year"].min(), "-", panel["Target_Year"].max())

print("\nTEMPORAL PARTITIONS")
print(
    panel.groupby("Partition")
    .agg(
        N=("Incidence_t_plus_1", "size"),
        Countries=("Country", "nunique"),
        First_Target_Year=("Target_Year", "min"),
        Last_Target_Year=("Target_Year", "max"),
        Mean_Incidence=("Incidence_t_plus_1", "mean"),
        SD_Incidence=("Incidence_t_plus_1", "std")
    )
    .round(3)
    .to_string()
)

print("\nYEAR-BY-YEAR TARGET SUMMARY")
print(
    panel.groupby("Target_Year")["Incidence_t_plus_1"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .round(3)
    .to_string()
)

print("\nPERSISTENCE BASELINE")

for partition in ["Train", "Validation", "Test"]:

    subset = panel.loc[
        panel["Partition"] == partition
    ]

    y_true = subset["Incidence_t_plus_1"].to_numpy()
    y_pred = subset["Persistence_Prediction"].to_numpy()

    mae = np.mean(np.abs(y_true - y_pred))
    rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))

    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - y_true.mean()) ** 2)

    r2 = 1 - ss_res / ss_tot if ss_tot > 0 else np.nan

    print(
        f"{partition:10s} | "
        f"N={len(subset):3d} | "
        f"MAE={mae:8.3f} | "
        f"RMSE={rmse:8.3f} | "
        f"R2={r2:8.4f}"
    )

print("\nINTEGRITY CHECKS")
print("261 pairs:", len(panel) == 261)
print("29 countries:", panel["Country"].nunique() == 29)
print("No excluded rows:", (panel["Partition"] != "Excluded").all())
print(
    "Correct target alignment:",
    (panel["Target_Year"] == panel["Year"] + 1).all()
)
print(
    "Unique country-target-year keys:",
    not panel.duplicated(["Country", "Target_Year"]).any()
)
print(
    "All targets finite:",
    np.isfinite(panel["Incidence_t_plus_1"]).all()
)

assert len(panel) == 261
assert (panel["Partition"] != "Excluded").all()

print("\nCELL 82 COMPLETE — NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 83 — BIML 2.0 FEATURE ALIGNMENT AUDIT
# NEW WHO PANEL + EXISTING B1/B2 BIOLOGICAL FEATURES
# READ ONLY
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path("/home/gato/Desktop/BIML_Research")

paths = {
    "WHO": Path(
        "/home/gato/Desktop/BIML_/data/"
        "BIML_WHO_core_2015_2024.csv"
    ),
    "B1": ROOT / (
        "biml_pipeline/outputs/stage05/"
        "b1_monthly_posterior_v1.csv"
    ),
    "B2": ROOT / (
        "biml_pipeline/outputs/stage10/"
        "stage10_b2_corrected_monthly.csv"
    )
}

print("=" * 78)
print("CELL 83 — FEATURE ALIGNMENT AUDIT")
print("=" * 78)

datasets = {}

for name, path in paths.items():

    print(f"\n{name} DATASET")
    print("Path:", path)
    print("Exists:", path.exists())

    if not path.exists():
        continue

    df = pd.read_csv(path)

    datasets[name] = df

    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())

    if "Country" in df.columns:
        print("Countries:", df["Country"].nunique())

    if "Year" in df.columns:
        print("Year range:", df["Year"].min(), "-", df["Year"].max())

    keys = [
        c for c in ["Country", "Year", "Month"]
        if c in df.columns
    ]

    print("Keys:", keys)
    print("Duplicate keys:", df.duplicated(keys).sum())

if all(k in datasets for k in ["WHO", "B1", "B2"]):

    who = datasets["WHO"].copy()
    b1 = datasets["B1"].copy()
    b2 = datasets["B2"].copy()

    for df in [who, b1, b2]:
        df["Country"] = df["Country"].astype(str).str.strip()
        df["Year"] = df["Year"].astype(int)

    # Restrict monthly datasets to WHO country-years.
    annual_keys = who[["Country", "Year"]].drop_duplicates()

    b1 = b1.merge(
        annual_keys,
        on=["Country", "Year"],
        how="inner",
        validate="many_to_one"
    )

    b2 = b2.merge(
        annual_keys,
        on=["Country", "Year"],
        how="inner",
        validate="many_to_one"
    )

    print("\n" + "=" * 78)
    print("MATCHED WHO STUDY POPULATION")
    print("=" * 78)

    print("WHO annual rows:", len(who))
    print("B1 monthly rows:", len(b1))
    print("B2 monthly rows:", len(b2))

    for name, df in [("B1", b1), ("B2", b2)]:

        counts = df.groupby(
            ["Country", "Year"]
        ).size()

        print(f"\n{name} MONTHLY COVERAGE")
        print("Country-years:", len(counts))
        print("Minimum months:", counts.min())
        print("Maximum months:", counts.max())
        print("Exactly 12 months:", (counts == 12).sum())

        missing_keys = annual_keys.merge(
            df[["Country", "Year"]].drop_duplicates(),
            on=["Country", "Year"],
            how="left",
            indicator=True
        )

        print(
            "Missing WHO country-years:",
            (missing_keys["_merge"] == "left_only").sum()
        )

    # Compare the two biological datasets on identical months.
    monthly = b1.merge(
        b2,
        on=["Country", "Year", "Month"],
        how="outer",
        indicator=True,
        suffixes=("_B1", "_B2"),
        validate="one_to_one"
    )

    print("\nB1/B2 MONTHLY JOIN")
    print(monthly["_merge"].value_counts().to_string())

    print("\nTEMPERATURE COLUMNS")
    for name, df in [("WHO", who), ("B1", b1), ("B2", b2)]:
        cols = [
            c for c in df.columns
            if "temp" in c.lower() or "t2m" in c.lower()
        ]
        print(name, ":", cols)

    print("\nBIOLOGICAL FEATURE COLUMNS")

    for name, df in [("B1", b1), ("B2", b2)]:
        biological = [
            c for c in df.columns
            if c not in ["Country", "Year", "Month"]
        ]
        print(f"\n{name}:")
        for col in biological:
            print(" ", col)

print("\n" + "=" * 78)
print("CELL 83 COMPLETE — NO FILES MODIFIED")
print("=" * 78)
```

```python

# ============================================================
# CELL 84 — REVISED BIML ANNUAL BIOLOGICAL FEATURE MATRIX
# BIML 2.0 | NEW WHO STUDY POPULATION | READ ONLY
# ============================================================

from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path("/home/gato/Desktop/BIML_Research")

who = pd.read_csv(
    "/home/gato/Desktop/BIML_/data/"
    "BIML_WHO_core_2015_2024.csv"
)

b1 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage05/"
    "b1_monthly_posterior_v1.csv"
)

b2 = pd.read_csv(
    ROOT / "biml_pipeline/outputs/stage10/"
    "stage10_b2_corrected_monthly.csv"
)

for df in (who, b1, b2):
    df["Country"] = df["Country"].astype(str).str.strip()
    df["Year"] = df["Year"].astype(int)

keys = ["Country", "Year"]

# Restrict biological data to WHO country-years.
study_keys = who[keys].drop_duplicates()

b1 = b1.merge(
    study_keys, on=keys, how="inner",
    validate="many_to_one"
)

b2 = b2.merge(
    study_keys, on=keys, how="inner",
    validate="many_to_one"
)

# Villena annual features: 12-month means of monthly
# posterior summaries, NOT an annual posterior distribution.
b1_annual = (
    b1.groupby(keys, as_index=False)
    .agg(
        B1_full_mean=(
            "B1_posterior_full_median", "mean"
        ),
        B1_full_sd_monthly=(
            "B1_posterior_full_median", "std"
        ),
        B1_noEFD_mean=(
            "B1_posterior_noEFD_sensitivity_median", "mean"
        ),
        B1_months=("Month", "nunique")
    )
)

# Preserve the exact Suh temperature-support mask.
b2["Supported"] = (
    pd.to_numeric(
        b2["B2_temperature_supported"],
        errors="coerce"
    ) == 1
)

suh_columns = {
    "B2_competence_supported_mean":
        "B2_Suh_vector_competence_pooled_median",
    "B2_EIP_supported_mean":
        "B2_Suh_EIP_mean_days_pooled_median",
    "B2_biting_supported_mean":
        "B2_An_stephensi_biting_median",
    "B2_index_supported_mean":
        "B2_deterministic_competence_biting_survival_index"
}

# Mask unsupported months explicitly.
for output_col, source_col in suh_columns.items():
    b2[output_col] = pd.to_numeric(
        b2[source_col], errors="coerce"
    ).where(b2["Supported"])

agg = {
    output_col: (output_col, "mean")
    for output_col in suh_columns
}
agg["B2_supported_months"] = ("Supported", "sum")
agg["B2_total_months"] = ("Month", "nunique")

b2_annual = (
    b2.groupby(keys, as_index=False)
    .agg(**agg)
)

b2_annual["B2_full_year_supported"] = (
    b2_annual["B2_supported_months"] == 12
)

# Join to WHO. WHO incidence is the ONLY outcome.
annual = (
    who.merge(
        b1_annual, on=keys,
        how="left", validate="one_to_one"
    )
    .merge(
        b2_annual, on=keys,
        how="left", validate="one_to_one"
    )
)

print("=" * 78)
print("CELL 84 — REVISED ANNUAL BIOLOGICAL MATRIX")
print("=" * 78)

print("\nSHAPE:", annual.shape)
print("COUNTRIES:", annual["Country"].nunique())
print("YEARS:", annual["Year"].min(), "-", annual["Year"].max())

print("\nVILLENA COVERAGE")
print(annual["B1_months"].value_counts().sort_index())

print("\nSUH SUPPORTED MONTHS")
print(
    annual["B2_supported_months"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nFULL-YEAR SUH SUPPORT")
print(
    annual["B2_full_year_supported"]
    .value_counts()
    .to_string()
)

print("\nMISSING BIOLOGICAL FEATURES")
print(
    annual[
        ["B1_full_mean", "B1_noEFD_mean"]
        + list(suh_columns)
    ]
    .isna()
    .sum()
    .to_string()
)

print("\nBIOLOGICAL FEATURE SUMMARY")
print(
    annual[
        ["B1_full_mean", "B1_noEFD_mean"]
        + list(suh_columns)
    ]
    .describe()
    .round(5)
    .to_string()
)

print("\nINTEGRITY CHECKS")
checks = {
    "290 annual rows": len(annual) == 290,
    "29 countries": annual["Country"].nunique() == 29,
    "Unique annual keys":
        not annual.duplicated(keys).any(),
    "12 Villena months":
        annual["B1_months"].eq(12).all(),
    "12 Suh calendar months":
        annual["B2_total_months"].eq(12).all(),
    "WHO target unchanged":
        annual["incidence_b"].notna().all(),
    "No old incidence columns":
        not any(
            c in annual.columns
            for c in [
                "Malaria_incidence_per_1000",
                "Incidence_lag1"
            ]
        )
}

for name, passed in checks.items():
    print(f"{name}: {passed}")

assert all(checks.values())

print("\nCELL 84 COMPLETE — NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 85 — PROSPECTIVE BIML FEATURE MATRIX
# BIML 2.0 | ONE-YEAR-AHEAD | READ ONLY
# Requires: annual DataFrame from Cell 84
# ============================================================

import numpy as np
import pandas as pd

assert "annual" in globals(), "Run Cell 84 first."

keys = ["Country", "Year"]

assert len(annual) == 290
assert not annual.duplicated(keys).any()

# All predictor columns originate in year t.
predictors = [
    "Country",
    "Year",
    "incidence_b",
    "ERA5_T2M_C_mean",
    "Rainfall_total_mm",
    "RH_pct_mean",
    "NDVI_mean",
    "NDWI_mean",
    "MODIS_Night_LST_C_mean",
    "Landsat_LST_C_mean",
    "itn_use",
    "itn_access",
    "B1_full_mean",
    "B1_noEFD_mean",
    "B2_competence_supported_mean",
    "B2_EIP_supported_mean",
    "B2_biting_supported_mean",
    "B2_index_supported_mean",
    "B2_supported_months",
    "B2_full_year_supported"
]

missing = [c for c in predictors if c not in annual.columns]
assert not missing, f"Missing columns: {missing}"

# Prepare next-year WHO incidence.
future = annual[
    ["Country", "Year", "incidence_b"]
].copy()

future["Year"] -= 1

future = future.rename(
    columns={"incidence_b": "Target_Incidence_t_plus_1"}
)

# Construct prospective observations.
prospective = annual[predictors].merge(
    future,
    on=["Country", "Year"],
    how="inner",
    validate="one_to_one"
)

prospective["Target_Year"] = prospective["Year"] + 1

prospective["Partition"] = np.select(
    [
        prospective["Target_Year"].between(2016, 2020),
        prospective["Target_Year"].between(2021, 2022),
        prospective["Target_Year"].between(2023, 2024)
    ],
    ["Train", "Validation", "Test"],
    default="Excluded"
)

# Explicit interpretation of incidence_b in predictor year.
prospective = prospective.rename(
    columns={"incidence_b": "Incidence_t"}
)

# The existing Suh annual summaries are supported-month means.
# Keep their coverage indicators for audit and sensitivity analyses.
prospective["B2_partial_year"] = (
    ~prospective["B2_full_year_supported"].astype(bool)
)

print("=" * 78)
print("CELL 85 — PROSPECTIVE FEATURE MATRIX")
print("=" * 78)

print("\nSHAPE:", prospective.shape)
print("COUNTRIES:", prospective["Country"].nunique())
print(
    "PREDICTOR YEARS:",
    prospective["Year"].min(),
    prospective["Year"].max()
)
print(
    "TARGET YEARS:",
    prospective["Target_Year"].min(),
    prospective["Target_Year"].max()
)

print("\nPARTITION COUNTS")
print(
    prospective.groupby("Partition")
    .agg(
        N=("Target_Incidence_t_plus_1", "size"),
        Countries=("Country", "nunique"),
        First_Target=("Target_Year", "min"),
        Last_Target=("Target_Year", "max"),
        Partial_Suh=("B2_partial_year", "sum")
    )
    .to_string()
)

print("\nMISSING VALUES")
print(
    prospective.isna()
    .sum()
    .loc[lambda s: s > 0]
    .to_string()
    if prospective.isna().any().any()
    else "No missing values"
)

print("\nSUH SUPPORT BY PARTITION")
print(
    pd.crosstab(
        prospective["Partition"],
        prospective["B2_full_year_supported"]
    ).to_string()
)

print("\nFEATURE GROUPS")
feature_groups = {
    "H": ["Incidence_t"],
    "E": [
        "ERA5_T2M_C_mean",
        "Rainfall_total_mm",
        "RH_pct_mean",
        "NDVI_mean",
        "NDWI_mean",
        "MODIS_Night_LST_C_mean",
        "Landsat_LST_C_mean"
    ],
    "I": ["itn_use", "itn_access"],
    "B1": ["B1_full_mean"],
    "B1_sensitivity": ["B1_noEFD_mean"],
    "B2": [
        "B2_competence_supported_mean",
        "B2_EIP_supported_mean",
        "B2_biting_supported_mean",
        "B2_index_supported_mean"
    ]
}

for name, columns in feature_groups.items():
    print(f"{name}: {columns}")

print("\nINTEGRITY CHECKS")

checks = {
    "261 prospective observations":
        len(prospective) == 261,
    "29 countries":
        prospective["Country"].nunique() == 29,
    "No duplicate predictor keys":
        not prospective.duplicated(["Country", "Year"]).any(),
    "No duplicate target keys":
        not prospective.duplicated(
            ["Country", "Target_Year"]
        ).any(),
    "Correct temporal alignment":
        (prospective["Target_Year"] ==
         prospective["Year"] + 1).all(),
    "All targets finite":
        np.isfinite(
            prospective["Target_Incidence_t_plus_1"]
        ).all(),
    "No excluded partitions":
        prospective["Partition"].ne("Excluded").all(),
    "No old incidence fields":
        not any(
            c in prospective.columns
            for c in [
                "Malaria_incidence_per_1000",
                "Incidence_lag1"
            ]
        )
}

for name, passed in checks.items():
    print(f"{name}: {passed}")

assert all(checks.values())

print("\nCELL 85 COMPLETE — NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 86 — FEATURE INTEGRITY AND REDUNDANCY AUDIT
# BIML 2.0 | PROSPECTIVE DATA | READ ONLY
# Requires: prospective and feature_groups from Cell 85
# ============================================================

import numpy as np
import pandas as pd

assert "prospective" in globals()
assert "feature_groups" in globals()

df = prospective.copy()

groups = feature_groups

feature_columns = (
    groups["H"]
    + groups["E"]
    + groups["I"]
    + groups["B1"]
    + groups["B1_sensitivity"]
    + groups["B2"]
    + ["B2_supported_months"]
)

print("=" * 78)
print("CELL 86 — FEATURE INTEGRITY AND REDUNDANCY AUDIT")
print("=" * 78)

print("\nFEATURE DISTRIBUTIONS")
summary = df[feature_columns].agg(
    ["count", "mean", "std", "min", "median", "max"]
).T

print(summary.round(4).to_string())

print("\nCONSTANT OR NEAR-CONSTANT FEATURES")
for col in feature_columns:
    unique = df[col].nunique(dropna=True)
    std = df[col].std()
    if unique <= 2 or np.isclose(std, 0):
        print(col, "| unique:", unique, "| std:", std)

print("\nINTERVENTION INDICATORS")
print(
    df[groups["I"]]
    .describe()
    .round(4)
    .to_string()
)

print("\nINTERVENTION CORRELATION")
print(
    df[groups["I"]]
    .corr()
    .round(4)
    .to_string()
)

print("\nENVIRONMENTAL FEATURE CORRELATIONS")
print(
    df[groups["E"]]
    .corr()
    .round(3)
    .to_string()
)

print("\nBIOLOGICAL FEATURE CORRELATIONS WITH TEMPERATURE")

temperature = "ERA5_T2M_C_mean"

biological_columns = (
    groups["B1"]
    + groups["B1_sensitivity"]
    + groups["B2"]
)

for col in biological_columns:
    corr = df[[temperature, col]].corr().iloc[0, 1]
    print(f"{col:38s} Pearson r = {corr: .4f}")

print("\nBIOLOGICAL FEATURE CORRELATION MATRIX")
print(
    df[biological_columns]
    .corr()
    .round(3)
    .to_string()
)

print("\nSUH SUPPORT AND TEMPERATURE")
print(
    df.groupby("B2_full_year_supported")
    .agg(
        N=("Country", "size"),
        Mean_Temperature=(temperature, "mean"),
        SD_Temperature=(temperature, "std"),
        Mean_Supported_Months=(
            "B2_supported_months", "mean"
        ),
        Mean_Incidence_t=("Incidence_t", "mean"),
        Mean_Target=(
            "Target_Incidence_t_plus_1", "mean"
        )
    )
    .round(3)
    .to_string()
)

print("\nSUH SUPPORT BY COUNTRY")
support_country = (
    df.groupby("Country")
    .agg(
        Observations=("Year", "size"),
        Partial_Years=("B2_partial_year", "sum"),
        Mean_Supported_Months=(
            "B2_supported_months", "mean"
        )
    )
    .sort_values(
        ["Partial_Years", "Country"],
        ascending=[False, True]
    )
)

print(support_country.to_string())

print("\nPARTITION-WISE FEATURE AVAILABILITY")
for part in ["Train", "Validation", "Test"]:
    sub = df.loc[df["Partition"] == part]
    print(
        f"{part:10s} | N={len(sub):3d} | "
        f"Missing={int(sub[feature_columns].isna().sum().sum())} | "
        f"Partial Suh={int(sub['B2_partial_year'].sum())}"
    )

print("\nCELL 86 COMPLETE — NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 87 — ITN REDUNDANCY AND TEMPORAL VARIATION AUDIT
# BIML 2.0 | NEW WHO DATA ONLY | READ ONLY
# Requires: prospective from Cell 85
# ============================================================

import numpy as np
import pandas as pd

assert "prospective" in globals()

df = prospective.copy().sort_values(["Country", "Year"])

print("=" * 78)
print("CELL 87 — INTERVENTION INDICATOR AUDIT")
print("=" * 78)

x = df["itn_use"].to_numpy(dtype=float)
y = df["itn_access"].to_numpy(dtype=float)

# Relationship between the two ITN indicators.
slope, intercept = np.polyfit(x, y, 1)
fitted = slope * x + intercept

print("\nITN USE VS ACCESS")
print("Pearson correlation:", np.corrcoef(x, y)[0, 1])
print("Linear slope:", slope)
print("Linear intercept:", intercept)
print("Maximum absolute residual:", np.max(np.abs(y - fitted)))
print("Mean absolute residual:", np.mean(np.abs(y - fitted)))

print("\nACCESS / USE RATIO")
ratio = y / x
print(pd.Series(ratio).describe().round(6).to_string())

print("\nWITHIN-COUNTRY TEMPORAL VARIATION")
country_stats = (
    df.groupby("Country")
    .agg(
        Use_Min=("itn_use", "min"),
        Use_Max=("itn_use", "max"),
        Use_SD=("itn_use", "std"),
        Access_SD=("itn_access", "std")
    )
)

country_stats["Use_Range"] = (
    country_stats["Use_Max"] - country_stats["Use_Min"]
)

print(country_stats.round(4).to_string())

print("\nCOUNTRIES WITH NEAR-ZERO ITN VARIATION")
print(
    (country_stats["Use_Range"] < 1e-6).sum(),
    "of",
    len(country_stats)
)

print("\nYEARLY MEAN ITN INDICATORS")
print(
    df.groupby("Year")[["itn_use", "itn_access"]]
    .mean()
    .round(3)
    .to_string()
)

print("\nYEAR-TO-YEAR CHANGES")

for col in ["itn_use", "itn_access"]:

    changes = df.groupby("Country")[col].diff()

    print(
        f"{col:12s} | "
        f"Mean absolute change={changes.abs().mean():.4f} | "
        f"Zero changes={(changes.dropna() == 0).sum()} | "
        f"Valid changes={changes.notna().sum()}"
    )

print("\nCOUNTRY AND YEAR EXPLANATORY POWER")

# Descriptive in-sample decomposition only.
# Not used for forecasting or feature selection.
for col in ["itn_use", "itn_access"]:

    values = df[col]
    grand_mean = values.mean()

    country_means = df.groupby("Country")[col].transform("mean")
    year_means = df.groupby("Year")[col].transform("mean")

    total_ss = np.sum((values - grand_mean) ** 2)

    country_r2 = 1 - (
        np.sum((values - country_means) ** 2) / total_ss
    )

    year_r2 = 1 - (
        np.sum((values - year_means) ** 2) / total_ss
    )

    print(
        f"{col:12s} | "
        f"Country-only R2={country_r2:.4f} | "
        f"Year-only R2={year_r2:.4f}"
    )

print("\nCELL 87 COMPLETE — NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 88 — FREEZE PROSPECTIVE EXPERIMENT PROTOCOL
# BIML 2.0 | NO MODEL FITTING | READ ONLY
# Requires: prospective and feature_groups
# ============================================================

import numpy as np
import pandas as pd

assert "prospective" in globals()
assert "feature_groups" in globals()

df = prospective.copy()

H = feature_groups["H"]
E = feature_groups["E"]
B1 = feature_groups["B1"]
B2 = feature_groups["B2"]

SUPPORT = ["B2_supported_months"]

# Explicit matched-support control:
# E + support vs E + support + B2.
# This isolates the incremental B2 contribution.
experiments = {
    "P1_H": H,
    "P2_E": E,
    "P3_E_B1": E + B1,
    "P4_E_SUPPORT": E + SUPPORT,
    "P4_E_SUPPORT_B2": E + SUPPORT + B2,
    "P5_H_E": H + E,
    "P6_H_E_B1": H + E + B1,
    "P7_H_E_SUPPORT": H + E + SUPPORT,
    "P7_H_E_SUPPORT_B2": H + E + SUPPORT + B2,
    "P8_H_E_SUPPORT_B1_B2":
        H + E + SUPPORT + B1 + B2,
}

sensitivity_experiments = {
    "S1_E_B1_noEFD":
        E + feature_groups["B1_sensitivity"],
    "S2_H_E_B1_noEFD":
        H + E + feature_groups["B1_sensitivity"],
    "S3_H_E_ITN":
        H + E + ["itn_use"],
    "S4_H_E_B1_ITN":
        H + E + B1 + ["itn_use"],
}

models = {
    "Ridge": {
        "alpha": 10.0
    },
    "RandomForest": {
        "n_estimators": 400,
        "min_samples_leaf": 3
    },
    "HistGradientBoosting": {
        "max_iter": 250,
        "learning_rate": 0.05,
        "max_leaf_nodes": 15,
        "l2_regularization": 1.0
    },
    "XGBoost": {
        "n_estimators": 300,
        "max_depth": 3,
        "learning_rate": 0.05,
        "reg_lambda": 5.0
    },
    "LightGBM": {
        "n_estimators": 300,
        "learning_rate": 0.05,
        "num_leaves": 7,
        "min_child_samples": 10,
        "reg_lambda": 5.0
    },
    "RBF_SVR": {
        "C": 10.0,
        "epsilon": 1.0
    }
}

RANDOM_STATE = 20261016

print("=" * 78)
print("CELL 88 — PROSPECTIVE EXPERIMENT PROTOCOL")
print("=" * 78)

print("\nPRIMARY EXPERIMENTS")
for name, columns in experiments.items():
    print(f"{name:30s} | {len(columns):2d} features")
    print("  ", columns)

print("\nSENSITIVITY EXPERIMENTS")
for name, columns in sensitivity_experiments.items():
    print(f"{name:30s} | {len(columns):2d} features")
    print("  ", columns)

print("\nMODEL FAMILIES")
for name, parameters in models.items():
    print(name, parameters)

print("\nTEMPORAL PARTITIONS")
print(
    df.groupby("Partition")
    .agg(
        N=("Country", "size"),
        Target_Start=("Target_Year", "min"),
        Target_End=("Target_Year", "max")
    )
    .to_string()
)

print("\nVALIDATION CHECKS")

all_sets = {
    **experiments,
    **sensitivity_experiments
}

checks = {
    "All feature columns exist":
        all(
            col in df.columns
            for cols in all_sets.values()
            for col in cols
        ),
    "No duplicated features":
        all(
            len(cols) == len(set(cols))
            for cols in all_sets.values()
        ),
    "No target in predictors":
        all(
            "Target_Incidence_t_plus_1" not in cols
            for cols in all_sets.values()
        ),
    "No old incidence predictors":
        all(
            col not in [
                "Malaria_incidence_per_1000",
                "Incidence_lag1"
            ]
            for cols in all_sets.values()
            for col in cols
        ),
    "ITN access excluded":
        all(
            "itn_access" not in cols
            for cols in all_sets.values()
        ),
    "261 observations":
        len(df) == 261,
    "Training years correct":
        set(
            df.loc[
                df["Partition"] == "Train",
                "Target_Year"
            ]
        ) == set(range(2016, 2021)),
    "Validation years correct":
        set(
            df.loc[
                df["Partition"] == "Validation",
                "Target_Year"
            ]
        ) == {2021, 2022},
    "Test years correct":
        set(
            df.loc[
                df["Partition"] == "Test",
                "Target_Year"
            ]
        ) == {2023, 2024}
}

for name, passed in checks.items():
    print(f"{name}: {passed}")

assert all(checks.values())

print("\nRANDOM STATE:", RANDOM_STATE)
print("PRIMARY EXPERIMENTS:", len(experiments))
print("SENSITIVITY EXPERIMENTS:", len(sensitivity_experiments))
print("MODEL FAMILIES:", len(models))

print("\nCELL 88 COMPLETE — PROTOCOL DEFINED IN MEMORY")
print("NO MODELS TRAINED — NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 89 — MODEL ENVIRONMENT AND PIPELINE PREPARATION
# BIML 2.0 | NO MODEL FITTING | NO FILES MODIFIED
# Requires: Cell 88
# ============================================================

import importlib.util
import numpy as np
import pandas as pd
import sklearn

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.ensemble import (
    RandomForestRegressor,
    HistGradientBoostingRegressor
)
from sklearn.svm import SVR

assert "prospective" in globals()
assert "experiments" in globals()
assert "models" in globals()
assert "RANDOM_STATE" in globals()

print("=" * 78)
print("CELL 89 — MODEL ENVIRONMENT AND PIPELINE PREPARATION")
print("=" * 78)

print("\nPACKAGE AVAILABILITY")
print("scikit-learn:", sklearn.__version__)

for package in ["xgboost", "lightgbm"]:
    spec = importlib.util.find_spec(package)
    print(f"{package}: {'AVAILABLE' if spec else 'MISSING'}")

# Stop explicitly if either frozen model family is unavailable.
assert importlib.util.find_spec("xgboost") is not None, (
    "XGBoost is missing. Install it in the active Jupyter environment."
)

assert importlib.util.find_spec("lightgbm") is not None, (
    "LightGBM is missing. Install it in the active Jupyter environment."
)

from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

def build_model(model_name):

    params = models[model_name].copy()

    if model_name == "Ridge":
        estimator = Ridge(**params)
        scaling = True

    elif model_name == "RandomForest":
        estimator = RandomForestRegressor(
            **params,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
        scaling = False

    elif model_name == "HistGradientBoosting":
        estimator = HistGradientBoostingRegressor(
            **params,
            random_state=RANDOM_STATE
        )
        scaling = False

    elif model_name == "XGBoost":
        estimator = XGBRegressor(
            **params,
            random_state=RANDOM_STATE,
            n_jobs=-1,
            objective="reg:squarederror",
            verbosity=0
        )
        scaling = False

    elif model_name == "LightGBM":
        estimator = LGBMRegressor(
            **params,
            random_state=RANDOM_STATE,
            n_jobs=-1,
            verbosity=-1
        )
        scaling = False

    elif model_name == "RBF_SVR":
        estimator = SVR(
            kernel="rbf",
            **params
        )
        scaling = True

    else:
        raise ValueError(
            f"Unknown model family: {model_name}"
        )

    steps = [
        ("imputer", SimpleImputer(strategy="median"))
    ]

    if scaling:
        steps.append(("scaler", StandardScaler()))

    steps.append(("model", estimator))

    return Pipeline(steps)


print("\nPIPELINE CONSTRUCTION")

pipelines = {}

for name in models:
    pipeline = build_model(name)
    pipelines[name] = pipeline

    print(
        f"{name:22s} | "
        + " -> ".join(pipeline.named_steps.keys())
    )

print("\nDATA PARTITIONS")

train = prospective[
    prospective["Partition"] == "Train"
].copy()

validation = prospective[
    prospective["Partition"] == "Validation"
].copy()

test = prospective[
    prospective["Partition"] == "Test"
].copy()

print("Train:", train.shape)
print("Validation:", validation.shape)
print("Test:", test.shape)

assert len(train) == 145
assert len(validation) == 58
assert len(test) == 58

print("\nFEATURE CONFIGURATION CHECK")

for name, columns in experiments.items():

    x_train = train[columns]

    assert np.isfinite(
        x_train.to_numpy(dtype=float)
    ).all()

    print(
        f"{name:30s} | "
        f"{x_train.shape[1]:2d} features | "
        f"{x_train.shape[0]:3d} training rows"
    )

print("\nFINAL STATUS")
print("Model families prepared:", len(pipelines))
print("Primary configurations:", len(experiments))
print("All models unfitted:", all(
    not hasattr(p.named_steps["model"], "n_features_in_")
    for p in pipelines.values()
))

print("\nCELL 89 COMPLETE — NO MODELS TRAINED")
```

```python

# ============================================================
# CELL 90 — PRIMARY MODEL TRAINING AND VALIDATION
# BIML 2.0 | 10 CONFIGURATIONS × 6 MODEL FAMILIES
# TRAIN: 2016–2020 | VALIDATION: 2021–2022
# TEST: NOT ACCESSED | NO FILES MODIFIED
# ============================================================

import time
import warnings
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "prospective" in globals()
assert "experiments" in globals()
assert "pipelines" in globals()

# Select only training and validation observations.
train90 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val90 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

target = "Target_Incidence_t_plus_1"

assert len(train90) == 145
assert len(val90) == 58
assert train90["Target_Year"].max() == 2020
assert val90["Target_Year"].min() == 2021

y_train = train90[target].to_numpy(dtype=float)
y_val = val90[target].to_numpy(dtype=float)

results90 = []
predictions90 = []
fitted90 = {}

print("=" * 78)
print("CELL 90 — PROSPECTIVE TRAINING AND VALIDATION")
print("=" * 78)

print("Training observations:", len(train90))
print("Validation observations:", len(val90))
print("Primary configurations:", len(experiments))
print("Model families:", len(pipelines))
print("Expected fits:", len(experiments) * len(pipelines))

start_all = time.time()

for experiment_name, features in experiments.items():

    X_train = train90[features]
    X_val = val90[features]

    print(f"\nEXPERIMENT: {experiment_name}")

    for model_name, pipeline in pipelines.items():

        start = time.time()

        estimator = clone(pipeline)

        with warnings.catch_warnings():
            warnings.filterwarnings(
                "ignore",
                category=UserWarning
            )

            estimator.fit(X_train, y_train)

        y_pred = estimator.predict(X_val)

        mae = mean_absolute_error(y_val, y_pred)
        rmse = np.sqrt(
            mean_squared_error(y_val, y_pred)
        )
        r2 = r2_score(y_val, y_pred)

        elapsed = time.time() - start

        results90.append({
            "Experiment": experiment_name,
            "Model": model_name,
            "N": len(y_val),
            "MAE": mae,
            "RMSE": rmse,
            "R2": r2,
            "Fit_Seconds": elapsed
        })

        for i, (_, row) in enumerate(val90.iterrows()):

            predictions90.append({
                "Experiment": experiment_name,
                "Model": model_name,
                "Country": row["Country"],
                "Target_Year": row["Target_Year"],
                "Observed": y_val[i],
                "Predicted": y_pred[i]
            })

        fitted90[(experiment_name, model_name)] = estimator

        print(
            f"  {model_name:22s} | "
            f"MAE={mae:8.3f} | "
            f"RMSE={rmse:8.3f} | "
            f"R2={r2:8.4f} | "
            f"{elapsed:6.2f}s"
        )

results90 = pd.DataFrame(results90)
predictions90 = pd.DataFrame(predictions90)

print("\n" + "=" * 78)
print("VALIDATION RESULTS — SORTED BY MAE")
print("=" * 78)

print(
    results90.sort_values("MAE")
    .round(4)
    .to_string(index=False)
)

print("\nTOP MODEL PER EXPERIMENT")

best90 = (
    results90.sort_values(["Experiment", "MAE"])
    .groupby("Experiment", as_index=False)
    .first()
)

print(
    best90[
        ["Experiment", "Model", "MAE", "RMSE", "R2"]
    ]
    .round(4)
    .to_string(index=False)
)

print("\nPERSISTENCE VALIDATION BENCHMARK")

persistence_pred = val90["Incidence_t"].to_numpy(dtype=float)

persistence_mae = mean_absolute_error(
    y_val, persistence_pred
)

persistence_rmse = np.sqrt(
    mean_squared_error(y_val, persistence_pred)
)

persistence_r2 = r2_score(
    y_val, persistence_pred
)

print(
    f"MAE={persistence_mae:.4f} | "
    f"RMSE={persistence_rmse:.4f} | "
    f"R2={persistence_r2:.4f}"
)

print("\nVALIDATION MODELS BEATING PERSISTENCE")

better = results90.loc[
    results90["MAE"] < persistence_mae
].sort_values("MAE")

print(
    better[
        ["Experiment", "Model", "MAE", "RMSE", "R2"]
    ]
    .round(4)
    .to_string(index=False)
    if len(better)
    else "None of the models improved on persistence MAE."
)

print("\nINTEGRITY CHECKS")

assert len(results90) == 60
assert len(predictions90) == 60 * 58
assert predictions90["Target_Year"].max() == 2022
assert predictions90["Target_Year"].min() == 2021
assert np.isfinite(results90[["MAE", "RMSE", "R2"]]).all().all()

print("Model fits:", len(results90))
print("Validation predictions:", len(predictions90))
print("Elapsed seconds:", round(time.time() - start_all, 2))

print("\nCELL 90 COMPLETE — VALIDATION ONLY")
print("HELD-OUT TEST NOT EVALUATED")
print("NO FILES MODIFIED")
```

```python

# ============================================================
# CELL 91 — PAIRED BIOLOGICAL IMPROVEMENT ANALYSIS
# BIML 2.0 | VALIDATION ONLY | NO MODEL FITTING
# ============================================================

import numpy as np
import pandas as pd

assert "predictions90" in globals()
assert "val90" in globals()

comparisons = [
    ("B1_environment", "P2_E", "P3_E_B1"),
    ("B2_environment", "P4_E_SUPPORT", "P4_E_SUPPORT_B2"),
    ("B1_history_environment", "P5_H_E", "P6_H_E_B1"),
    ("B2_history_environment",
     "P7_H_E_SUPPORT", "P7_H_E_SUPPORT_B2"),
    ("Combined_vs_B2",
     "P7_H_E_SUPPORT_B2", "P8_H_E_SUPPORT_B1_B2")
]

models_to_check = [
    "Ridge",
    "RandomForest",
    "HistGradientBoosting",
    "XGBoost",
    "LightGBM",
    "RBF_SVR"
]

rows = []
paired_details91 = []

for label, baseline, augmented in comparisons:

    for model in models_to_check:

        base = predictions90.loc[
            (predictions90["Experiment"] == baseline)
            & (predictions90["Model"] == model)
        ][
            ["Country", "Target_Year", "Observed", "Predicted"]
        ].rename(columns={"Predicted": "Base_Pred"})

        aug = predictions90.loc[
            (predictions90["Experiment"] == augmented)
            & (predictions90["Model"] == model)
        ][
            ["Country", "Target_Year", "Observed", "Predicted"]
        ].rename(columns={
            "Observed": "Observed_Aug",
            "Predicted": "Aug_Pred"
        })

        paired = base.merge(
            aug,
            on=["Country", "Target_Year"],
            how="inner",
            validate="one_to_one"
        )

        assert len(paired) == 58
        assert np.allclose(
            paired["Observed"], paired["Observed_Aug"]
        )

        paired["Base_AE"] = (
            paired["Observed"] - paired["Base_Pred"]
        ).abs()

        paired["Aug_AE"] = (
            paired["Observed"] - paired["Aug_Pred"]
        ).abs()

        paired["AE_Improvement"] = (
            paired["Base_AE"] - paired["Aug_AE"]
        )

        improvement = paired["AE_Improvement"].mean()

        rows.append({
            "Comparison": label,
            "Model": model,
            "Base_MAE": paired["Base_AE"].mean(),
            "Aug_MAE": paired["Aug_AE"].mean(),
            "Delta_MAE": improvement,
            "Improvement_pct": (
                100 * improvement / paired["Base_AE"].mean()
            ),
            "Improved_Observations": int(
                (paired["AE_Improvement"] > 0).sum()
            ),
            "Worsened_Observations": int(
                (paired["AE_Improvement"] < 0).sum()
            ),
            "Improved_Countries": int(
                (
                    paired.groupby("Country")["AE_Improvement"]
                    .mean() > 0
                ).sum()
            )
        })

        paired["Comparison"] = label
        paired["Model"] = model

        paired_details91.append(paired)

results91 = pd.DataFrame(rows)
paired_details91 = pd.concat(
    paired_details91, ignore_index=True
)

print("=" * 78)
print("CELL 91 — PAIRED BIOLOGICAL IMPROVEMENTS")
print("=" * 78)

print("\nALL MATCHED COMPARISONS")
print(
    results91.sort_values(
        ["Comparison", "Model"]
    )
    .round(3)
    .to_string(index=False)
)

print("\nRANDOM FOREST COMPARISONS")
print(
    results91.loc[
        results91["Model"] == "RandomForest"
    ]
    .round(3)
    .to_string(index=False)
)

print("\nXGBOOST COMPARISONS")
print(
    results91.loc[
        results91["Model"] == "XGBoost"
    ]
    .round(3)
    .to_string(index=False)
)

print("\nVALIDATION INTEGRITY")
print("Comparisons:", len(comparisons))
print("Model families:", len(models_to_check))
print("Paired comparisons:", len(results91))
print("Paired observation rows:", len(paired_details91))
print(
    "Unique target years:",
    sorted(paired_details91["Target_Year"].unique())
)

assert len(results91) == 30
assert len(paired_details91) == 30 * 58
assert set(
    paired_details91["Target_Year"]
) == {2021, 2022}

print("\nCELL 91 COMPLETE — VALIDATION ONLY")
print("NO MODELS FITTED — NO TEST DATA ACCESSED")
```

```python

# ============================================================
# CELL 92 — COUNTRY-CLUSTER BOOTSTRAP
# BIML 2.0 | VALIDATION ONLY | 2,000 REPLICATES
# NO MODEL TRAINING | NO TEST DATA | NO FILES MODIFIED
# ============================================================

import numpy as np
import pandas as pd

assert "paired_details91" in globals()
assert "results91" in globals()

N_BOOT = 2000
BOOT_SEED = 20261016

rng = np.random.default_rng(BOOT_SEED)

countries92 = sorted(
    paired_details91["Country"].unique()
)

assert len(countries92) == 29

# Shared country resamples across all comparisons
# preserve comparability between bootstrap estimates.
bootstrap_indices92 = rng.integers(
    low=0,
    high=len(countries92),
    size=(N_BOOT, len(countries92))
)

bootstrap_countries92 = np.asarray(countries92)[
    bootstrap_indices92
]

rows92 = []
bootstrap_distributions92 = {}

for (comparison, model), group in paired_details91.groupby(
    ["Comparison", "Model"],
    sort=True
):

    # Each country has exactly two validation target years.
    country_improvement = (
        group.groupby("Country")["AE_Improvement"]
        .mean()
        .reindex(countries92)
    )

    assert country_improvement.notna().all()

    values = country_improvement.to_numpy(dtype=float)

    # Each bootstrap replicate samples 29 countries
    # with replacement, retaining their paired-year structure.
    bootstrap_delta = values[bootstrap_indices92].mean(axis=1)

    point_delta = float(values.mean())

    ci_low, ci_high = np.percentile(
        bootstrap_delta, [2.5, 97.5]
    )

    rows92.append({
        "Comparison": comparison,
        "Model": model,
        "Delta_MAE": point_delta,
        "CI_2.5": ci_low,
        "CI_97.5": ci_high,
        "CI_Excludes_Zero": bool(
            ci_low > 0 or ci_high < 0
        ),
        "Bootstrap_P_Positive": float(
            np.mean(bootstrap_delta > 0)
        )
    })

    bootstrap_distributions92[(comparison, model)] = (
        bootstrap_delta
    )

results92 = pd.DataFrame(rows92)

print("=" * 78)
print("CELL 92 — COUNTRY-CLUSTER BOOTSTRAP RESULTS")
print("=" * 78)

print("\nALL PAIRED COMPARISONS")
print(
    results92.round(4).to_string(index=False)
)

print("\nRANDOM FOREST — BIOLOGICAL COMPARISONS")
print(
    results92.loc[
        results92["Model"] == "RandomForest"
    ].round(4).to_string(index=False)
)

print("\nXGBOOST — BIOLOGICAL COMPARISONS")
print(
    results92.loc[
        results92["Model"] == "XGBoost"
    ].round(4).to_string(index=False)
)

print("\nSUMMARY")
print("Countries:", len(countries92))
print("Bootstrap replicates:", N_BOOT)
print("Paired comparisons:", len(results92))
print(
    "Intervals excluding zero:",
    int(results92["CI_Excludes_Zero"].sum())
)

assert len(results92) == 30
assert all(
    len(values) == N_BOOT
    for values in bootstrap_distributions92.values()
)

print("\nCELL 92 COMPLETE — VALIDATION ONLY")
print("NO MODELS FITTED — TEST PERIOD UNTOUCHED")
```

```python

# ============================================================
# CELL 93 — HISTORY-FREE ML VS BIML DIAGNOSTIC AUDIT
# BIML 2.0 | VALIDATION ONLY | NO MODEL FITTING
# ============================================================

import numpy as np
import pandas as pd

assert "predictions90" in globals()
assert "results90" in globals()

history_free = [
    "P2_E",
    "P3_E_B1",
    "P4_E_SUPPORT",
    "P4_E_SUPPORT_B2"
]

print("=" * 76)
print("CELL 93 — HISTORY-FREE ML VS BIML")
print("=" * 76)

# 1. Full results for all six algorithms
history_free_results93 = (
    results90.loc[
        results90["Experiment"].isin(history_free)
    ]
    .sort_values(["Experiment", "MAE"])
    .copy()
)

print("\nALL HISTORY-FREE MODEL RESULTS")
print(
    history_free_results93[
        ["Experiment", "Model", "MAE", "RMSE", "R2"]
    ]
    .round(4)
    .to_string(index=False)
)

# 2. Country-level performance for matched XGBoost comparison
def extract_predictions93(experiment, model):

    subset = predictions90.loc[
        (predictions90["Experiment"] == experiment)
        & (predictions90["Model"] == model)
    ].copy()

    assert len(subset) == 58

    return subset

base93 = extract_predictions93(
    "P4_E_SUPPORT", "XGBoost"
)

biml93 = extract_predictions93(
    "P4_E_SUPPORT_B2", "XGBoost"
)

paired93 = base93.merge(
    biml93,
    on=["Country", "Target_Year"],
    suffixes=("_ML", "_BIML"),
    validate="one_to_one"
)

assert len(paired93) == 58

assert np.allclose(
    paired93["Observed_ML"],
    paired93["Observed_BIML"]
)

paired93["ML_AE"] = (
    paired93["Observed_ML"]
    - paired93["Predicted_ML"]
).abs()

paired93["BIML_AE"] = (
    paired93["Observed_BIML"]
    - paired93["Predicted_BIML"]
).abs()

paired93["Improvement"] = (
    paired93["ML_AE"] - paired93["BIML_AE"]
)

country93 = (
    paired93.groupby("Country")
    .agg(
        ML_MAE=("ML_AE", "mean"),
        BIML_MAE=("BIML_AE", "mean"),
        Improvement=("Improvement", "mean"),
        Mean_Incidence=("Observed_ML", "mean")
    )
    .reset_index()
    .sort_values("Improvement", ascending=False)
)

print("\nCOUNTRY-LEVEL XGBOOST COMPARISON")
print(
    country93.round(3).to_string(index=False)
)

print("\nCOUNTRY IMPROVEMENT SUMMARY")
print("Countries improved:",
      int((country93["Improvement"] > 0).sum()))
print("Countries worsened:",
      int((country93["Improvement"] < 0).sum()))

print("\nVALIDATION YEAR COMPARISON")
print(
    paired93.groupby("Target_Year")
    .agg(
        ML_MAE=("ML_AE", "mean"),
        BIML_MAE=("BIML_AE", "mean"),
        Improvement=("Improvement", "mean")
    )
    .round(3)
    .to_string()
)

print("\nINCIDENCE-LEVEL COMPARISON")

# Fixed bins for descriptive diagnostics only.
# These are NOT early-warning thresholds.
bins93 = [-np.inf, 25, 100, np.inf]
labels93 = ["Below_25", "25_to_100", "Above_100"]

paired93["Incidence_Group"] = pd.cut(
    paired93["Observed_ML"],
    bins=bins93,
    labels=labels93
)

print(
    paired93.groupby(
        "Incidence_Group",
        observed=True
    )
    .agg(
        N=("ML_AE", "size"),
        ML_MAE=("ML_AE", "mean"),
        BIML_MAE=("BIML_AE", "mean"),
        Improvement=("Improvement", "mean")
    )
    .round(3)
    .to_string()
)

print("\nCELL 93 COMPLETE")
print("NO MODELS TRAINED — NO TEST DATA ACCESSED")
```

```python

# ============================================================
# CELL 94 — COUNTRY INFLUENCE ANALYSIS
# BIML 2.0 | HISTORY-FREE XGBOOST
# VALIDATION ONLY | NO MODEL FITTING
# ============================================================

import numpy as np
import pandas as pd

assert "paired93" in globals()
assert "country93" in globals()

print("=" * 76)
print("CELL 94 — COUNTRY INFLUENCE AND ROBUSTNESS")
print("=" * 76)

overall_improvement94 = paired93["Improvement"].mean()

countries94 = sorted(paired93["Country"].unique())

rows94 = []

for country in countries94:

    subset = paired93.loc[
        paired93["Country"] != country
    ]

    remaining_improvement = subset["Improvement"].mean()

    rows94.append({
        "Excluded_Country": country,
        "Remaining_N": len(subset),
        "MAE_Improvement": remaining_improvement,
        "Influence": (
            overall_improvement94 - remaining_improvement
        )
    })

influence94 = (
    pd.DataFrame(rows94)
    .sort_values("Influence", ascending=False)
)

print("\nORIGINAL XGBOOST B2 IMPROVEMENT")
print(f"MAE reduction: {overall_improvement94:.4f}")

print("\nLEAVE-ONE-COUNTRY-OUT RESULTS")
print(
    influence94.round(4).to_string(index=False)
)

print("\nROBUSTNESS SUMMARY")
print(
    "Minimum remaining improvement:",
    round(influence94["MAE_Improvement"].min(), 4)
)
print(
    "Maximum remaining improvement:",
    round(influence94["MAE_Improvement"].max(), 4)
)
print(
    "Countries whose exclusion reverses improvement:",
    int((influence94["MAE_Improvement"] <= 0).sum())
)

# Exclude the three countries with the largest
# positive contributions to the original improvement.
top3 = (
    country93.sort_values(
        "Improvement", ascending=False
    )["Country"]
    .head(3)
    .tolist()
)

without_top3 = paired93.loc[
    ~paired93["Country"].isin(top3)
]

print("\nEXCLUDING THREE LARGEST BENEFICIARIES")
print("Excluded:", top3)
print("Remaining observations:", len(without_top3))
print(
    "Remaining MAE improvement:",
    round(without_top3["Improvement"].mean(), 4)
)

# Robust descriptive summaries across countries.
country_effects94 = (
    country93["Improvement"].to_numpy(dtype=float)
)

print("\nCOUNTRY-LEVEL EFFECT DISTRIBUTION")
print(
    pd.Series(country_effects94)
    .describe(percentiles=[.10, .25, .50, .75, .90])
    .round(4)
)

print("\nINTEGRITY CHECKS")
assert len(countries94) == 29
assert len(influence94) == 29
assert (influence94["Remaining_N"] == 56).all()
assert np.isclose(
    overall_improvement94,
    7.1399,
    atol=0.001
)

print("\nCELL 94 COMPLETE — VALIDATION ONLY")
print("NO MODELS TRAINED — TEST PERIOD UNTOUCHED")
```

```python

# ============================================================
# CELL 95 — RESIDUAL-LEARNING ML VS BIML
# BIML 2.0 | TRAINING + VALIDATION ONLY
# NO TEST DATA | NO FILES MODIFIED
# ============================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "prospective" in globals()
assert "feature_groups" in globals()

train95 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val95 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

assert len(train95) == 145
assert len(val95) == 58

target = "Target_Incidence_t_plus_1"

# Target: next-year change in incidence
delta_train95 = (
    train95[target] - train95["Incidence_t"]
).to_numpy(dtype=float)

y_val95 = val95[target].to_numpy(dtype=float)

# Matched feature configurations
H = feature_groups["H"]
E = feature_groups["E"]
B1 = feature_groups["B1"]
B2 = feature_groups["B2"]

residual_experiments95 = {
    "Residual_ML":
        H + E + ["B2_supported_months"],

    "Residual_BIML_B2":
        H + E + ["B2_supported_months"] + B2,

    "Residual_BIML_B1_B2":
        H + E + ["B2_supported_months"] + B1 + B2
}

models95 = {
    "RandomForest": RandomForestRegressor(
        n_estimators=400,
        min_samples_leaf=3,
        random_state=20261016,
        n_jobs=-1
    ),
    "XGBoost": XGBRegressor(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.05,
        reg_lambda=5,
        objective="reg:squarederror",
        random_state=20261016,
        n_jobs=-1
    )
}

results95 = []
predictions95 = []
fitted95 = {}

# Persistence baseline
persistence95 = val95["Incidence_t"].to_numpy(dtype=float)

def metrics95(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2": r2_score(y_true, y_pred)
    }

results95.append({
    "Experiment": "Persistence",
    "Model": "No_Fit",
    **metrics95(y_val95, persistence95)
})

print("=" * 75)
print("CELL 95 — RESIDUAL-LEARNING EXPERIMENT")
print("=" * 75)

for experiment, features in residual_experiments95.items():

    for model_name, estimator in models95.items():

        pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", clone(estimator))
        ])

        # Learn the incidence change using TRAIN only
        pipeline.fit(
            train95[features],
            delta_train95
        )

        predicted_change = pipeline.predict(
            val95[features]
        )

        # Restore incidence level
        predicted_incidence = (
            val95["Incidence_t"].to_numpy(dtype=float)
            + predicted_change
        )

        # Incidence cannot be negative
        predicted_incidence = np.maximum(
            predicted_incidence, 0
        )

        fitted95[(experiment, model_name)] = pipeline

        results95.append({
            "Experiment": experiment,
            "Model": model_name,
            **metrics95(y_val95, predicted_incidence)
        })

        for country, year, observed, predicted in zip(
            val95["Country"],
            val95["Target_Year"],
            y_val95,
            predicted_incidence
        ):
            predictions95.append({
                "Experiment": experiment,
                "Model": model_name,
                "Country": country,
                "Target_Year": year,
                "Observed": observed,
                "Predicted": predicted
            })

results95 = pd.DataFrame(results95)
predictions95 = pd.DataFrame(predictions95)

print("\nVALIDATION RESULTS")
print(
    results95.sort_values("MAE")
    .round(4)
    .to_string(index=False)
)

print("\nIMPROVEMENT RELATIVE TO PERSISTENCE")

persistence_mae95 = results95.loc[
    results95["Experiment"] == "Persistence",
    "MAE"
].iloc[0]

comparison95 = results95.loc[
    results95["Experiment"] != "Persistence"
].copy()

comparison95["Improvement_vs_Persistence_pct"] = (
    100 * (
        persistence_mae95 - comparison95["MAE"]
    ) / persistence_mae95
)

print(
    comparison95[
        [
            "Experiment",
            "Model",
            "MAE",
            "Improvement_vs_Persistence_pct"
        ]
    ]
    .sort_values("MAE")
    .round(4)
    .to_string(index=False)
)

assert len(results95) == 7
assert len(predictions95) == 348
assert set(predictions95["Target_Year"]) == {
    2021, 2022
}

print("\nCELL 95 COMPLETE")
print("TEST DATA NOT ACCESSED")
```

```python

# ============================================================
# CELL 96 — MATCHED RESIDUAL ML VS BIML COMPARISON
# BIML 2.0 | VALIDATION ONLY
# ============================================================

import numpy as np
import pandas as pd

assert "results95" in globals()
assert "predictions95" in globals()

comparisons96 = [
    ("Residual_ML", "Residual_BIML_B2"),
    ("Residual_ML", "Residual_BIML_B1_B2"),
    ("Residual_BIML_B2", "Residual_BIML_B1_B2")
]

rows96 = []

for base, augmented in comparisons96:

    for model in ["RandomForest", "XGBoost"]:

        base_row = results95.loc[
            (results95["Experiment"] == base)
            & (results95["Model"] == model)
        ].iloc[0]

        augmented_row = results95.loc[
            (results95["Experiment"] == augmented)
            & (results95["Model"] == model)
        ].iloc[0]

        rows96.append({
            "Comparison": f"{base} -> {augmented}",
            "Model": model,
            "Base_MAE": base_row["MAE"],
            "BIML_MAE": augmented_row["MAE"],
            "MAE_Reduction": (
                base_row["MAE"] - augmented_row["MAE"]
            ),
            "Base_RMSE": base_row["RMSE"],
            "BIML_RMSE": augmented_row["RMSE"],
            "RMSE_Reduction": (
                base_row["RMSE"] - augmented_row["RMSE"]
            ),
            "Base_R2": base_row["R2"],
            "BIML_R2": augmented_row["R2"],
            "R2_Gain": (
                augmented_row["R2"] - base_row["R2"]
            )
        })

results96 = pd.DataFrame(rows96)

print("=" * 78)
print("CELL 96 — MATCHED RESIDUAL ML VS BIML")
print("=" * 78)

print(
    results96.round(5).to_string(index=False)
)

print("\nBIML IMPROVEMENT COUNTS")

for metric in [
    "MAE_Reduction",
    "RMSE_Reduction",
    "R2_Gain"
]:
    print(
        metric,
        int((results96[metric] > 0).sum()),
        "of",
        len(results96)
    )

print("\nBEST VALIDATION MODELS BY METRIC")

for metric, ascending in [
    ("MAE", True),
    ("RMSE", True),
    ("R2", False)
]:
    best = results95.sort_values(
        metric,
        ascending=ascending
    ).iloc[0]

    print(
        f"{metric}: {best['Experiment']} / "
        f"{best['Model']} = {best[metric]:.5f}"
    )

assert len(results96) == 6

print("\nCELL 96 COMPLETE — VALIDATION ONLY")
print("NO MODEL TRAINING — NO TEST DATA ACCESSED")
```

```python

# ============================================================
# CELL 97 — FOCUSED RESIDUAL RANDOM FOREST DEVELOPMENT
# BIML 2.0 | TRAINING + VALIDATION ONLY
# NO TEST DATA | NO FILES MODIFIED
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "prospective" in globals()
assert "feature_groups" in globals()

train97 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val97 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

assert len(train97) == 145
assert len(val97) == 58

target97 = "Target_Incidence_t_plus_1"

delta_train97 = (
    train97[target97] - train97["Incidence_t"]
).to_numpy(dtype=float)

y_val97 = val97[target97].to_numpy(dtype=float)

base_features97 = (
    feature_groups["H"]
    + feature_groups["E"]
    + ["B2_supported_months"]
)

biml_features97 = (
    base_features97 + feature_groups["B2"]
)

# Small, predefined development grid.
# These are not exhaustive hyperparameter searches.
configurations97 = {
    "RF_original": {
        "n_estimators": 400,
        "min_samples_leaf": 3,
        "max_features": 1.0
    },
    "RF_regularized": {
        "n_estimators": 500,
        "min_samples_leaf": 5,
        "max_features": 0.8
    },
    "RF_flexible": {
        "n_estimators": 500,
        "min_samples_leaf": 2,
        "max_features": 0.8
    }
}

results97 = []
predictions97 = []
fitted97 = {}

for config_name, config in configurations97.items():

    for experiment, features in {
        "Residual_ML": base_features97,
        "Residual_BIML_B2": biml_features97
    }.items():

        model = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "regressor",
                RandomForestRegressor(
                    **config,
                    random_state=20261016,
                    n_jobs=-1
                )
            )
        ])

        model.fit(
            train97[features],
            delta_train97
        )

        predicted_delta = model.predict(
            val97[features]
        )

        predicted_incidence = np.maximum(
            val97["Incidence_t"].to_numpy(dtype=float)
            + predicted_delta,
            0
        )

        fitted97[(config_name, experiment)] = model

        results97.append({
            "Configuration": config_name,
            "Experiment": experiment,
            "MAE": mean_absolute_error(
                y_val97, predicted_incidence
            ),
            "RMSE": np.sqrt(
                mean_squared_error(
                    y_val97, predicted_incidence
                )
            ),
            "R2": r2_score(
                y_val97, predicted_incidence
            )
        })

        for country, year, observed, predicted in zip(
            val97["Country"],
            val97["Target_Year"],
            y_val97,
            predicted_incidence
        ):
            predictions97.append({
                "Configuration": config_name,
                "Experiment": experiment,
                "Country": country,
                "Target_Year": year,
                "Observed": observed,
                "Predicted": predicted
            })

results97 = pd.DataFrame(results97)
predictions97 = pd.DataFrame(predictions97)

print("=" * 76)
print("CELL 97 — RESIDUAL RANDOM FOREST DEVELOPMENT")
print("=" * 76)

print("\nALL VALIDATION RESULTS")
print(
    results97.sort_values("RMSE")
    .round(5)
    .to_string(index=False)
)

print("\nMATCHED ML VS BIML COMPARISONS")

paired_results97 = []

for config_name in configurations97:

    ml = results97.loc[
        (results97["Configuration"] == config_name)
        & (results97["Experiment"] == "Residual_ML")
    ].iloc[0]

    biml = results97.loc[
        (results97["Configuration"] == config_name)
        & (results97["Experiment"] == "Residual_BIML_B2")
    ].iloc[0]

    paired_results97.append({
        "Configuration": config_name,
        "MAE_Reduction": ml["MAE"] - biml["MAE"],
        "RMSE_Reduction": ml["RMSE"] - biml["RMSE"],
        "R2_Gain": biml["R2"] - ml["R2"]
    })

paired_results97 = pd.DataFrame(paired_results97)

print(
    paired_results97.round(5).to_string(index=False)
)

assert len(results97) == 6
assert len(predictions97) == 348
assert set(predictions97["Target_Year"]) == {2021, 2022}

print("\nCELL 97 COMPLETE")
print("NO TEST DATA ACCESSED")
```

```python

# ============================================================
# CELL 93 — FITTED AUTOREGRESSIVE BENCHMARK
# BIML 2.0 | TRAINING + VALIDATION ONLY
# NO TEST DATA | NO FILES MODIFIED
# ============================================================

import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "train90" in globals()
assert "val90" in globals()
assert "results90" in globals()

target = "Target_Incidence_t_plus_1"

X_train93 = train90[["Incidence_t"]]
X_val93 = val90[["Incidence_t"]]

y_train93 = train90[target].to_numpy(dtype=float)
y_val93 = val90[target].to_numpy(dtype=float)

assert len(X_train93) == 145
assert len(X_val93) == 58

# Fitted autoregressive model:
# I(t+1) = intercept + coefficient * I(t)
ar93 = LinearRegression()
ar93.fit(X_train93, y_train93)

pred_ar93 = ar93.predict(X_val93)

# Parameter-free persistence:
# I(t+1) = I(t)
pred_persistence93 = X_val93["Incidence_t"].to_numpy()

def evaluate93(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2": r2_score(y_true, y_pred),
        "Bias": float(np.mean(y_pred - y_true))
    }

rows93 = [
    {
        "Model": "Persistence",
        **evaluate93(y_val93, pred_persistence93)
    },
    {
        "Model": "Fitted_AR1",
        **evaluate93(y_val93, pred_ar93)
    }
]

# Include best previously fitted history-only ML.
history_results93 = results90.loc[
    results90["Experiment"] == "P1_H"
].sort_values("MAE")

print("=" * 75)
print("CELL 93 — FITTED AUTOREGRESSIVE BENCHMARK")
print("=" * 75)

print("\nFITTED AR(1)-STYLE RELATIONSHIP")
print(f"Intercept: {ar93.intercept_:.6f}")
print(f"Incidence coefficient: {ar93.coef_[0]:.6f}")

print("\nBASELINE COMPARISON")
print(
    pd.DataFrame(rows93)
    .round(4)
    .to_string(index=False)
)

print("\nPREVIOUS HISTORY-ONLY ML RESULTS")
print(
    history_results93[
        ["Model", "MAE", "RMSE", "R2"]
    ]
    .round(4)
    .to_string(index=False)
)

# Save validation predictions in memory for future
# paired comparisons and early-warning evaluation.
baseline_predictions93 = val90[
    ["Country", "Target_Year"]
].copy()

baseline_predictions93["Observed"] = y_val93
baseline_predictions93["Persistence"] = pred_persistence93
baseline_predictions93["Fitted_AR1"] = pred_ar93

assert set(baseline_predictions93["Target_Year"]) == {
    2021, 2022
}
assert np.isfinite(
    baseline_predictions93[
        ["Observed", "Persistence", "Fitted_AR1"]
    ].to_numpy()
).all()

print("\nCELL 93 COMPLETE — VALIDATION ONLY")
print("NO TEST DATA ACCESSED")
```

```python

# ============================================================
# CELL 98 — FREEZE FINAL BIML EXPERIMENT SPECIFICATIONS
# BIML 2.0 | NO MODEL FITTING | NO TEST ACCESS
# ============================================================

import copy
import json
import pandas as pd

assert "feature_groups" in globals()
assert "prospective" in globals()

H = list(feature_groups["H"])
E = list(feature_groups["E"])
B2 = list(feature_groups["B2"])

frozen_protocol98 = {
    "protocol_name": "BIML_2_0_Final_Validation_Selected",
    "seed": 20261016,
    "target": "Target_Incidence_t_plus_1",
    "forecast_horizon": "One year ahead",
    "training_target_years": [2016, 2017, 2018, 2019, 2020],
    "validation_target_years": [2021, 2022],
    "test_target_years": [2023, 2024],
    "selection_basis": "2021-2022 validation performance",
    "experiments": {
        "History_Free": {
            "model": "XGBoost",
            "learning_target": "Incidence",
            "primary_metric": "MAE",
            "ml_features": E + ["B2_supported_months"],
            "biml_features": E + ["B2_supported_months"] + B2,
            "parameters": {
                "n_estimators": 300,
                "max_depth": 3,
                "learning_rate": 0.05,
                "reg_lambda": 5,
                "objective": "reg:squarederror",
                "random_state": 20261016
            }
        },
        "History_Enabled_Residual": {
            "model": "RandomForest",
            "learning_target": "Incidence_change",
            "primary_metric": "RMSE",
            "ml_features": H + E + ["B2_supported_months"],
            "biml_features": H + E + ["B2_supported_months"] + B2,
            "parameters": {
                "n_estimators": 400,
                "min_samples_leaf": 3,
                "max_features": 1.0,
                "random_state": 20261016,
                "n_jobs": -1
            }
        }
    },
    "report_metrics": ["MAE", "RMSE", "R2"],
    "additional_baseline": "Persistence",
    "test_evaluation_policy": (
        "Single held-out evaluation after validation-based "
        "model selection. No test-driven retuning."
    )
}

# Defensive copy so subsequent changes to feature_groups
# do not modify the frozen protocol.
frozen_protocol98 = copy.deepcopy(frozen_protocol98)

print("=" * 76)
print("CELL 98 — FROZEN EXPERIMENT SPECIFICATIONS")
print("=" * 76)

print(json.dumps(frozen_protocol98, indent=2))

print("\nPARTITION COUNTS")
print(
    prospective.groupby("Partition")
    .agg(
        N=("Country", "size"),
        Countries=("Country", "nunique"),
        Min_Target_Year=("Target_Year", "min"),
        Max_Target_Year=("Target_Year", "max")
    )
    .to_string()
)

assert len(
    frozen_protocol98["experiments"]
) == 2

assert set(
    prospective.loc[
        prospective["Partition"] == "Test",
        "Target_Year"
    ]
) == {2023, 2024}

print("\nCELL 98 COMPLETE")
print("SPECIFICATIONS FROZEN IN MEMORY")
print("NO MODELS TRAINED — NO TEST OUTCOMES ACCESSED")
```

```python

# ============================================================
# CELL 99 — FINAL HELD-OUT TEST EVALUATION
# BIML 2.0 | FROZEN MODELS | TEST YEARS 2023–2024
# NO RETRAINING | NO HYPERPARAMETER TUNING
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "frozen_protocol98" in globals()
assert "fitted90" in globals()
assert "fitted95" in globals()
assert "prospective" in globals()

test99 = prospective.loc[
    prospective["Partition"] == "Test"
].copy()

assert len(test99) == 58
assert test99["Country"].nunique() == 29
assert set(test99["Target_Year"]) == {2023, 2024}
assert not test99.duplicated(
    ["Country", "Target_Year"]
).any()

target99 = frozen_protocol98["target"]
y_test99 = test99[target99].to_numpy(dtype=float)

def metrics99(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, y_pred)
        ),
        "R2": r2_score(y_true, y_pred)
    }

results99 = []
prediction_rows99 = []

def record99(family, experiment, predictions):
    predictions = np.asarray(
        predictions, dtype=float
    )

    assert len(predictions) == len(test99)
    assert np.isfinite(predictions).all()

    results99.append({
        "Family": family,
        "Experiment": experiment,
        "N": len(predictions),
        **metrics99(y_test99, predictions)
    })

    for country, year, observed, predicted in zip(
        test99["Country"],
        test99["Target_Year"],
        y_test99,
        predictions
    ):
        prediction_rows99.append({
            "Family": family,
            "Experiment": experiment,
            "Country": country,
            "Target_Year": int(year),
            "Observed": float(observed),
            "Predicted": float(predicted)
        })

# ------------------------------------------------------------
# 1. Persistence baseline
# ------------------------------------------------------------
record99(
    "Baseline",
    "Persistence",
    test99["Incidence_t"].to_numpy(dtype=float)
)

# ------------------------------------------------------------
# 2. History-free ML and BIML
# Reuse fitted XGBoost pipelines from Cell 90
# ------------------------------------------------------------
history_free99 = frozen_protocol98[
    "experiments"
]["History_Free"]

history_free_pairs99 = {
    "HistoryFree_ML": (
        "P4_E_SUPPORT",
        history_free99["ml_features"]
    ),
    "HistoryFree_BIML_B2": (
        "P4_E_SUPPORT_B2",
        history_free99["biml_features"]
    )
}

for name, (experiment, features) in (
    history_free_pairs99.items()
):
    model = fitted90[(experiment, "XGBoost")]

    prediction = model.predict(
        test99[features]
    )

    prediction = np.maximum(prediction, 0)

    record99(
        "History_Free",
        name,
        prediction
    )

# ------------------------------------------------------------
# 3. History-enabled residual ML and BIML
# Reuse fitted Random Forest pipelines from Cell 95
# ------------------------------------------------------------
residual99 = frozen_protocol98[
    "experiments"
]["History_Enabled_Residual"]

residual_pairs99 = {
    "Residual_ML": residual99["ml_features"],
    "Residual_BIML_B2": residual99["biml_features"]
}

for name, features in residual_pairs99.items():

    model = fitted95[(name, "RandomForest")]

    predicted_change = model.predict(
        test99[features]
    )

    predicted_incidence = np.maximum(
        test99["Incidence_t"].to_numpy(dtype=float)
        + predicted_change,
        0
    )

    record99(
        "History_Enabled_Residual",
        name,
        predicted_incidence
    )

results99 = pd.DataFrame(results99)
predictions99 = pd.DataFrame(prediction_rows99)

print("=" * 78)
print("CELL 99 — FINAL HELD-OUT TEST RESULTS")
print("=" * 78)

print("\nALL FROZEN MODEL RESULTS")
print(
    results99.sort_values("RMSE")
    .round(5)
    .to_string(index=False)
)

print("\nMATCHED ML VS BIML IMPROVEMENTS")

comparisons99 = [
    (
        "History_Free",
        "HistoryFree_ML",
        "HistoryFree_BIML_B2"
    ),
    (
        "History_Enabled_Residual",
        "Residual_ML",
        "Residual_BIML_B2"
    )
]

paired_results99 = []

for family, ml_name, biml_name in comparisons99:

    ml = results99.loc[
        (results99["Family"] == family)
        & (results99["Experiment"] == ml_name)
    ].iloc[0]

    biml = results99.loc[
        (results99["Family"] == family)
        & (results99["Experiment"] == biml_name)
    ].iloc[0]

    paired_results99.append({
        "Family": family,
        "ML_MAE": ml["MAE"],
        "BIML_MAE": biml["MAE"],
        "MAE_Reduction": ml["MAE"] - biml["MAE"],
        "ML_RMSE": ml["RMSE"],
        "BIML_RMSE": biml["RMSE"],
        "RMSE_Reduction": ml["RMSE"] - biml["RMSE"],
        "ML_R2": ml["R2"],
        "BIML_R2": biml["R2"],
        "R2_Gain": biml["R2"] - ml["R2"]
    })

paired_results99 = pd.DataFrame(paired_results99)

print(
    paired_results99.round(5).to_string(index=False)
)

print("\nINTEGRITY CHECKS")

assert len(results99) == 5
assert len(predictions99) == 290
assert predictions99.groupby(
    "Experiment"
).size().eq(58).all()

print("Frozen configurations evaluated: 2")
print("Models evaluated: 4 + persistence")
print("Countries:", test99["Country"].nunique())
print("Test years:", sorted(test99["Target_Year"].unique()))
print("No models retrained")
print("No hyperparameters changed")

print("\nCELL 99 COMPLETE — FINAL TEST EVALUATION")
```

```python

# ============================================================
# CELL 100 — EARLY-WARNING THRESHOLD AUDIT
# BIML 2.0 | TRAINING-DERIVED THRESHOLDS
# NO MODEL FITTING | NO TEST OUTCOME INSPECTION
# ============================================================

import numpy as np
import pandas as pd

assert "prospective" in globals()
assert "predictions99" in globals()

target100 = "Target_Incidence_t_plus_1"

train100 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

assert len(train100) == 145

y_train100 = train100[target100].to_numpy(dtype=float)

thresholds100 = {
    "Median": float(np.quantile(y_train100, 0.50)),
    "P75": float(np.quantile(y_train100, 0.75)),
    "P90": float(np.quantile(y_train100, 0.90))
}

rows100 = []

for name, threshold in thresholds100.items():

    high = y_train100 >= threshold

    rows100.append({
        "Threshold": name,
        "Incidence_Cutoff": threshold,
        "Training_High": int(high.sum()),
        "Training_Low": int((~high).sum()),
        "Training_High_pct": 100 * high.mean()
    })

threshold_audit100 = pd.DataFrame(rows100)

print("=" * 76)
print("CELL 100 — EARLY-WARNING THRESHOLD AUDIT")
print("=" * 76)

print("\nTRAINING INCIDENCE DISTRIBUTION")
print(
    pd.Series(y_train100)
    .describe(percentiles=[0.25, 0.50, 0.75, 0.90])
    .round(3)
)

print("\nTRAINING-DERIVED WARNING THRESHOLDS")
print(
    threshold_audit100.round(3).to_string(index=False)
)

print("\nFROZEN TEST PREDICTION INVENTORY")
print(
    predictions99.groupby(
        ["Family", "Experiment"]
    )
    .size()
    .to_string()
)

assert len(threshold_audit100) == 3
assert predictions99["Experiment"].nunique() == 5

print("\nCELL 100 COMPLETE")
print("NO CLASSIFICATION SCORES COMPUTED")
print("NO MODELS TRAINED")
```

```python

# ============================================================
# CELL 101 — EARLY-WARNING CLASSIFICATION EVALUATION
# BIML 2.0 | FROZEN P75 THRESHOLD
# FINAL TEST PREDICTIONS | NO RETRAINING
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score
)

assert "predictions99" in globals()
assert "thresholds100" in globals()

THRESHOLD101 = float(thresholds100["P75"])

assert np.isclose(THRESHOLD101, 345.1)

pred101 = predictions99.copy()

# Apply identical cutoff to observations and predictions.
pred101["Actual_High"] = (
    pred101["Observed"] >= THRESHOLD101
).astype(int)

pred101["Predicted_High"] = (
    pred101["Predicted"] >= THRESHOLD101
).astype(int)

results101 = []
confusion_matrices101 = {}

for (family, experiment), group in pred101.groupby(
    ["Family", "Experiment"],
    sort=True
):

    y_true = group["Actual_High"].to_numpy()
    y_pred = group["Predicted_High"].to_numpy()

    cm = confusion_matrix(
        y_true, y_pred, labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    confusion_matrices101[experiment] = cm

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )

    results101.append({
        "Family": family,
        "Experiment": experiment,
        "N": len(group),
        "Actual_High": int(y_true.sum()),
        "Predicted_High": int(y_pred.sum()),
        "TN": int(tn),
        "FP": int(fp),
        "FN": int(fn),
        "TP": int(tp),
        "Precision": precision_score(
            y_true, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_true, y_pred, zero_division=0
        ),
        "Specificity": specificity,
        "F1": f1_score(
            y_true, y_pred, zero_division=0
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_true, y_pred
        )
    })

results101 = pd.DataFrame(results101)

print("=" * 78)
print("CELL 101 — EARLY-WARNING CLASSIFICATION")
print("=" * 78)

print(f"\nFrozen incidence threshold: {THRESHOLD101:.3f}")
print("Positive class: annual incidence >= threshold")

print("\nCLASSIFICATION PERFORMANCE")
print(
    results101[
        [
            "Experiment",
            "N",
            "Actual_High",
            "Predicted_High",
            "Precision",
            "Recall",
            "Specificity",
            "F1",
            "Balanced_Accuracy",
            "FN",
            "FP"
        ]
    ]
    .round(4)
    .sort_values("F1", ascending=False)
    .to_string(index=False)
)

print("\nCONFUSION MATRICES")
print("Rows: actual [Low, High]")
print("Columns: predicted [Low, High]")

for experiment, cm in confusion_matrices101.items():

    print(f"\n{experiment}")
    print(
        pd.DataFrame(
            cm,
            index=["Actual_Low", "Actual_High"],
            columns=["Predicted_Low", "Predicted_High"]
        ).to_string()
    )

print("\nINTEGRITY CHECKS")

assert len(results101) == 5
assert results101["N"].eq(58).all()
assert results101["Actual_High"].nunique() == 1

assert pred101.groupby(
    ["Country", "Target_Year"]
)["Actual_High"].nunique().eq(1).all()

print("Models evaluated:", len(results101))
print("Frozen threshold:", THRESHOLD101)
print("No models retrained")
print("No threshold optimization")

print("\nCELL 101 COMPLETE — FINAL TEST CLASSIFICATION")
```

```python

# ============================================================
# CELL 102 — CORRECTED GAM BASELINE
# BIML 2.0 | TRAIN 2016–2020 | VALIDATION 2021–2022
# TRAINING-ONLY SPLINE BOUNDARIES | NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm

from statsmodels.gam.api import GLMGam, BSplines
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "prospective" in globals()
assert "feature_groups" in globals()

train102 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val102 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

target102 = "Target_Incidence_t_plus_1"

feature_sets102 = {
    "GAM_H": list(feature_groups["H"]),
    "GAM_H_E": (
        list(feature_groups["H"])
        + list(feature_groups["E"])
    )
}

results102 = []
predictions102 = {}
fitted102 = {}

for name, features in feature_sets102.items():

    # --------------------------------------------------------
    # 1. Training-only preprocessing
    # --------------------------------------------------------
    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(
        train102[features]
    )
    X_val_imp = imputer.transform(
        val102[features]
    )

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train_imp)
    X_val_scaled = scaler.transform(X_val_imp)

    # --------------------------------------------------------
    # 2. Define spline boundaries using TRAINING data only
    # --------------------------------------------------------
    lower = X_train_scaled.min(axis=0)
    upper = X_train_scaled.max(axis=0)

    span = upper - lower
    margin = np.maximum(span * 0.10, 0.10)

    lower_bound = lower - margin
    upper_bound = upper + margin

    knot_kwargs = [
        {
            "lower_bound": float(lower_bound[j]),
            "upper_bound": float(upper_bound[j])
        }
        for j in range(len(features))
    ]

    # Clip validation predictors only if they exceed
    # the predefined training-derived boundaries.
    X_val_safe = np.clip(
        X_val_scaled,
        lower_bound,
        upper_bound
    )

    clipped_values = int(
        np.sum(X_val_safe != X_val_scaled)
    )

    # --------------------------------------------------------
    # 3. Construct training spline basis
    # --------------------------------------------------------
    spline = BSplines(
        X_train_scaled,
        df=[4] * len(features),
        degree=[3] * len(features),
        variable_names=features,
        knot_kwds=knot_kwargs
    )

    # --------------------------------------------------------
    # 4. Fit regularized Gaussian GAM
    # --------------------------------------------------------
    model = GLMGam(
        endog=train102[target102].to_numpy(dtype=float),
        exog=np.ones((len(train102), 1)),
        smoother=spline,
        family=sm.families.Gaussian(),
        alpha=10.0
    )

    fitted = model.fit()

    # --------------------------------------------------------
    # 5. Predict validation using training spline knots
    # --------------------------------------------------------
    X_val_smooth = spline.transform(X_val_safe)

    prediction = fitted.predict(
        exog=np.ones((len(val102), 1)),
        exog_smooth=X_val_smooth
    )

    prediction = np.maximum(
        np.asarray(prediction, dtype=float),
        0
    )

    y_true = val102[target102].to_numpy(dtype=float)

    results102.append({
        "Model": name,
        "N": len(y_true),
        "MAE": mean_absolute_error(
            y_true, prediction
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, prediction)
        ),
        "R2": r2_score(
            y_true, prediction
        ),
        "Clipped_Values": clipped_values
    })

    predictions102[name] = prediction

    fitted102[name] = {
        "model": fitted,
        "spline": spline,
        "imputer": imputer,
        "scaler": scaler,
        "features": features,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound
    }

results102 = pd.DataFrame(results102)

print("=" * 76)
print("CELL 102 — CORRECTED GAM VALIDATION RESULTS")
print("=" * 76)

print(
    results102.sort_values("RMSE")
    .round(5)
    .to_string(index=False)
)

print("\nEXISTING VALIDATION BASELINES")
print("Persistence: MAE=12.03690, RMSE=21.42910, R2=0.95320")
print("Fitted AR1: MAE=12.47410, RMSE=22.09250, R2=0.95030")
print("Residual BIML B2 RF: MAE=13.19400, RMSE=20.58268, R2=0.95684")

print("\nCELL 102 COMPLETE")
print("Spline boundaries derived from training predictors")
print("No validation outcomes used for preprocessing")
print("No test data accessed")
```

```python

# ============================================================
# CELL 103 — RESIDUAL GAM BASELINE
# BIML 2.0 | AUTOREGRESSIVE ANCHOR + SMOOTH ENVIRONMENT
# TRAIN 2016–2020 | VALIDATION 2021–2022
# NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm

from statsmodels.gam.api import GLMGam, BSplines
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

train103 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val103 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

target103 = "Target_Incidence_t_plus_1"

environment103 = list(feature_groups["E"])

# Historical incidence is the autoregressive anchor.
history_train103 = train103["Incidence_t"].to_numpy(dtype=float)
history_val103 = val103["Incidence_t"].to_numpy(dtype=float)

# Predict annual incidence change.
delta_train103 = (
    train103[target103].to_numpy(dtype=float)
    - history_train103
)

y_val103 = val103[target103].to_numpy(dtype=float)

# ------------------------------------------------------------
# Training-only environmental preprocessing
# ------------------------------------------------------------
imputer103 = SimpleImputer(strategy="median")

E_train103 = imputer103.fit_transform(
    train103[environment103]
)
E_val103 = imputer103.transform(
    val103[environment103]
)

scaler103 = StandardScaler()

E_train103 = scaler103.fit_transform(E_train103)
E_val103 = scaler103.transform(E_val103)

# Training-derived spline boundaries.
lower103 = E_train103.min(axis=0)
upper103 = E_train103.max(axis=0)

span103 = upper103 - lower103
margin103 = np.maximum(span103 * 0.10, 0.10)

lower_bound103 = lower103 - margin103
upper_bound103 = upper103 + margin103

E_val_safe103 = np.clip(
    E_val103,
    lower_bound103,
    upper_bound103
)

knot_kwargs103 = [
    {
        "lower_bound": float(lower_bound103[j]),
        "upper_bound": float(upper_bound103[j])
    }
    for j in range(len(environment103))
]

spline103 = BSplines(
    E_train103,
    df=[4] * len(environment103),
    degree=[3] * len(environment103),
    variable_names=environment103,
    knot_kwds=knot_kwargs103
)

E_val_smooth103 = spline103.transform(E_val_safe103)

# Include intercept and incidence in the linear component.
X_linear_train103 = sm.add_constant(
    history_train103,
    has_constant="add"
)

X_linear_val103 = sm.add_constant(
    history_val103,
    has_constant="add"
)

# Limited smoothing penalty candidates.
alphas103 = [0.1, 1.0, 10.0]

results103 = []
fitted103 = {}
predictions103 = {}

for alpha in alphas103:

    model = GLMGam(
        endog=delta_train103,
        exog=X_linear_train103,
        smoother=spline103,
        family=sm.families.Gaussian(),
        alpha=alpha
    )

    fitted = model.fit()

    predicted_delta = fitted.predict(
        exog=X_linear_val103,
        exog_smooth=E_val_smooth103
    )

    predicted_incidence = np.maximum(
        history_val103 + predicted_delta,
        0
    )

    name = f"Residual_GAM_alpha_{alpha}"

    results103.append({
        "Model": name,
        "Alpha": alpha,
        "MAE": mean_absolute_error(
            y_val103, predicted_incidence
        ),
        "RMSE": np.sqrt(
            mean_squared_error(
                y_val103, predicted_incidence
            )
        ),
        "R2": r2_score(
            y_val103, predicted_incidence
        )
    })

    fitted103[name] = fitted
    predictions103[name] = predicted_incidence

results103 = pd.DataFrame(results103)

print("=" * 76)
print("CELL 103 — RESIDUAL GAM VALIDATION")
print("=" * 76)

print(
    results103.sort_values("RMSE")
    .round(5)
    .to_string(index=False)
)

print("\nREFERENCE BASELINES")
print("Persistence: RMSE=21.42910, R2=0.95320")
print("Fitted AR1: RMSE=22.09250, R2=0.95030")
print("Residual ML RF: RMSE=22.02184, R2=0.95059")
print("Residual BIML B2 RF: RMSE=20.58268, R2=0.95684")

print("\nCELL 103 COMPLETE")
print("No test data accessed")
```

```python

# ============================================================
# CELL 104 — MATCHED RESIDUAL GAM VS BIOGAM
# BIML 2.0 | VALIDATION 2021–2022
# SAME REGULARIZATION | NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm

from statsmodels.gam.api import GLMGam, BSplines
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "prospective" in globals()
assert "feature_groups" in globals()

train104 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val104 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

target104 = "Target_Incidence_t_plus_1"

history_train104 = train104[
    "Incidence_t"
].to_numpy(dtype=float)

history_val104 = val104[
    "Incidence_t"
].to_numpy(dtype=float)

delta_train104 = (
    train104[target104].to_numpy(dtype=float)
    - history_train104
)

y_val104 = val104[target104].to_numpy(dtype=float)

feature_sets104 = {
    "Residual_GAM": list(feature_groups["E"]),
    "Residual_BioGAM_B2": (
        list(feature_groups["E"])
        + list(feature_groups["B2"])
    )
}

ALPHA104 = 10.0

results104 = []
predictions104 = {}
fitted104 = {}

for name, features in feature_sets104.items():

    imputer = SimpleImputer(strategy="median")

    X_train = imputer.fit_transform(
        train104[features]
    )
    X_val = imputer.transform(
        val104[features]
    )

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X_train)
    X_val = scaler.transform(X_val)

    lower = X_train.min(axis=0)
    upper = X_train.max(axis=0)

    span = upper - lower
    margin = np.maximum(span * 0.10, 0.10)

    lower_bound = lower - margin
    upper_bound = upper + margin

    X_val_safe = np.clip(
        X_val, lower_bound, upper_bound
    )

    knot_kwargs = [
        {
            "lower_bound": float(lower_bound[j]),
            "upper_bound": float(upper_bound[j])
        }
        for j in range(len(features))
    ]

    spline = BSplines(
        X_train,
        df=[4] * len(features),
        degree=[3] * len(features),
        variable_names=features,
        knot_kwds=knot_kwargs
    )

    X_linear_train = sm.add_constant(
        history_train104,
        has_constant="add"
    )

    X_linear_val = sm.add_constant(
        history_val104,
        has_constant="add"
    )

    model = GLMGam(
        endog=delta_train104,
        exog=X_linear_train,
        smoother=spline,
        family=sm.families.Gaussian(),
        alpha=ALPHA104
    )

    fitted = model.fit()

    X_val_smooth = spline.transform(X_val_safe)

    predicted_delta = fitted.predict(
        exog=X_linear_val,
        exog_smooth=X_val_smooth
    )

    predicted_incidence = np.maximum(
        history_val104 + predicted_delta,
        0
    )

    results104.append({
        "Model": name,
        "N": len(y_val104),
        "MAE": mean_absolute_error(
            y_val104, predicted_incidence
        ),
        "RMSE": np.sqrt(
            mean_squared_error(
                y_val104, predicted_incidence
            )
        ),
        "R2": r2_score(
            y_val104, predicted_incidence
        ),
        "Clipped_Values": int(
            np.sum(X_val_safe != X_val)
        )
    })

    predictions104[name] = predicted_incidence

    fitted104[name] = {
        "model": fitted,
        "spline": spline,
        "imputer": imputer,
        "scaler": scaler,
        "features": features,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound
    }

results104 = pd.DataFrame(results104)

gam104 = results104.loc[
    results104["Model"] == "Residual_GAM"
].iloc[0]

bio104 = results104.loc[
    results104["Model"] == "Residual_BioGAM_B2"
].iloc[0]

print("=" * 76)
print("CELL 104 — MATCHED GAM VS BIOGAM")
print("=" * 76)

print("\nVALIDATION PERFORMANCE")
print(
    results104.round(5).to_string(index=False)
)

print("\nBIOLOGICAL CONTRIBUTION")
print(
    "MAE reduction:",
    round(gam104["MAE"] - bio104["MAE"], 5)
)
print(
    "RMSE reduction:",
    round(gam104["RMSE"] - bio104["RMSE"], 5)
)
print(
    "R2 gain:",
    round(bio104["R2"] - gam104["R2"], 5)
)

print("\nCELL 104 COMPLETE")
print("Matched regularization:", ALPHA104)
print("No test data accessed")
```

```python

# ============================================================
# CELL 105 — INCIDENCE CHANGE PREDICTABILITY
# BIML 2.0 | TRAIN + VALIDATION ONLY
# NO MODEL FITTING | NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "prospective" in globals()
assert "feature_groups" in globals()

df105 = prospective.loc[
    prospective["Partition"].isin(["Train", "Validation"])
].copy()

df105["Delta_Incidence"] = (
    df105["Target_Incidence_t_plus_1"]
    - df105["Incidence_t"]
)

df105["Abs_Delta"] = df105["Delta_Incidence"].abs()

print("=" * 76)
print("CELL 105 — INCIDENCE CHANGE PREDICTABILITY")
print("=" * 76)

print("\nANNUAL INCIDENCE CHANGE BY PARTITION")
print(
    df105.groupby("Partition")[
        ["Delta_Incidence", "Abs_Delta"]
    ]
    .agg(["mean", "median", "std", "max"])
    .round(3)
    .to_string()
)

print("\nMAGNITUDE OF ANNUAL CHANGES")

for partition, group in df105.groupby("Partition"):

    abs_delta = group["Abs_Delta"]

    print(f"\n{partition}")
    print(f"N: {len(group)}")
    print(f"Median absolute change: {abs_delta.median():.3f}")
    print(f"75th percentile: {abs_delta.quantile(0.75):.3f}")
    print(f"90th percentile: {abs_delta.quantile(0.90):.3f}")
    print(f"Maximum absolute change: {abs_delta.max():.3f}")

# ------------------------------------------------------------
# Biological feature association with incidence changes
# ------------------------------------------------------------
biological105 = (
    list(feature_groups["B1"])
    + list(feature_groups["B2"])
)

rows105 = []

for feature in biological105:

    for partition, group in df105.groupby("Partition"):

        valid = group[
            [feature, "Delta_Incidence", "Abs_Delta"]
        ].dropna()

        rows105.append({
            "Feature": feature,
            "Partition": partition,
            "N": len(valid),
            "Corr_Delta": (
                valid[feature].corr(
                    valid["Delta_Incidence"]
                )
            ),
            "Corr_AbsDelta": (
                valid[feature].corr(
                    valid["Abs_Delta"]
                )
            )
        })

correlations105 = pd.DataFrame(rows105)

print("\nBIOLOGICAL ASSOCIATIONS WITH INCIDENCE CHANGE")
print(
    correlations105.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# Largest changes in the validation period
# ------------------------------------------------------------
print("\nLARGEST VALIDATION INCIDENCE CHANGES")

largest105 = (
    df105.loc[
        df105["Partition"] == "Validation",
        [
            "Country",
            "Target_Year",
            "Incidence_t",
            "Target_Incidence_t_plus_1",
            "Delta_Incidence",
            "Abs_Delta"
        ]
    ]
    .sort_values("Abs_Delta", ascending=False)
    .head(10)
)

print(
    largest105.round(3).to_string(index=False)
)

print("\nCELL 105 COMPLETE")
print("No model training")
print("No test data accessed")
```

```python

# ============================================================
# CELL 106 — INCIDENCE CHANGE DIRECTION PREDICTION
# BIML 2.0 | TRAIN + VALIDATION ONLY
# NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

assert "prospective" in globals()
assert "feature_groups" in globals()

train106 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val106 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

target106 = "Target_Incidence_t_plus_1"

for frame in (train106, val106):
    frame["Delta"] = (
        frame[target106] - frame["Incidence_t"]
    )
    frame["Direction"] = (
        frame["Delta"] > 0
    ).astype(int)

feature_sets106 = {
    "Direction_ML": (
        list(feature_groups["H"])
        + list(feature_groups["E"])
        + ["B2_supported_months"]
    ),
    "Direction_BIML_B2": (
        list(feature_groups["H"])
        + list(feature_groups["E"])
        + ["B2_supported_months"]
        + list(feature_groups["B2"])
    )
}

results106 = []
predictions106 = {}
fitted106 = {}

for name, features in feature_sets106.items():

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=400,
                min_samples_leaf=3,
                max_features=0.8,
                class_weight="balanced",
                random_state=20261016,
                n_jobs=-1
            )
        )
    ])

    model.fit(
        train106[features],
        train106["Direction"]
    )

    prediction = model.predict(
        val106[features]
    )

    actual = val106["Direction"].to_numpy()

    results106.append({
        "Model": name,
        "N": len(actual),
        "Accuracy": accuracy_score(
            actual, prediction
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            actual, prediction
        ),
        "Precision": precision_score(
            actual, prediction,
            zero_division=0
        ),
        "Recall": recall_score(
            actual, prediction,
            zero_division=0
        ),
        "F1": f1_score(
            actual, prediction,
            zero_division=0
        )
    })

    predictions106[name] = prediction
    fitted106[name] = model

results106 = pd.DataFrame(results106)

print("=" * 76)
print("CELL 106 — INCIDENCE CHANGE DIRECTION")
print("=" * 76)

print("\nCLASS DISTRIBUTION")

for name, frame in [
    ("Train", train106),
    ("Validation", val106)
]:
    counts = frame["Direction"].value_counts()
    print(
        f"{name}: "
        f"Increase={counts.get(1, 0)}, "
        f"No_Increase={counts.get(0, 0)}"
    )

print("\nVALIDATION RESULTS")
print(
    results106.round(5).to_string(index=False)
)

print("\nCONFUSION MATRICES")
print("Rows: actual [No increase, Increase]")
print("Columns: predicted [No increase, Increase]")

for name, prediction in predictions106.items():

    cm = confusion_matrix(
        val106["Direction"],
        prediction,
        labels=[0, 1]
    )

    print(f"\n{name}")
    print(cm)

print("\nCELL 106 COMPLETE")
print("No test data accessed")
```

```python

# ============================================================
# CELL 107 — INCIDENCE CHANGE MAGNITUDE
# BIML 2.0 | ML VS BIML B2
# TRAIN 2016–2020 | VALIDATION 2021–2022
# NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "prospective" in globals()
assert "feature_groups" in globals()

train107 = prospective.loc[
    prospective["Partition"] == "Train"
].copy()

val107 = prospective.loc[
    prospective["Partition"] == "Validation"
].copy()

target107 = "Target_Incidence_t_plus_1"

for frame in (train107, val107):
    frame["Delta"] = (
        frame[target107] - frame["Incidence_t"]
    )
    frame["Magnitude"] = frame["Delta"].abs()

feature_sets107 = {
    "Magnitude_ML": (
        list(feature_groups["H"])
        + list(feature_groups["E"])
        + ["B2_supported_months"]
    ),
    "Magnitude_BIML_B2": (
        list(feature_groups["H"])
        + list(feature_groups["E"])
        + ["B2_supported_months"]
        + list(feature_groups["B2"])
    )
}

y_train107 = np.log1p(
    train107["Magnitude"].to_numpy(dtype=float)
)

y_val107 = val107["Magnitude"].to_numpy(dtype=float)

results107 = []
predictions107 = {}
fitted107 = {}

for name, features in feature_sets107.items():

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=400,
                min_samples_leaf=3,
                max_features=0.8,
                random_state=20261016,
                n_jobs=-1
            )
        )
    ])

    model.fit(
        train107[features],
        y_train107
    )

    predicted_log_magnitude = model.predict(
        val107[features]
    )

    predicted_magnitude = np.maximum(
        np.expm1(predicted_log_magnitude),
        0
    )

    results107.append({
        "Model": name,
        "N": len(y_val107),
        "MAE": mean_absolute_error(
            y_val107, predicted_magnitude
        ),
        "RMSE": np.sqrt(
            mean_squared_error(
                y_val107, predicted_magnitude
            )
        ),
        "R2": r2_score(
            y_val107, predicted_magnitude
        )
    })

    predictions107[name] = predicted_magnitude
    fitted107[name] = model

results107 = pd.DataFrame(results107)

print("=" * 76)
print("CELL 107 — MAGNITUDE PREDICTION")
print("=" * 76)

print("\nVALIDATION RESULTS")
print(
    results107.round(5).to_string(index=False)
)

ml107 = results107.loc[
    results107["Model"] == "Magnitude_ML"
].iloc[0]

biml107 = results107.loc[
    results107["Model"] == "Magnitude_BIML_B2"
].iloc[0]

print("\nBIOLOGICAL CONTRIBUTION")
print(
    "MAE reduction:",
    round(ml107["MAE"] - biml107["MAE"], 5)
)
print(
    "RMSE reduction:",
    round(ml107["RMSE"] - biml107["RMSE"], 5)
)
print(
    "R2 gain:",
    round(biml107["R2"] - ml107["R2"], 5)
)

print("\nREFERENCE")
print(
    "Validation mean absolute incidence change:",
    round(y_val107.mean(), 5)
)
print(
    "Validation median absolute incidence change:",
    round(np.median(y_val107), 5)
)

print("\nCELL 107 COMPLETE")
print("No test data accessed")
```

```python

# ============================================================
# CELL 109 — MONTHLY BIOLOGICAL FEATURE ENGINEERING
# BIML 2.0 | B1 + B2
# NO OUTCOME ACCESS | NO MODEL TRAINING
# ============================================================

import numpy as np
import pandas as pd

assert "b1_monthly108" in globals()
assert "b2_monthly108" in globals()

keys109 = ["Country", "Year", "Month"]
annual_keys109 = ["Country", "Year"]

b1_109 = b1_monthly108.copy()
b2_109 = b2_monthly108.copy()

# ------------------------------------------------------------
# 1. Validate monthly records
# ------------------------------------------------------------

for name, frame in [
    ("B1", b1_109),
    ("B2", b2_109)
]:
    assert not frame.duplicated(keys109).any(), (
        f"{name}: duplicate monthly keys"
    )

    assert frame["Month"].between(1, 12).all(), (
        f"{name}: invalid month"
    )

# ------------------------------------------------------------
# 2. B1 posterior and seasonal features
# ------------------------------------------------------------

b1_value109 = "B1_posterior_full_median"

b1_109["B1_interval_width"] = (
    b1_109["B1_posterior_full_q975"]
    - b1_109["B1_posterior_full_q025"]
)

b1_features109 = (
    b1_109.groupby(annual_keys109)
    .agg(
        B1_season_mean=(
            b1_value109, "mean"
        ),
        B1_season_std=(
            b1_value109, "std"
        ),
        B1_season_max=(
            b1_value109, "max"
        ),
        B1_season_min=(
            b1_value109, "min"
        ),
        B1_uncertainty_width_mean=(
            "B1_interval_width", "mean"
        ),
        B1_probability_positive_mean=(
            "B1_posterior_full_probability_positive",
            "mean"
        ),
        B1_months=(
            "Month", "nunique"
        )
    )
    .reset_index()
)

# ------------------------------------------------------------
# 3. B2 supported-month features
# ------------------------------------------------------------

support109 = b2_109["B2_temperature_supported"]

assert support109.notna().all(), (
    "Missing B2 support flags"
)

assert (
    b2_109.loc[
        ~support109,
        "B2_deterministic_competence_biting_survival_index"
    ].isna().all()
), "Unsupported B2 index unexpectedly contains values"

b2_supported109 = b2_109.loc[support109].copy()

b2_index109 = (
    "B2_deterministic_competence_biting_survival_index"
)

b2_supported109["B2_EIP_inverse"] = (
    1.0 /
    b2_supported109[
        "B2_Suh_EIP_mean_days_pooled_median"
    ]
)

b2_features109 = (
    b2_supported109.groupby(annual_keys109)
    .agg(
        B2_index_season_mean=(
            b2_index109, "mean"
        ),
        B2_index_season_std=(
            b2_index109, "std"
        ),
        B2_index_season_max=(
            b2_index109, "max"
        ),
        B2_index_season_min=(
            b2_index109, "min"
        ),
        B2_competence_season_mean=(
            "B2_Suh_vector_competence_pooled_median",
            "mean"
        ),
        B2_biting_season_mean=(
            "B2_An_stephensi_biting_median",
            "mean"
        ),
        B2_EIP_season_mean=(
            "B2_Suh_EIP_mean_days_pooled_median",
            "mean"
        ),
        B2_survival_season_mean=(
            "B2_feed1_survival_at_mean_EIP",
            "mean"
        ),
        B2_inverse_EIP_mean=(
            "B2_EIP_inverse",
            "mean"
        )
    )
    .reset_index()
)

# ------------------------------------------------------------
# 4. Upper-quartile seasonal intensity
# ------------------------------------------------------------

def upper_quartile_mean109(series):
    values = series.dropna().to_numpy(dtype=float)

    if len(values) == 0:
        return np.nan

    threshold = np.quantile(values, 0.75)

    return float(
        values[values >= threshold].mean()
    )

b2_upper109 = (
    b2_supported109.groupby(annual_keys109)[
        b2_index109
    ]
    .apply(upper_quartile_mean109)
    .rename("B2_index_upper_quartile_mean")
    .reset_index()
)

b2_features109 = b2_features109.merge(
    b2_upper109,
    on=annual_keys109,
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 5. Coverage from all original B2 months
# ------------------------------------------------------------

b2_coverage109 = (
    b2_109.groupby(annual_keys109)
    .agg(
        B2_months_total=("Month", "nunique"),
        B2_supported_months=(
            "B2_temperature_supported", "sum"
        )
    )
    .reset_index()
)

b2_coverage109["B2_supported_fraction"] = (
    b2_coverage109["B2_supported_months"]
    / b2_coverage109["B2_months_total"]
)

b2_features109 = b2_coverage109.merge(
    b2_features109,
    on=annual_keys109,
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 6. Peak-to-mean ratio
# ------------------------------------------------------------

denominator109 = b2_features109[
    "B2_index_season_mean"
].replace(0, np.nan)

b2_features109["B2_index_peak_to_mean"] = (
    b2_features109["B2_index_season_max"]
    / denominator109
)

# ------------------------------------------------------------
# 7. Merge annual B1 and B2 features
# ------------------------------------------------------------

bio_features109 = b1_features109.merge(
    b2_features109,
    on=annual_keys109,
    how="outer",
    validate="one_to_one",
    indicator=True
)

assert (
    bio_features109["_merge"] == "both"
).all(), "B1/B2 country-year coverage mismatch"

bio_features109 = bio_features109.drop(
    columns="_merge"
)

assert not bio_features109.duplicated(
    annual_keys109
).any()

assert (
    bio_features109["B1_months"] == 12
).all()

assert (
    bio_features109["B2_months_total"] == 12
).all()

# ------------------------------------------------------------
# 8. Inspect engineered feature table
# ------------------------------------------------------------

print("=" * 76)
print("CELL 109 — MONTHLY BIOLOGICAL FEATURE ENGINEERING")
print("=" * 76)

print("\nANNUAL FEATURE TABLE")
print("Shape:", bio_features109.shape)

print("\nFEATURE COLUMNS")
print(
    bio_features109.columns.tolist()
)

print("\nB2 SUPPORTED-MONTH DISTRIBUTION")
print(
    bio_features109[
        "B2_supported_months"
    ].value_counts().sort_index().to_string()
)

print("\nMISSING FEATURE VALUES")
print(
    bio_features109.isna()
    .sum()
    .loc[lambda s: s > 0]
    .to_string()
)

print("\nEXAMPLE COUNTRY-YEAR FEATURES")
print(
    bio_features109.head(5)
    .round(5)
    .to_string(index=False)
)

print("\nCELL 109 COMPLETE")
print("No outcome data accessed")
print("No models fitted")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 110 — BIOLOGICAL FEATURE ALIGNMENT AND AUDIT
# BIML 2.0 | MONTHLY-DERIVED BIOLOGICAL FEATURES
# NO MODEL FITTING | NO TEST TARGET INSPECTION
# ============================================================

import numpy as np
import pandas as pd

assert "prospective" in globals()
assert "bio_features109" in globals()

# ------------------------------------------------------------
# 1. Identify the predictor year
# ------------------------------------------------------------

base110 = prospective.copy()

assert "Target_Year" in base110.columns
assert "Country" in base110.columns
assert "Partition" in base110.columns

base110["Biological_Year"] = (
    pd.to_numeric(
        base110["Target_Year"],
        errors="raise"
    ).astype(int) - 1
)

print("=" * 76)
print("CELL 110 — BIOLOGICAL FEATURE ALIGNMENT")
print("=" * 76)

print("\nFORECAST YEAR RELATIONSHIP")
print(
    base110[
        ["Country", "Target_Year", "Biological_Year", "Partition"]
    ].head(8).to_string(index=False)
)

# ------------------------------------------------------------
# 2. Join annual biological features
# ------------------------------------------------------------

bio110 = bio_features109.rename(
    columns={"Year": "Biological_Year"}
).copy()

assert not bio110.duplicated(
    ["Country", "Biological_Year"]
).any()

enhanced110 = base110.merge(
    bio110,
    on=["Country", "Biological_Year"],
    how="left",
    validate="many_to_one",
    indicator="_bio_merge110",
    suffixes=("", "_monthly110")
)

assert len(enhanced110) == len(base110), (
    "Row count changed during merge"
)

print("\nMERGE COVERAGE")
print(
    enhanced110["_bio_merge110"]
    .value_counts(dropna=False)
    .to_string()
)

missing110 = enhanced110.loc[
    enhanced110["_bio_merge110"] != "both",
    ["Country", "Target_Year", "Biological_Year", "Partition"]
]

if len(missing110):
    print("\nUNMATCHED OBSERVATIONS")
    print(missing110.to_string(index=False))

assert len(missing110) == 0, (
    "Some forecasting rows lack biological features"
)

# ------------------------------------------------------------
# 3. Check original versus reconstructed annual means
# ------------------------------------------------------------

comparisons110 = [
    ("B1_full_mean", "B1_season_mean"),
    (
        "B2_index_supported_mean",
        "B2_index_season_mean"
    ),
    (
        "B2_competence_supported_mean",
        "B2_competence_season_mean"
    ),
    (
        "B2_EIP_supported_mean",
        "B2_EIP_season_mean"
    ),
    (
        "B2_biting_supported_mean",
        "B2_biting_season_mean"
    )
]

audit110 = []

for old_col, new_col in comparisons110:

    assert old_col in enhanced110.columns, old_col
    assert new_col in enhanced110.columns, new_col

    old_values = pd.to_numeric(
        enhanced110[old_col],
        errors="coerce"
    )

    new_values = pd.to_numeric(
        enhanced110[new_col],
        errors="coerce"
    )

    valid = old_values.notna() & new_values.notna()

    assert valid.any(), (
        f"No valid comparisons for {old_col}"
    )

    diff = (
        old_values[valid] - new_values[valid]
    ).abs()

    audit110.append({
        "Original": old_col,
        "Reconstructed": new_col,
        "N": int(valid.sum()),
        "Max_Abs_Diff": float(diff.max()),
        "Mean_Abs_Diff": float(diff.mean())
    })

audit110 = pd.DataFrame(audit110)

print("\nORIGINAL VS RECONSTRUCTED ANNUAL FEATURES")
print(
    audit110.to_string(
        index=False,
        float_format=lambda x: f"{x:.10f}"
    )
)

# ------------------------------------------------------------
# 4. Verify partition integrity
# ------------------------------------------------------------

print("\nPARTITION INTEGRITY")
print(
    enhanced110["Partition"]
    .value_counts()
    .to_string()
)

assert (
    enhanced110["Partition"].to_numpy()
    == base110["Partition"].to_numpy()
).all()

assert (
    enhanced110["Target_Year"].to_numpy()
    == base110["Target_Year"].to_numpy()
).all()

# ------------------------------------------------------------
# 5. Check newly engineered features
# ------------------------------------------------------------

new_features110 = [
    "B1_season_std",
    "B1_season_max",
    "B1_uncertainty_width_mean",
    "B1_probability_positive_mean",
    "B2_index_season_std",
    "B2_index_season_max",
    "B2_index_upper_quartile_mean",
    "B2_index_peak_to_mean",
    "B2_supported_fraction"
]

print("\nNEW FEATURE MISSINGNESS")
print(
    enhanced110[new_features110]
    .isna()
    .sum()
    .to_string()
)

print("\nB2 COVERAGE BY PARTITION")
print(
    enhanced110.groupby("Partition")[
        "B2_supported_months_monthly110"
        if "B2_supported_months_monthly110"
        in enhanced110.columns
        else "B2_supported_months"
    ]
    .agg(["min", "median", "max"])
    .to_string()
)

print("\nCELL 110 COMPLETE")
print("Biological features aligned to predictor year")
print("No model fitting")
print("No test target inspection")
```

```python

# ============================================================
# CELL 111 — ENHANCED BIML VALIDATION EXPERIMENT
# BIML 2.0 | CONTROLLED FEATURE ABLATION
# TRAIN 2016–2020 | VALIDATION 2021–2022
# NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "enhanced110" in globals()
assert "feature_groups" in globals()

train111 = enhanced110.loc[
    enhanced110["Partition"] == "Train"
].copy()

val111 = enhanced110.loc[
    enhanced110["Partition"] == "Validation"
].copy()

target111 = "Target_Incidence_t_plus_1"

history111 = list(feature_groups["H"])
environment111 = list(feature_groups["E"])
biology111 = list(feature_groups["B2"])

# Predefined compact seasonal representation
seasonal111 = [
    "B1_season_std",
    "B1_uncertainty_width_mean",
    "B2_index_season_std",
    "B2_index_season_max",
    "B2_index_peak_to_mean",
    "B2_supported_fraction"
]

coverage111 = ["B2_supported_months"]

feature_sets111 = {
    "ML": (
        history111
        + environment111
        + coverage111
    ),
    "BIML_B2": (
        history111
        + environment111
        + coverage111
        + biology111
    ),
    "Enhanced_BIML": (
        history111
        + environment111
        + coverage111
        + biology111
        + seasonal111
    )
}

# Prevent accidental duplicate predictors
for name, features in feature_sets111.items():
    assert len(features) == len(set(features)), (
        f"Duplicate features in {name}"
    )

    missing = [
        c for c in features
        if c not in enhanced110.columns
    ]

    assert not missing, (
        f"{name}: missing columns {missing}"
    )

y_train111 = (
    train111[target111].to_numpy(dtype=float)
    - train111["Incidence_t"].to_numpy(dtype=float)
)

y_val111 = val111[target111].to_numpy(dtype=float)

history_val111 = val111[
    "Incidence_t"
].to_numpy(dtype=float)

results111 = []
predictions111 = {}
fitted111 = {}

def metrics111(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, y_pred)
        ),
        "R2": r2_score(y_true, y_pred)
    }

# ------------------------------------------------------------
# Persistence baseline
# ------------------------------------------------------------

results111.append({
    "Model": "Persistence",
    "N": len(y_val111),
    **metrics111(y_val111, history_val111)
})

predictions111["Persistence"] = history_val111.copy()

# ------------------------------------------------------------
# Matched Random Forest experiments
# ------------------------------------------------------------

for name, features in feature_sets111.items():

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=400,
                min_samples_leaf=3,
                max_features=1.0,
                random_state=20261016,
                n_jobs=-1
            )
        )
    ])

    model.fit(
        train111[features],
        y_train111
    )

    predicted_delta = model.predict(
        val111[features]
    )

    predicted_incidence = np.maximum(
        history_val111 + predicted_delta,
        0
    )

    results111.append({
        "Model": name,
        "N": len(y_val111),
        **metrics111(
            y_val111,
            predicted_incidence
        )
    })

    predictions111[name] = predicted_incidence
    fitted111[name] = model

results111 = pd.DataFrame(results111)

print("=" * 76)
print("CELL 111 — ENHANCED BIML VALIDATION")
print("=" * 76)

print("\nFEATURE COUNTS")
for name, features in feature_sets111.items():
    print(f"{name}: {len(features)} predictors")

print("\nVALIDATION RESULTS")
print(
    results111.sort_values("RMSE")
    .round(5)
    .to_string(index=False)
)

# ------------------------------------------------------------
# Matched biological contributions
# ------------------------------------------------------------

indexed111 = results111.set_index("Model")

for baseline, candidate in [
    ("ML", "BIML_B2"),
    ("BIML_B2", "Enhanced_BIML"),
    ("ML", "Enhanced_BIML")
]:

    mae_reduction = (
        indexed111.loc[baseline, "MAE"]
        - indexed111.loc[candidate, "MAE"]
    )

    rmse_reduction = (
        indexed111.loc[baseline, "RMSE"]
        - indexed111.loc[candidate, "RMSE"]
    )

    print(f"\n{baseline} → {candidate}")
    print(f"MAE reduction: {mae_reduction:.5f}")
    print(f"RMSE reduction: {rmse_reduction:.5f}")

print("\nCELL 111 COMPLETE")
print("Train rows:", len(train111))
print("Validation rows:", len(val111))
print("No test target access")
```

```python

# ============================================================
# CELL 112 — ROLLING-ORIGIN BIML EVALUATION
# BIML 2.0 | TEMPORAL ROBUSTNESS AUDIT
# DEVELOPMENT YEARS ONLY | NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "enhanced110" in globals()
assert "feature_sets111" in globals()

data112 = enhanced110.loc[
    enhanced110["Target_Year"] <= 2022
].copy()

target112 = "Target_Incidence_t_plus_1"

evaluation_years112 = [2019, 2020, 2021, 2022]

results112 = []
predictions112 = {}

for evaluation_year in evaluation_years112:

    train = data112.loc[
        data112["Target_Year"] < evaluation_year
    ].copy()

    val = data112.loc[
        data112["Target_Year"] == evaluation_year
    ].copy()

    assert train["Target_Year"].max() < evaluation_year

    if len(train) == 0 or len(val) == 0:
        raise ValueError(
            f"Insufficient observations for {evaluation_year}"
        )

    y_train_delta = (
        train[target112].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    y_val = val[target112].to_numpy(dtype=float)

    history_val = val[
        "Incidence_t"
    ].to_numpy(dtype=float)

    # Persistence
    persistence_metrics = {
        "Evaluation_Year": evaluation_year,
        "Model": "Persistence",
        "Train_N": len(train),
        "Val_N": len(val),
        "MAE": mean_absolute_error(
            y_val, history_val
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, history_val)
        ),
        "R2": r2_score(y_val, history_val)
    }

    results112.append(persistence_metrics)

    predictions112[
        (evaluation_year, "Persistence")
    ] = history_val.copy()

    # Matched RF experiments
    for name, features in feature_sets111.items():

        model = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "regressor",
                RandomForestRegressor(
                    n_estimators=400,
                    min_samples_leaf=3,
                    max_features=1.0,
                    random_state=20261016,
                    n_jobs=-1
                )
            )
        ])

        model.fit(
            train[features],
            y_train_delta
        )

        predicted_delta = model.predict(
            val[features]
        )

        predicted_incidence = np.maximum(
            history_val + predicted_delta,
            0
        )

        results112.append({
            "Evaluation_Year": evaluation_year,
            "Model": name,
            "Train_N": len(train),
            "Val_N": len(val),
            "MAE": mean_absolute_error(
                y_val, predicted_incidence
            ),
            "RMSE": np.sqrt(
                mean_squared_error(
                    y_val, predicted_incidence
                )
            ),
            "R2": r2_score(
                y_val, predicted_incidence
            )
        })

        predictions112[
            (evaluation_year, name)
        ] = predicted_incidence

results112 = pd.DataFrame(results112)

print("=" * 76)
print("CELL 112 — ROLLING-ORIGIN BIML EVALUATION")
print("=" * 76)

print("\nPER-YEAR RMSE")
print(
    results112.pivot(
        index="Evaluation_Year",
        columns="Model",
        values="RMSE"
    ).round(4).to_string()
)

print("\nPER-YEAR MAE")
print(
    results112.pivot(
        index="Evaluation_Year",
        columns="Model",
        values="MAE"
    ).round(4).to_string()
)

print("\nBIOLOGICAL CONTRIBUTION BY YEAR")

rmse112 = results112.pivot(
    index="Evaluation_Year",
    columns="Model",
    values="RMSE"
)

contribution112 = pd.DataFrame({
    "B2_vs_ML_RMSE_reduction": (
        rmse112["ML"] - rmse112["BIML_B2"]
    ),
    "Enhanced_vs_B2_RMSE_reduction": (
        rmse112["BIML_B2"]
        - rmse112["Enhanced_BIML"]
    ),
    "B2_vs_Persistence_RMSE_reduction": (
        rmse112["Persistence"]
        - rmse112["BIML_B2"]
    )
})

print(
    contribution112.round(4).to_string()
)

print("\nNUMBER OF YEARS WITH POSITIVE RMSE CONTRIBUTION")
print(
    (contribution112 > 0).sum().to_string()
)

print("\nCELL 112 COMPLETE")
print("Evaluation years:", evaluation_years112)
print("No test data accessed")
```

```python

# ============================================================
# CELL 113 — BIOLOGICAL VARIATION DECOMPOSITION
# BIML 2.0 | BETWEEN-COUNTRY VS WITHIN-COUNTRY
# DEVELOPMENT PERIOD ONLY | NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "enhanced110" in globals()

data113 = enhanced110.loc[
    enhanced110["Target_Year"] <= 2022
].copy()

features113 = [
    "B1_full_mean",
    "B2_competence_supported_mean",
    "B2_EIP_supported_mean",
    "B2_biting_supported_mean",
    "B2_index_supported_mean",
    "B2_index_season_std",
    "B2_index_season_max"
]

rows113 = []

for feature in features113:

    df = data113[
        ["Country", "Target_Year", feature]
    ].dropna().copy()

    country_mean = df.groupby("Country")[
        feature
    ].transform("mean")

    within = df[feature] - country_mean

    total_variance = df[feature].var(ddof=0)

    between_variance = country_mean.var(ddof=0)

    within_variance = within.var(ddof=0)

    rows113.append({
        "Feature": feature,
        "N": len(df),
        "Countries": df["Country"].nunique(),
        "Total_Variance": total_variance,
        "Between_Share": (
            between_variance / total_variance
            if total_variance > 0 else np.nan
        ),
        "Within_Share": (
            within_variance / total_variance
            if total_variance > 0 else np.nan
        ),
        "Within_SD": within.std(ddof=0)
    })

results113 = pd.DataFrame(rows113)

print("=" * 76)
print("CELL 113 — BIOLOGICAL VARIATION DECOMPOSITION")
print("=" * 76)

print("\nBETWEEN VS WITHIN COUNTRY VARIATION")
print(
    results113[
        [
            "Feature",
            "N",
            "Countries",
            "Between_Share",
            "Within_Share",
            "Within_SD"
        ]
    ].round(5).to_string(index=False)
)

print("\nCHECK: VARIANCE SHARES SUM TO ONE")

results113["Share_Sum"] = (
    results113["Between_Share"]
    + results113["Within_Share"]
)

print(
    results113[
        ["Feature", "Share_Sum"]
    ].round(8).to_string(index=False)
)

print("\nCELL 113 COMPLETE")
print("Descriptive decomposition only")
print("No test data accessed")
```

```python

# ============================================================
# CELL 114 — BIOLOGICAL TEMPORAL CHANGE FEATURES
# BIML 2.0 | YEAR-OVER-YEAR BIOLOGICAL DYNAMICS
# NO MODEL FITTING | NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "bio_features109" in globals()
assert "enhanced110" in globals()

bio114 = bio_features109.copy()

bio114 = bio114.sort_values(
    ["Country", "Year"]
).reset_index(drop=True)

biological_levels114 = [
    "B1_season_mean",
    "B2_competence_season_mean",
    "B2_EIP_season_mean",
    "B2_biting_season_mean",
    "B2_index_season_mean",
    "B2_index_season_std",
    "B2_index_season_max"
]

change_features114 = []

for feature in biological_levels114:

    previous = bio114.groupby("Country")[
        feature
    ].shift(1)

    change_name = f"{feature}_change"

    bio114[change_name] = (
        bio114[feature] - previous
    )

    change_features114.append(change_name)

# ------------------------------------------------------------
# 1. Confirm lag continuity
# ------------------------------------------------------------

previous_year114 = bio114.groupby("Country")[
    "Year"
].shift(1)

valid_previous114 = (
    bio114["Year"] - previous_year114 == 1
)

for feature in change_features114:
    bio114.loc[
        ~valid_previous114, feature
    ] = np.nan

# ------------------------------------------------------------
# 2. Merge with forecasting observations
# ------------------------------------------------------------

change_table114 = bio114[
    ["Country", "Year"] + change_features114
].rename(
    columns={"Year": "Biological_Year"}
)

enhanced114 = enhanced110.merge(
    change_table114,
    on=["Country", "Biological_Year"],
    how="left",
    validate="many_to_one"
)

assert len(enhanced114) == len(enhanced110)

# ------------------------------------------------------------
# 3. Audit development-period changes
# ------------------------------------------------------------

development114 = enhanced114.loc[
    enhanced114["Target_Year"] <= 2022
].copy()

rows114 = []

for feature in change_features114:

    values = development114[feature].dropna()

    rows114.append({
        "Feature": feature,
        "N": len(values),
        "Mean": values.mean(),
        "SD": values.std(),
        "Median_Abs_Change": values.abs().median(),
        "P90_Abs_Change": values.abs().quantile(0.90),
        "Zero_Fraction": (
            (values.abs() < 1e-12).mean()
        )
    })

results114 = pd.DataFrame(rows114)

print("=" * 76)
print("CELL 114 — BIOLOGICAL TEMPORAL CHANGES")
print("=" * 76)

print("\nCHANGE FEATURE SUMMARY")
print(
    results114.round(6).to_string(index=False)
)

print("\nMISSING CHANGES BY PARTITION")
print(
    enhanced114.groupby("Partition")[
        change_features114
    ].apply(
        lambda x: x.isna().sum().sum()
    ).to_string()
)

print("\nEXAMPLE TEMPORAL ALIGNMENT")
print(
    enhanced114[
        [
            "Country",
            "Target_Year",
            "Biological_Year",
            "B2_index_season_mean",
            "B2_index_season_mean_change"
        ]
    ].head(10).round(6).to_string(index=False)
)

print("\nCELL 114 COMPLETE")
print("No models fitted")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 115 — BIML LEVELS VS BIOLOGICAL CHANGES
# BIML 2.0 | ROLLING-ORIGIN EVALUATION
# DEVELOPMENT YEARS 2016–2022 ONLY
# NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "enhanced114" in globals()
assert "feature_groups" in globals()

data115 = enhanced114.loc[
    enhanced114["Target_Year"] <= 2022
].copy()

target115 = "Target_Incidence_t_plus_1"

base_features115 = (
    list(feature_groups["H"])
    + list(feature_groups["E"])
    + ["B2_supported_months"]
)

level_features115 = list(feature_groups["B2"])

change_features115 = [
    "B2_competence_season_mean_change",
    "B2_EIP_season_mean_change",
    "B2_biting_season_mean_change",
    "B2_index_season_mean_change"
]

feature_sets115 = {
    "ML": base_features115,
    "BIML_Levels": (
        base_features115
        + level_features115
    ),
    "BIML_Changes": (
        base_features115
        + change_features115
    ),
    "BIML_Levels_Changes": (
        base_features115
        + level_features115
        + change_features115
    )
}

for name, features in feature_sets115.items():
    assert len(features) == len(set(features)), name
    assert set(features).issubset(data115.columns), name

evaluation_years115 = [2019, 2020, 2021, 2022]

results115 = []
predictions115 = {}

for year in evaluation_years115:

    train = data115.loc[
        data115["Target_Year"] < year
    ].copy()

    val = data115.loc[
        data115["Target_Year"] == year
    ].copy()

    assert len(train) > 0
    assert len(val) > 0
    assert train["Target_Year"].max() < year

    y_train_delta = (
        train[target115].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    y_val = val[target115].to_numpy(dtype=float)

    history_val = val[
        "Incidence_t"
    ].to_numpy(dtype=float)

    # Persistence
    results115.append({
        "Year": year,
        "Model": "Persistence",
        "N": len(val),
        "MAE": mean_absolute_error(
            y_val, history_val
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, history_val)
        ),
        "R2": r2_score(y_val, history_val)
    })

    predictions115[(year, "Persistence")] = history_val

    # Matched residual models
    for name, features in feature_sets115.items():

        model = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "regressor",
                RandomForestRegressor(
                    n_estimators=400,
                    min_samples_leaf=3,
                    max_features=1.0,
                    random_state=20261016,
                    n_jobs=-1
                )
            )
        ])

        model.fit(
            train[features],
            y_train_delta
        )

        predicted_delta = model.predict(
            val[features]
        )

        prediction = np.maximum(
            history_val + predicted_delta,
            0
        )

        results115.append({
            "Year": year,
            "Model": name,
            "N": len(val),
            "MAE": mean_absolute_error(
                y_val, prediction
            ),
            "RMSE": np.sqrt(
                mean_squared_error(y_val, prediction)
            ),
            "R2": r2_score(y_val, prediction)
        })

        predictions115[(year, name)] = prediction

results115 = pd.DataFrame(results115)

print("=" * 76)
print("CELL 115 — BIOLOGICAL LEVELS VS CHANGES")
print("=" * 76)

print("\nFEATURE COUNTS")
for name, features in feature_sets115.items():
    print(name, len(features))

print("\nROLLING-ORIGIN RMSE")
print(
    results115.pivot(
        index="Year",
        columns="Model",
        values="RMSE"
    ).round(4).to_string()
)

print("\nROLLING-ORIGIN MAE")
print(
    results115.pivot(
        index="Year",
        columns="Model",
        values="MAE"
    ).round(4).to_string()
)

# ------------------------------------------------------------
# Biological contribution relative to conventional ML
# ------------------------------------------------------------

rmse115 = results115.pivot(
    index="Year",
    columns="Model",
    values="RMSE"
)

contribution115 = pd.DataFrame({
    "Levels_vs_ML": (
        rmse115["ML"]
        - rmse115["BIML_Levels"]
    ),
    "Changes_vs_ML": (
        rmse115["ML"]
        - rmse115["BIML_Changes"]
    ),
    "LevelsChanges_vs_ML": (
        rmse115["ML"]
        - rmse115["BIML_Levels_Changes"]
    ),
    "LevelsChanges_vs_Levels": (
        rmse115["BIML_Levels"]
        - rmse115["BIML_Levels_Changes"]
    )
})

print("\nRMSE CONTRIBUTION (POSITIVE = IMPROVEMENT)")
print(
    contribution115.round(4).to_string()
)

print("\nYEARS WITH POSITIVE CONTRIBUTION")
print(
    (contribution115 > 0).sum().to_string()
)

print("\nCELL 115 COMPLETE")
print("No test data accessed")
```

```python

# ============================================================
# CELL 116 — DIAGNOSE 2020 FORECASTING FAILURE
# BIML 2.0 | ERROR ATTRIBUTION
# EXISTING ROLLING-ORIGIN PREDICTIONS ONLY
# NO MODEL TRAINING | NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "data115" in globals()
assert "predictions115" in globals()

YEAR116 = 2020

models116 = [
    "Persistence",
    "ML",
    "BIML_Levels",
    "BIML_Changes",
    "BIML_Levels_Changes"
]

validation116 = data115.loc[
    data115["Target_Year"] == YEAR116
].copy().reset_index(drop=True)

actual116 = validation116[
    "Target_Incidence_t_plus_1"
].to_numpy(dtype=float)

history116 = validation116[
    "Incidence_t"
].to_numpy(dtype=float)

diagnostics116 = validation116[
    ["Country", "Target_Year", "Incidence_t",
     "Target_Incidence_t_plus_1"]
].copy()

diagnostics116["Actual_Change"] = (
    actual116 - history116
)

for model_name in models116:

    prediction = np.asarray(
        predictions115[(YEAR116, model_name)],
        dtype=float
    )

    assert len(prediction) == len(actual116)

    diagnostics116[
        f"{model_name}_Prediction"
    ] = prediction

    diagnostics116[
        f"{model_name}_Predicted_Change"
    ] = prediction - history116

    diagnostics116[
        f"{model_name}_Abs_Error"
    ] = np.abs(actual116 - prediction)

    diagnostics116[
        f"{model_name}_Squared_Error"
    ] = (actual116 - prediction) ** 2

print("=" * 76)
print("CELL 116 — 2020 FORECASTING ERROR ATTRIBUTION")
print("=" * 76)

print("\nTOP 10 COUNTRIES BY BIML-CHANGES SQUARED ERROR")

top116 = diagnostics116.sort_values(
    "BIML_Changes_Squared_Error",
    ascending=False
).head(10)

columns116 = [
    "Country",
    "Incidence_t",
    "Target_Incidence_t_plus_1",
    "Actual_Change",
    "BIML_Changes_Predicted_Change",
    "BIML_Changes_Abs_Error",
    "Persistence_Abs_Error"
]

print(
    top116[columns116]
    .round(3)
    .to_string(index=False)
)

print("\nSQUARED ERROR CONCENTRATION")

for model_name in models116:

    errors = diagnostics116[
        f"{model_name}_Squared_Error"
    ].sort_values(ascending=False)

    total = errors.sum()

    top3_share = (
        errors.head(3).sum() / total
        if total > 0 else np.nan
    )

    print(
        f"{model_name:22s} "
        f"Top-3 share: {top3_share:.2%}"
    )

print("\nDIRECTION ERRORS")

actual_direction = (
    diagnostics116["Actual_Change"] > 0
)

for model_name in models116:

    predicted_direction = (
        diagnostics116[
            f"{model_name}_Predicted_Change"
        ] > 0
    )

    accuracy = (
        predicted_direction == actual_direction
    ).mean()

    print(
        f"{model_name:22s} "
        f"Direction accuracy: {accuracy:.3f}"
    )

print("\nCELL 116 COMPLETE")
print("Existing predictions only")
print("No model fitting")
print("No test data accessed")
```

```python

# ============================================================
# CELL 117 — RESIDUAL CORRECTION CALIBRATION AUDIT
# BIML 2.0 | FORECAST CORRECTION MAGNITUDES
# EXISTING DEVELOPMENT PREDICTIONS ONLY
# NO NEW MODEL FITTING | NO TEST ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "data115" in globals()
assert "predictions115" in globals()

models117 = [
    "ML",
    "BIML_Levels",
    "BIML_Changes",
    "BIML_Levels_Changes"
]

rows117 = []

for year in [2019, 2020, 2021, 2022]:

    subset = data115.loc[
        data115["Target_Year"] == year
    ].copy()

    actual_change = (
        subset["Target_Incidence_t_plus_1"].to_numpy(dtype=float)
        - subset["Incidence_t"].to_numpy(dtype=float)
    )

    history = subset[
        "Incidence_t"
    ].to_numpy(dtype=float)

    for name in models117:

        predicted = np.asarray(
            predictions115[(year, name)],
            dtype=float
        )

        predicted_change = predicted - history

        rows117.append({
            "Year": year,
            "Model": name,
            "N": len(subset),
            "Actual_Mean_Abs_Change": np.mean(
                np.abs(actual_change)
            ),
            "Predicted_Mean_Abs_Change": np.mean(
                np.abs(predicted_change)
            ),
            "Actual_Median_Abs_Change": np.median(
                np.abs(actual_change)
            ),
            "Predicted_Median_Abs_Change": np.median(
                np.abs(predicted_change)
            ),
            "Direction_Accuracy": np.mean(
                (actual_change > 0)
                == (predicted_change > 0)
            ),
            "Correction_Actual_Correlation": (
                np.corrcoef(
                    actual_change,
                    predicted_change
                )[0, 1]
                if np.std(actual_change) > 0
                and np.std(predicted_change) > 0
                else np.nan
            )
        })

results117 = pd.DataFrame(rows117)

print("=" * 76)
print("CELL 117 — RESIDUAL CORRECTION CALIBRATION")
print("=" * 76)

print("\nMEAN ABSOLUTE CORRECTION BY YEAR")
print(
    results117[
        [
            "Year",
            "Model",
            "Actual_Mean_Abs_Change",
            "Predicted_Mean_Abs_Change",
            "Correction_Actual_Correlation"
        ]
    ].round(4).to_string(index=False)
)

print("\nMEDIAN ABSOLUTE CORRECTION BY YEAR")
print(
    results117[
        [
            "Year",
            "Model",
            "Actual_Median_Abs_Change",
            "Predicted_Median_Abs_Change"
        ]
    ].round(4).to_string(index=False)
)

print("\nDIRECTION ACCURACY")
print(
    results117.pivot(
        index="Year",
        columns="Model",
        values="Direction_Accuracy"
    ).round(3).to_string()
)

print("\nCELL 117 COMPLETE")
print("No model fitting")
print("No correction weights optimized")
print("No test data accessed")
```

```python

# ============================================================
# CELL 118 — HISTORICAL INCIDENCE TRAJECTORY AUDIT
# BIML 2.0 | LAGGED INCIDENCE INFORMATION
# NO MODEL FITTING | NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "enhanced114" in globals()

trajectory118 = enhanced114.copy()

trajectory118 = trajectory118.sort_values(
    ["Country", "Target_Year"]
).reset_index(drop=True)

# Incidence_t corresponds to Target_Year - 1
trajectory118["Incidence_Year"] = (
    trajectory118["Target_Year"] - 1
)

# Previous available observation
trajectory118["Incidence_t_minus_1"] = (
    trajectory118.groupby("Country")[
        "Incidence_t"
    ].shift(1)
)

trajectory118["Previous_Incidence_Year"] = (
    trajectory118.groupby("Country")[
        "Incidence_Year"
    ].shift(1)
)

# Enforce exact one-year continuity
continuous118 = (
    trajectory118["Incidence_Year"]
    - trajectory118["Previous_Incidence_Year"]
    == 1
)

trajectory118.loc[
    ~continuous118,
    "Incidence_t_minus_1"
] = np.nan

trajectory118["Incidence_Change_t"] = (
    trajectory118["Incidence_t"]
    - trajectory118["Incidence_t_minus_1"]
)

trajectory118["Incidence_Relative_Change_t"] = (
    trajectory118["Incidence_Change_t"]
    / trajectory118[
        "Incidence_t_minus_1"
    ].clip(lower=1.0)
)

# Restrict descriptive analysis to development years
development118 = trajectory118.loc[
    trajectory118["Target_Year"] <= 2022
].copy()

print("=" * 76)
print("CELL 118 — HISTORICAL INCIDENCE TRAJECTORIES")
print("=" * 76)

print("\nMISSING HISTORICAL INCIDENCE BY PARTITION")
print(
    trajectory118.groupby("Partition")[
        [
            "Incidence_t_minus_1",
            "Incidence_Change_t",
            "Incidence_Relative_Change_t"
        ]
    ].apply(
        lambda x: x.isna().sum()
    ).to_string()
)

print("\nHISTORICAL CHANGE DISTRIBUTION")
print(
    development118[
        [
            "Incidence_Change_t",
            "Incidence_Relative_Change_t"
        ]
    ].describe().round(4).to_string()
)

print("\nEXAMPLE TEMPORAL ALIGNMENT")
print(
    trajectory118[
        [
            "Country",
            "Target_Year",
            "Incidence_Year",
            "Incidence_t_minus_1",
            "Incidence_t",
            "Incidence_Change_t"
        ]
    ].head(12).round(4).to_string(index=False)
)

print("\nAVAILABLE HISTORY BY TARGET YEAR")
print(
    development118.groupby("Target_Year")[
        "Incidence_Change_t"
    ].agg(
        N="size",
        Available="count"
    ).to_string()
)

print("\nCELL 118 COMPLETE")
print("No models fitted")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 119 — HISTORY-AWARE BIML COMPARISON
# BIML 2.0 | ROLLING-ORIGIN DEVELOPMENT EVALUATION
# 2019–2022 | NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "trajectory118" in globals()
assert "feature_groups" in globals()

data119 = trajectory118.loc[
    trajectory118["Target_Year"] <= 2022
].copy()

# Explicitly flag unavailable first-year history
data119["History_Missing"] = (
    data119["Incidence_t_minus_1"].isna()
).astype(int)

target119 = "Target_Incidence_t_plus_1"

history_features119 = [
    "Incidence_t",
    "Incidence_t_minus_1",
    "Incidence_Change_t",
    "History_Missing"
]

environment119 = list(feature_groups["E"])

coverage119 = ["B2_supported_months"]

biological_levels119 = list(feature_groups["B2"])

biological_changes119 = [
    "B2_competence_season_mean_change",
    "B2_EIP_season_mean_change",
    "B2_biting_season_mean_change",
    "B2_index_season_mean_change"
]

base119 = (
    history_features119
    + environment119
    + coverage119
)

feature_sets119 = {
    "History_ML": base119,
    "History_BIML_Levels": (
        base119 + biological_levels119
    ),
    "History_BIML_Changes": (
        base119 + biological_changes119
    )
}

for name, features in feature_sets119.items():
    assert len(features) == len(set(features)), name
    assert set(features).issubset(data119.columns), name

years119 = [2019, 2020, 2021, 2022]

results119 = []
predictions119 = {}

for year in years119:

    train = data119.loc[
        data119["Target_Year"] < year
    ].copy()

    val = data119.loc[
        data119["Target_Year"] == year
    ].copy()

    assert train["Target_Year"].max() < year

    y_train_delta = (
        train[target119].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    y_val = val[target119].to_numpy(dtype=float)

    history_val = val[
        "Incidence_t"
    ].to_numpy(dtype=float)

    # Persistence benchmark
    results119.append({
        "Year": year,
        "Model": "Persistence",
        "Train_N": len(train),
        "Val_N": len(val),
        "MAE": mean_absolute_error(
            y_val, history_val
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_val, history_val)
        ),
        "R2": r2_score(y_val, history_val)
    })

    predictions119[(year, "Persistence")] = (
        history_val.copy()
    )

    # Matched models
    for name, features in feature_sets119.items():

        model = Pipeline([
            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),
            (
                "regressor",
                RandomForestRegressor(
                    n_estimators=400,
                    min_samples_leaf=3,
                    max_features=1.0,
                    random_state=20261016,
                    n_jobs=-1
                )
            )
        ])

        model.fit(
            train[features],
            y_train_delta
        )

        predicted_delta = model.predict(
            val[features]
        )

        prediction = np.maximum(
            history_val + predicted_delta,
            0
        )

        results119.append({
            "Year": year,
            "Model": name,
            "Train_N": len(train),
            "Val_N": len(val),
            "MAE": mean_absolute_error(
                y_val, prediction
            ),
            "RMSE": np.sqrt(
                mean_squared_error(
                    y_val, prediction
                )
            ),
            "R2": r2_score(
                y_val, prediction
            )
        })

        predictions119[(year, name)] = prediction

results119 = pd.DataFrame(results119)

print("=" * 76)
print("CELL 119 — HISTORY-AWARE BIML COMPARISON")
print("=" * 76)

print("\nFEATURE COUNTS")
for name, features in feature_sets119.items():
    print(f"{name}: {len(features)}")

print("\nROLLING-ORIGIN RMSE")
print(
    results119.pivot(
        index="Year",
        columns="Model",
        values="RMSE"
    ).round(4).to_string()
)

print("\nROLLING-ORIGIN MAE")
print(
    results119.pivot(
        index="Year",
        columns="Model",
        values="MAE"
    ).round(4).to_string()
)

# ------------------------------------------------------------
# Incremental biological contribution
# ------------------------------------------------------------

rmse119 = results119.pivot(
    index="Year",
    columns="Model",
    values="RMSE"
)

contribution119 = pd.DataFrame({
    "Levels_vs_HistoryML": (
        rmse119["History_ML"]
        - rmse119["History_BIML_Levels"]
    ),
    "Changes_vs_HistoryML": (
        rmse119["History_ML"]
        - rmse119["History_BIML_Changes"]
    ),
    "Levels_vs_Persistence": (
        rmse119["Persistence"]
        - rmse119["History_BIML_Levels"]
    ),
    "Changes_vs_Persistence": (
        rmse119["Persistence"]
        - rmse119["History_BIML_Changes"]
    )
})

print("\nRMSE CONTRIBUTION")
print("Positive values indicate improvement")
print(
    contribution119.round(4).to_string()
)

print("\nYEARS WITH POSITIVE CONTRIBUTION")
print(
    (contribution119 > 0).sum().to_string()
)

print("\nCELL 119 COMPLETE")
print("History-aware comparison finished")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 125 — BIML ALGORITHM COMPARISON
# BIML 2.0 | RF vs GB vs XGBOOST
#
# GAM = INDEPENDENT STATISTICAL BASELINE
# ML = HISTORY + ENVIRONMENT
# BIML = ML + MECHANISTIC B2 FEATURES
#
# ROLLING ORIGINS: 2019, 2020, 2021, 2022
# NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

try:
    from xgboost import XGBRegressor
except ImportError as exc:
    raise ImportError(
        "XGBoost is required. Install it using "
        "%pip install xgboost, then rerun Cell 125."
    ) from exc


# ------------------------------------------------------------
# 1. Verify previous research objects
# ------------------------------------------------------------

assert "trajectory118" in globals()
assert "feature_groups" in globals()
assert "results119" in globals()
assert "results120" in globals()

data125 = trajectory118.loc[
    trajectory118["Target_Year"] <= 2022
].copy()

target125 = "Target_Incidence_t_plus_1"

assert data125["Target_Year"].max() == 2022
assert data125["Country"].nunique() == 29


# ------------------------------------------------------------
# 2. Matched feature configurations
# ------------------------------------------------------------

history125 = [
    "Incidence_t",
    "Incidence_t_minus_1",
    "Incidence_Change_t"
]

environment125 = list(feature_groups["E"])

coverage125 = ["B2_supported_months"]

biological_levels125 = list(feature_groups["B2"])

biological_changes125 = [
    "B2_competence_season_mean_change",
    "B2_EIP_season_mean_change",
    "B2_biting_season_mean_change",
    "B2_index_season_mean_change"
]

# Match Cell 119's history-missing indicator.
data125["History_Missing"] = (
    data125["Incidence_t_minus_1"]
    .isna()
    .astype(int)
)

history125.append("History_Missing")

base125 = (
    history125
    + environment125
    + coverage125
)

feature_sets125 = {
    "ML": base125,

    "BIML_Levels": (
        base125 + biological_levels125
    ),

    "BIML_Changes": (
        base125 + biological_changes125
    )
}

for name, features in feature_sets125.items():

    assert len(features) == len(set(features)), name

    assert set(features).issubset(
        data125.columns
    ), name


# ------------------------------------------------------------
# 3. Fixed model configurations
# ------------------------------------------------------------

SEED125 = 20261016

def build_model125(algorithm):

    if algorithm == "RF":

        estimator = RandomForestRegressor(
            n_estimators=400,
            min_samples_leaf=3,
            max_features=1.0,
            random_state=SEED125,
            n_jobs=-1
        )

    elif algorithm == "GB":

        estimator = GradientBoostingRegressor(
            n_estimators=150,
            learning_rate=0.03,
            max_depth=2,
            min_samples_leaf=5,
            subsample=0.8,
            random_state=SEED125
        )

    elif algorithm == "XGBoost":

        estimator = XGBRegressor(
            n_estimators=200,
            learning_rate=0.03,
            max_depth=2,
            min_child_weight=5,
            subsample=0.8,
            colsample_bytree=0.9,
            reg_lambda=10.0,
            objective="reg:squarederror",
            random_state=SEED125,
            n_jobs=4,
            verbosity=0
        )

    else:
        raise ValueError(
            f"Unknown algorithm: {algorithm}"
        )

    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "regressor",
            estimator
        )
    ])


# ------------------------------------------------------------
# 4. Rolling-origin evaluation
# ------------------------------------------------------------

years125 = [2019, 2020, 2021, 2022]

algorithms125 = [
    "RF",
    "GB",
    "XGBoost"
]

results125 = []
predictions125 = {}
models125 = {}

for year in years125:

    train = data125.loc[
        data125["Target_Year"] < year
    ].copy()

    val = data125.loc[
        data125["Target_Year"] == year
    ].copy()

    assert train["Target_Year"].max() < year
    assert len(val) == 29

    y_train_delta = (
        train[target125].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    y_val = val[
        target125
    ].to_numpy(dtype=float)

    incidence_val = val[
        "Incidence_t"
    ].to_numpy(dtype=float)

    for algorithm in algorithms125:

        for config, features in feature_sets125.items():

            model = build_model125(algorithm)

            model.fit(
                train[features],
                y_train_delta
            )

            predicted_delta = model.predict(
                val[features]
            )

            prediction = np.maximum(
                incidence_val + predicted_delta,
                0.0
            )

            model_name = (
                f"{algorithm}_{config}"
            )

            results125.append({
                "Year": year,
                "Model": model_name,
                "Algorithm": algorithm,
                "Configuration": config,
                "Train_N": len(train),
                "Val_N": len(val),
                "MAE": mean_absolute_error(
                    y_val,
                    prediction
                ),
                "RMSE": np.sqrt(
                    mean_squared_error(
                        y_val,
                        prediction
                    )
                ),
                "R2": r2_score(
                    y_val,
                    prediction
                )
            })

            predictions125[
                (year, model_name)
            ] = prediction.copy()

            models125[
                (year, model_name)
            ] = model


results125 = pd.DataFrame(results125)


# ------------------------------------------------------------
# 5. Verify RF reproduction against Cell 119
# ------------------------------------------------------------

rf125 = results125.loc[
    results125["Algorithm"] == "RF"
].copy()

name_mapping125 = {
    "ML": "History_ML",
    "BIML_Levels": "History_BIML_Levels",
    "BIML_Changes": "History_BIML_Changes"
}

rf125["Original_Model"] = (
    rf125["Configuration"]
    .map(name_mapping125)
)

reference125 = results119[
    ["Year", "Model", "RMSE"]
].rename(
    columns={
        "Model": "Original_Model",
        "RMSE": "Original_RMSE"
    }
)

reproduction125 = rf125.merge(
    reference125,
    on=["Year", "Original_Model"],
    validate="one_to_one"
)

reproduction125["Difference"] = (
    reproduction125["RMSE"]
    - reproduction125["Original_RMSE"]
).abs()

max_difference125 = (
    reproduction125["Difference"].max()
)

assert max_difference125 < 1e-4, (
    "RF results differ from Cell 119. "
    "Check preprocessing and model settings."
)


# ------------------------------------------------------------
# 6. Add standalone GAM baseline for reporting
# ------------------------------------------------------------

gam125 = results120[
    ["Year", "MAE", "RMSE", "R2"]
].copy()

gam125["Model"] = "GAM"

comparison125 = pd.concat([
    results125[
        ["Year", "Model", "MAE", "RMSE", "R2"]
    ],
    gam125[
        ["Year", "Model", "MAE", "RMSE", "R2"]
    ]
], ignore_index=True)


# ------------------------------------------------------------
# 7. Report performance
# ------------------------------------------------------------

print("=" * 78)
print("CELL 125 — BIML ALGORITHM COMPARISON")
print("=" * 78)

print("\nFEATURE COUNTS")

for name, features in feature_sets125.items():
    print(f"{name}: {len(features)}")


print("\nRF REPRODUCTION CHECK")
print(
    "Maximum absolute RMSE difference:",
    round(float(max_difference125), 8)
)


print("\nROLLING-ORIGIN RMSE")

rmse125 = comparison125.pivot(
    index="Year",
    columns="Model",
    values="RMSE"
)

print(
    rmse125.round(4).to_string()
)


# ------------------------------------------------------------
# 8. Biological contribution within each algorithm
# ------------------------------------------------------------

contribution_rows125 = []

for algorithm in algorithms125:

    for year in years125:

        block = results125.loc[
            (results125["Year"] == year)
            & (results125["Algorithm"] == algorithm)
        ].set_index("Configuration")

        ml_rmse = block.loc["ML", "RMSE"]

        for config in [
            "BIML_Levels",
            "BIML_Changes"
        ]:

            biml_rmse = block.loc[
                config, "RMSE"
            ]

            contribution_rows125.append({
                "Year": year,
                "Algorithm": algorithm,
                "Biological_Set": config,
                "ML_RMSE": ml_rmse,
                "BIML_RMSE": biml_rmse,
                "Biological_Improvement": (
                    ml_rmse - biml_rmse
                )
            })


contribution125 = pd.DataFrame(
    contribution_rows125
)

print("\nBIOLOGICAL CONTRIBUTION BY ALGORITHM")
print("Positive = biological features improve RMSE")

print(
    contribution125.pivot_table(
        index=[
            "Algorithm",
            "Biological_Set"
        ],
        columns="Year",
        values="Biological_Improvement"
    ).round(4).to_string()
)


print("\nYEARS WITH POSITIVE BIOLOGICAL CONTRIBUTION")

print(
    contribution125.groupby(
        ["Algorithm", "Biological_Set"]
    )["Biological_Improvement"]
    .apply(lambda x: int((x > 0).sum()))
    .to_string()
)


print("\nMEAN ANNUAL RMSE")
print(
    comparison125.groupby("Model")["RMSE"]
    .mean()
    .sort_values()
    .round(4)
    .to_string()
)


print("\nCELL 125 COMPLETE")
print("RF, GB and XGBoost evaluated")
print("GAM retained as independent baseline")
print("No GAM-BIML combination")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 126 — GRADIENT BOOSTING BIML DIAGNOSTIC
# BIML 2.0 | BIOLOGICAL FEATURE UTILIZATION
#
# Uses existing Cell 125 fitted models.
# No retraining. No test target access.
# ============================================================

import numpy as np
import pandas as pd

assert "models125" in globals()
assert "predictions125" in globals()
assert "feature_sets125" in globals()
assert "data125" in globals()

years126 = [2019, 2020, 2021, 2022]

configs126 = [
    "ML",
    "BIML_Levels",
    "BIML_Changes"
]

importance_rows126 = []
forecast_rows126 = []

# ------------------------------------------------------------
# 1. Extract Gradient Boosting feature importance
# ------------------------------------------------------------

for year in years126:

    for config in configs126:

        model_name = f"GB_{config}"

        model = models125[(year, model_name)]

        estimator = model.named_steps["regressor"]

        features = feature_sets125[config]

        importance = estimator.feature_importances_

        assert len(features) == len(importance)

        for feature, score in zip(
            features, importance
        ):

            if feature in biological_levels125:
                group = "Biological_Level"

            elif feature in biological_changes125:
                group = "Biological_Change"

            elif feature in environment125:
                group = "Environment"

            elif feature in history125:
                group = "History"

            else:
                group = "Coverage"

            importance_rows126.append({
                "Year": year,
                "Model": model_name,
                "Feature": feature,
                "Group": group,
                "Importance": float(score)
            })

importance126 = pd.DataFrame(
    importance_rows126
)

# ------------------------------------------------------------
# 2. Compare GB and BIML forecasts
# ------------------------------------------------------------

for year in years126:

    val = data125.loc[
        data125["Target_Year"] == year
    ].copy()

    actual = val[
        "Target_Incidence_t_plus_1"
    ].to_numpy(dtype=float)

    ml_prediction = np.asarray(
        predictions125[(year, "GB_ML")],
        dtype=float
    )

    for config in [
        "BIML_Levels",
        "BIML_Changes"
    ]:

        biml_prediction = np.asarray(
            predictions125[
                (year, f"GB_{config}")
            ],
            dtype=float
        )

        for i in range(len(val)):

            forecast_rows126.append({
                "Year": year,
                "Country": val.iloc[i]["Country"],
                "Configuration": config,
                "Actual": actual[i],
                "ML_Prediction": ml_prediction[i],
                "BIML_Prediction": biml_prediction[i],
                "Forecast_Shift": (
                    biml_prediction[i]
                    - ml_prediction[i]
                ),
                "ML_Absolute_Error": abs(
                    actual[i] - ml_prediction[i]
                ),
                "BIML_Absolute_Error": abs(
                    actual[i] - biml_prediction[i]
                )
            })

forecasts126 = pd.DataFrame(
    forecast_rows126
)

forecasts126["Error_Improvement"] = (
    forecasts126["ML_Absolute_Error"]
    - forecasts126["BIML_Absolute_Error"]
)

# ------------------------------------------------------------
# 3. Importance by feature group
# ------------------------------------------------------------

group_importance126 = (
    importance126.groupby(
        ["Year", "Model", "Group"]
    )["Importance"]
    .sum()
    .reset_index()
)

print("=" * 78)
print("CELL 126 — GRADIENT BOOSTING BIML DIAGNOSTIC")
print("=" * 78)

print("\nFEATURE-GROUP IMPORTANCE")

print(
    group_importance126.pivot_table(
        index=["Year", "Model"],
        columns="Group",
        values="Importance",
        fill_value=0
    ).round(4).to_string()
)

# ------------------------------------------------------------
# 4. Biological feature importance
# ------------------------------------------------------------

biological_importance126 = importance126.loc[
    importance126["Group"].isin([
        "Biological_Level",
        "Biological_Change"
    ])
].copy()

print("\nINDIVIDUAL BIOLOGICAL FEATURE IMPORTANCE")

print(
    biological_importance126.pivot_table(
        index="Feature",
        columns="Year",
        values="Importance",
        aggfunc="mean"
    ).round(4).to_string()
)

# ------------------------------------------------------------
# 5. Forecast shift and error improvement
# ------------------------------------------------------------

summary126 = (
    forecasts126.groupby(
        ["Year", "Configuration"]
    )
    .agg(
        Mean_Absolute_Forecast_Shift=(
            "Forecast_Shift",
            lambda x: np.mean(np.abs(x))
        ),
        Mean_Error_Improvement=(
            "Error_Improvement",
            "mean"
        ),
        Countries_Improved=(
            "Error_Improvement",
            lambda x: int((x > 0).sum())
        ),
        Countries_Worsened=(
            "Error_Improvement",
            lambda x: int((x < 0).sum())
        )
    )
    .reset_index()
)

print("\nFORECAST EFFECT OF BIOLOGICAL FEATURES")

print(
    summary126.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# 6. Largest changes in GB+B2 Levels predictions
# ------------------------------------------------------------

levels126 = forecasts126.loc[
    forecasts126["Configuration"] == "BIML_Levels"
].copy()

levels126["Absolute_Shift"] = (
    levels126["Forecast_Shift"].abs()
)

print("\nLARGEST BIOLOGY-INDUCED FORECAST SHIFTS")

print(
    levels126.sort_values(
        "Absolute_Shift",
        ascending=False
    )[
        [
            "Year",
            "Country",
            "Actual",
            "ML_Prediction",
            "BIML_Prediction",
            "Forecast_Shift",
            "Error_Improvement"
        ]
    ].head(12).round(4).to_string(index=False)
)

print("\nCELL 126 COMPLETE")
print("Gradient Boosting biological utilization examined")
print("No models retrained")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 127 — GRADIENT BOOSTING REGULARIZATION
# BIML 2.0 | MATCHED ML VS BIML
#
# GAM remains an independent baseline.
# No GAM-BIML combination.
# Development years only: 2016–2022.
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "data125" in globals()
assert "feature_sets125" in globals()
assert "results125" in globals()
assert "results120" in globals()

data127 = data125.loc[
    data125["Target_Year"] <= 2022
].copy()

target127 = "Target_Incidence_t_plus_1"

# ------------------------------------------------------------
# 1. Prespecified configurations
# ------------------------------------------------------------

settings127 = {
    "GB_Original": {
        "n_estimators": 150,
        "learning_rate": 0.03,
        "max_depth": 2,
        "min_samples_leaf": 5,
        "subsample": 0.8
    },

    "GB_Regularized": {
        "n_estimators": 100,
        "learning_rate": 0.03,
        "max_depth": 2,
        "min_samples_leaf": 10,
        "subsample": 0.8
    },

    "GB_Strongly_Regularized": {
        "n_estimators": 100,
        "learning_rate": 0.02,
        "max_depth": 1,
        "min_samples_leaf": 15,
        "subsample": 0.8
    }
}

configs127 = [
    "ML",
    "BIML_Levels"
]

years127 = [2019, 2020, 2021, 2022]

results127 = []
predictions127 = {}

# ------------------------------------------------------------
# 2. Rolling-origin evaluation
# ------------------------------------------------------------

for year in years127:

    train = data127.loc[
        data127["Target_Year"] < year
    ].copy()

    val = data127.loc[
        data127["Target_Year"] == year
    ].copy()

    assert train["Target_Year"].max() < year
    assert len(val) == 29

    y_train_delta = (
        train[target127].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    y_val = val[
        target127
    ].to_numpy(dtype=float)

    incidence_val = val[
        "Incidence_t"
    ].to_numpy(dtype=float)

    for setting_name, params in settings127.items():

        for config in configs127:

            features = feature_sets125[config]

            model = Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="median")
                ),
                (
                    "regressor",
                    GradientBoostingRegressor(
                        **params,
                        random_state=20261016
                    )
                )
            ])

            model.fit(
                train[features],
                y_train_delta
            )

            predicted_delta = model.predict(
                val[features]
            )

            prediction = np.maximum(
                incidence_val + predicted_delta,
                0
            )

            results127.append({
                "Year": year,
                "Setting": setting_name,
                "Configuration": config,
                "Train_N": len(train),
                "Val_N": len(val),
                "MAE": mean_absolute_error(
                    y_val,
                    prediction
                ),
                "RMSE": np.sqrt(
                    mean_squared_error(
                        y_val,
                        prediction
                    )
                ),
                "R2": r2_score(
                    y_val,
                    prediction
                )
            })

            predictions127[
                (year, setting_name, config)
            ] = prediction.copy()

results127 = pd.DataFrame(results127)

# ------------------------------------------------------------
# 3. Verify original GB reproduction
# ------------------------------------------------------------

original127 = results127.loc[
    results127["Setting"] == "GB_Original"
].copy()

reference127 = results125.loc[
    results125["Algorithm"] == "GB",
    ["Year", "Configuration", "RMSE"]
].rename(
    columns={"RMSE": "Reference_RMSE"}
)

check127 = original127.merge(
    reference127,
    on=["Year", "Configuration"],
    validate="one_to_one"
)

max_difference127 = np.max(
    np.abs(
        check127["RMSE"]
        - check127["Reference_RMSE"]
    )
)

assert max_difference127 < 1e-8, (
    "Original GB does not reproduce Cell 125"
)

# ------------------------------------------------------------
# 4. Compare biological contribution
# ------------------------------------------------------------

rmse127 = results127.pivot_table(
    index=["Year", "Setting"],
    columns="Configuration",
    values="RMSE"
)

rmse127["Biological_Improvement"] = (
    rmse127["ML"]
    - rmse127["BIML_Levels"]
)

print("=" * 78)
print("CELL 127 — GB REGULARIZATION SENSITIVITY")
print("=" * 78)

print("\nORIGINAL GB REPRODUCTION")
print(
    "Maximum RMSE difference:",
    round(float(max_difference127), 10)
)

print("\nROLLING-ORIGIN RMSE")
print(
    rmse127.round(4).to_string()
)

print("\nBIOLOGICAL CONTRIBUTION")
print("Positive = B2 improves RMSE")

print(
    rmse127["Biological_Improvement"]
    .unstack("Setting")
    .round(4)
    .to_string()
)

print("\nMEAN ANNUAL RMSE BY CONFIGURATION")

summary127 = (
    results127.groupby(
        ["Setting", "Configuration"]
    )["RMSE"]
    .mean()
    .sort_values()
)

print(
    summary127.round(4).to_string()
)

print("\nYEARS WITH POSITIVE BIOLOGICAL CONTRIBUTION")

positive127 = (
    rmse127["Biological_Improvement"]
    .groupby("Setting")
    .apply(lambda x: int((x > 0).sum()))
)

print(
    positive127.to_string()
)

print("\nCELL 127 COMPLETE")
print("GB regularization sensitivity evaluated")
print("GAM unchanged as standalone baseline")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 128 — BIOLOGICAL FEATURE REDUNDANCY AUDIT
# BIML 2.0 | B2 VS ENVIRONMENTAL INFORMATION
#
# No model training
# No GAM-BIML combination
# Development data only (Target_Year <= 2022)
# ============================================================

import numpy as np
import pandas as pd

assert "data125" in globals()
assert "feature_groups" in globals()

data128 = data125.loc[
    data125["Target_Year"] <= 2022
].copy()

bio128 = list(feature_groups["B2"])

environment128 = list(feature_groups["E"])

temperature128 = [
    "ERA5_T2M_C_mean",
    "MODIS_Night_LST_C_mean",
    "Landsat_LST_C_mean"
]

assert set(
    bio128 + environment128
).issubset(data128.columns)

# ------------------------------------------------------------
# 1. Biological vs environmental correlations
# ------------------------------------------------------------

corr128 = data128[
    bio128 + environment128
].corr(method="spearman")

bio_env128 = corr128.loc[
    bio128,
    environment128
]

print("=" * 78)
print("CELL 128 — BIOLOGICAL FEATURE REDUNDANCY")
print("=" * 78)

print("\nB2 VS ENVIRONMENTAL SPEARMAN CORRELATIONS")
print(
    bio_env128.round(3).to_string()
)

# ------------------------------------------------------------
# 2. Strongest environmental association per B2 feature
# ------------------------------------------------------------

strongest128 = []

for feature in bio128:

    row = bio_env128.loc[feature].dropna()

    strongest_feature = row.abs().idxmax()

    strongest128.append({
        "Biological_Feature": feature,
        "Strongest_Environmental_Feature": strongest_feature,
        "Spearman_Rho": row[strongest_feature],
        "Absolute_Rho": abs(row[strongest_feature])
    })

strongest128 = pd.DataFrame(
    strongest128
)

print("\nSTRONGEST ENVIRONMENTAL ASSOCIATION")

print(
    strongest128.sort_values(
        "Absolute_Rho",
        ascending=False
    ).round(4).to_string(index=False)
)

# ------------------------------------------------------------
# 3. Correlations among B2 biological features
# ------------------------------------------------------------

print("\nWITHIN-B2 SPEARMAN CORRELATIONS")

print(
    data128[bio128]
    .corr(method="spearman")
    .round(3)
    .to_string()
)

# ------------------------------------------------------------
# 4. Between-country / within-country variance
# ------------------------------------------------------------

variance_rows128 = []

for feature in bio128:

    valid = data128[
        ["Country", feature]
    ].dropna()

    grand_mean = valid[feature].mean()

    group_stats = valid.groupby("Country")[feature].agg(
        ["mean", "count"]
    )

    between_ss = np.sum(
        group_stats["count"]
        * (
            group_stats["mean"] - grand_mean
        ) ** 2
    )

    within_ss = valid.groupby("Country")[feature].apply(
        lambda x: np.sum(
            (x - x.mean()) ** 2
        )
    ).sum()

    total_ss = between_ss + within_ss

    variance_rows128.append({
        "Feature": feature,
        "Between_Country_Fraction": (
            between_ss / total_ss
            if total_ss > 0 else np.nan
        ),
        "Within_Country_Fraction": (
            within_ss / total_ss
            if total_ss > 0 else np.nan
        )
    })

variance128 = pd.DataFrame(
    variance_rows128
)

print("\nBETWEEN-COUNTRY VS WITHIN-COUNTRY VARIATION")

print(
    variance128.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# 5. Within-country correlations with temperature
# ------------------------------------------------------------

within128 = data128[
    ["Country"] + bio128 + temperature128
].copy()

for feature in bio128 + temperature128:

    within128[feature] = (
        within128[feature]
        - within128.groupby("Country")[feature]
        .transform("mean")
    )

within_corr128 = within128[
    bio128 + temperature128
].corr(method="spearman")

print("\nWITHIN-COUNTRY B2 VS TEMPERATURE CORRELATIONS")

print(
    within_corr128.loc[
        bio128,
        temperature128
    ].round(3).to_string()
)

print("\nCELL 128 COMPLETE")
print("Biological redundancy audit finished")
print("No models trained")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 129 — B2 BIOLOGICAL FEATURE ABLATION
# BIML 2.0 | STRONGLY REGULARIZED GRADIENT BOOSTING
#
# Original BIML architecture preserved.
# GAM remains independent.
# Development data only, through 2022.
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error
)

assert "data125" in globals()
assert "feature_sets125" in globals()
assert "biological_levels125" in globals()
assert "settings127" in globals()
assert "results127" in globals()

data129 = data125.loc[
    data125["Target_Year"] <= 2022
].copy()

target129 = "Target_Incidence_t_plus_1"

base129 = list(feature_sets125["ML"])

bio129 = list(biological_levels125)

# ------------------------------------------------------------
# 1. Prespecified biological feature sets
# ------------------------------------------------------------

ablations129 = {
    "ML_No_Biology": [],

    "BIML_Competence": [
        "B2_competence_supported_mean"
    ],

    "BIML_EIP": [
        "B2_EIP_supported_mean"
    ],

    "BIML_Biting": [
        "B2_biting_supported_mean"
    ],

    "BIML_Index": [
        "B2_index_supported_mean"
    ],

    "BIML_Competence_Index": [
        "B2_competence_supported_mean",
        "B2_index_supported_mean"
    ],

    "BIML_All_B2": bio129
}

# ------------------------------------------------------------
# 2. Use fixed GB hyperparameters from Cell 127
# ------------------------------------------------------------

params129 = settings127[
    "GB_Strongly_Regularized"
].copy()

years129 = [2019, 2020, 2021, 2022]

results129 = []
predictions129 = {}
models129 = {}

# ------------------------------------------------------------
# 3. Rolling-origin ablation
# ------------------------------------------------------------

for year in years129:

    train = data129.loc[
        data129["Target_Year"] < year
    ].copy()

    val = data129.loc[
        data129["Target_Year"] == year
    ].copy()

    assert train["Target_Year"].max() < year
    assert len(val) == 29

    y_train = (
        train[target129].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    actual = val[
        target129
    ].to_numpy(dtype=float)

    incidence_val = val[
        "Incidence_t"
    ].to_numpy(dtype=float)

    for name, biological_features in ablations129.items():

        features = base129 + biological_features

        model = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "regressor",
                GradientBoostingRegressor(
                    **params129,
                    random_state=20261016
                )
            )
        ])

        model.fit(
            train[features],
            y_train
        )

        predicted_change = model.predict(
            val[features]
        )

        prediction = np.maximum(
            incidence_val + predicted_change,
            0.0
        )

        rmse = np.sqrt(
            mean_squared_error(
                actual,
                prediction
            )
        )

        mae = mean_absolute_error(
            actual,
            prediction
        )

        results129.append({
            "Year": year,
            "Model": name,
            "N_Features": len(features),
            "RMSE": rmse,
            "MAE": mae
        })

        predictions129[
            (year, name)
        ] = prediction.copy()

        models129[
            (year, name)
        ] = model


results129 = pd.DataFrame(results129)

# ------------------------------------------------------------
# 4. Verify baseline and full B2 reproduction
# ------------------------------------------------------------

reference129 = results127.loc[
    results127["Setting"]
    == "GB_Strongly_Regularized"
].copy()

mapping129 = {
    "ML_No_Biology": "ML",
    "BIML_All_B2": "BIML_Levels"
}

check129 = results129.loc[
    results129["Model"].isin(mapping129)
].copy()

check129["Configuration"] = (
    check129["Model"].map(mapping129)
)

check129 = check129.merge(
    reference129[
        ["Year", "Configuration", "RMSE"]
    ].rename(
        columns={"RMSE": "Reference_RMSE"}
    ),
    on=["Year", "Configuration"],
    validate="one_to_one"
)

max_diff129 = np.max(
    np.abs(
        check129["RMSE"]
        - check129["Reference_RMSE"]
    )
)

assert max_diff129 < 1e-8, (
    "Ablation reference models do not "
    "reproduce Cell 127."
)

# ------------------------------------------------------------
# 5. Calculate biological contribution
# ------------------------------------------------------------

rmse129 = results129.pivot(
    index="Year",
    columns="Model",
    values="RMSE"
)

baseline129 = rmse129["ML_No_Biology"]

contribution129 = (
    rmse129
    .rsub(baseline129, axis=0)
)

# ------------------------------------------------------------
# 6. Report
# ------------------------------------------------------------

print("=" * 78)
print("CELL 129 — B2 BIOLOGICAL FEATURE ABLATION")
print("=" * 78)

print("\nREPRODUCTION CHECK")
print(
    "Maximum RMSE difference:",
    round(float(max_diff129), 10)
)

print("\nROLLING-ORIGIN RMSE")
print(
    rmse129.round(4).to_string()
)

print("\nBIOLOGICAL CONTRIBUTION")
print("Positive = improvement over conventional GB")

print(
    contribution129.drop(
        columns="ML_No_Biology"
    ).round(4).to_string()
)

print("\nMEAN ANNUAL RMSE")

mean_rmse129 = (
    results129.groupby("Model")["RMSE"]
    .mean()
    .sort_values()
)

print(
    mean_rmse129.round(4).to_string()
)

print("\nYEARS WITH POSITIVE CONTRIBUTION")

positive129 = (
    contribution129.drop(
        columns="ML_No_Biology"
    ) > 0
).sum()

print(
    positive129.sort_values(
        ascending=False
    ).to_string()
)

print("\nCELL 129 COMPLETE")
print("Biological feature ablation completed")
print("Original BIML architecture preserved")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 130 — COUNTRY-LEVEL BIML IMPROVEMENT AUDIT
# BIML 2.0 | STRONGLY REGULARIZED GB
#
# No new training.
# No GAM-BIML combination.
# Development years only (2019–2022).
# ============================================================

import numpy as np
import pandas as pd

assert "data129" in globals()
assert "predictions129" in globals()
assert "results129" in globals()

years130 = [2019, 2020, 2021, 2022]

configs130 = [
    "BIML_Competence",
    "BIML_Competence_Index",
    "BIML_All_B2"
]

rows130 = []

# ------------------------------------------------------------
# 1. Reconstruct paired country-year predictions
# ------------------------------------------------------------

for year in years130:

    val = data129.loc[
        data129["Target_Year"] == year
    ].copy()

    actual = val[
        "Target_Incidence_t_plus_1"
    ].to_numpy(dtype=float)

    ml_prediction = np.asarray(
        predictions129[
            (year, "ML_No_Biology")
        ],
        dtype=float
    )

    assert len(actual) == len(ml_prediction)

    for config in configs130:

        biml_prediction = np.asarray(
            predictions129[(year, config)],
            dtype=float
        )

        assert len(actual) == len(biml_prediction)

        for i in range(len(val)):

            ml_error = (
                actual[i] - ml_prediction[i]
            )

            biml_error = (
                actual[i] - biml_prediction[i]
            )

            rows130.append({
                "Year": year,
                "Country": val.iloc[i]["Country"],
                "Configuration": config,
                "Actual": actual[i],
                "ML_Prediction": ml_prediction[i],
                "BIML_Prediction": biml_prediction[i],
                "ML_AE": abs(ml_error),
                "BIML_AE": abs(biml_error),
                "ML_SE": ml_error ** 2,
                "BIML_SE": biml_error ** 2,
                "AE_Improvement": (
                    abs(ml_error) - abs(biml_error)
                ),
                "SE_Improvement": (
                    ml_error ** 2 - biml_error ** 2
                )
            })

paired130 = pd.DataFrame(rows130)

# ------------------------------------------------------------
# 2. Annual improvement distribution
# ------------------------------------------------------------

annual130 = (
    paired130.groupby(
        ["Year", "Configuration"]
    )
    .agg(
        Countries=("Country", "nunique"),
        Countries_Improved=(
            "AE_Improvement",
            lambda x: int((x > 0).sum())
        ),
        Countries_Worsened=(
            "AE_Improvement",
            lambda x: int((x < 0).sum())
        ),
        Mean_AE_Improvement=(
            "AE_Improvement",
            "mean"
        ),
        Median_AE_Improvement=(
            "AE_Improvement",
            "median"
        ),
        Mean_SE_Improvement=(
            "SE_Improvement",
            "mean"
        )
    )
    .reset_index()
)

print("=" * 78)
print("CELL 130 — COUNTRY-LEVEL BIML AUDIT")
print("=" * 78)

print("\nANNUAL COUNTRY-LEVEL CONTRIBUTION")

print(
    annual130.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# 3. Overall country-level contribution
# ------------------------------------------------------------

country130 = (
    paired130.groupby(
        ["Configuration", "Country"]
    )
    .agg(
        Mean_AE_Improvement=(
            "AE_Improvement",
            "mean"
        ),
        Total_SE_Improvement=(
            "SE_Improvement",
            "sum"
        )
    )
    .reset_index()
)

print("\nCOUNTRIES WITH POSITIVE MEAN AE CONTRIBUTION")

print(
    country130.groupby("Configuration")[
        "Mean_AE_Improvement"
    ]
    .apply(lambda x: int((x > 0).sum()))
    .to_string()
)

# ------------------------------------------------------------
# 4. Largest positive and negative contributions
# ------------------------------------------------------------

for config in configs130:

    subset = country130.loc[
        country130["Configuration"] == config
    ].copy()

    print("\n" + "-" * 65)
    print(config)
    print("-" * 65)

    print("\nTOP 5 COUNTRIES BENEFITING FROM B2")

    print(
        subset.nlargest(
            5,
            "Total_SE_Improvement"
        )[
            [
                "Country",
                "Mean_AE_Improvement",
                "Total_SE_Improvement"
            ]
        ].round(4).to_string(index=False)
    )

    print("\nTOP 5 COUNTRIES HARMED BY B2")

    print(
        subset.nsmallest(
            5,
            "Total_SE_Improvement"
        )[
            [
                "Country",
                "Mean_AE_Improvement",
                "Total_SE_Improvement"
            ]
        ].round(4).to_string(index=False)
    )

# ------------------------------------------------------------
# 5. Influence of largest country contributions
# ------------------------------------------------------------

influence130 = []

for config in configs130:

    subset = country130.loc[
        country130["Configuration"] == config
    ].copy()

    total_gain = subset[
        "Total_SE_Improvement"
    ].sum()

    positive_gains = subset.loc[
        subset["Total_SE_Improvement"] > 0,
        "Total_SE_Improvement"
    ].sort_values(ascending=False)

    total_positive = positive_gains.sum()

    top3_share = (
        positive_gains.head(3).sum()
        / total_positive
        if total_positive > 0
        else np.nan
    )

    influence130.append({
        "Configuration": config,
        "Net_SE_Improvement": total_gain,
        "Top3_Share_of_Positive_Gains": top3_share,
        "Countries_With_Positive_SE_Gain": (
            subset["Total_SE_Improvement"] > 0
        ).sum()
    })

influence130 = pd.DataFrame(influence130)

print("\nINFLUENCE DIAGNOSTIC")

print(
    influence130.round(4).to_string(index=False)
)

print("\nCELL 130 COMPLETE")
print("Country-level biological contributions audited")
print("No models retrained")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 131 — COUNTRY-INFLUENCE SENSITIVITY
# BIML | EXISTING GRADIENT BOOSTING PREDICTIONS
#
# No model retraining
# No GAM-BIML combination
# Development years 2019–2022 only
# ============================================================

import numpy as np
import pandas as pd

assert "paired130" in globals()

configs131 = [
    "BIML_Competence",
    "BIML_Competence_Index",
    "BIML_All_B2"
]

rows131 = []

for config in configs131:

    block = paired130.loc[
        paired130["Configuration"] == config
    ].copy()

    countries = sorted(block["Country"].unique())

    # Include full sample as reference
    for excluded in [None] + countries:

        subset = (
            block if excluded is None
            else block.loc[block["Country"] != excluded]
        )

        # Calculate annual RMSE first, then average years
        annual = subset.groupby("Year").agg(
            ML_MSE=("ML_SE", "mean"),
            BIML_MSE=("BIML_SE", "mean")
        )

        annual["ML_RMSE"] = np.sqrt(
            annual["ML_MSE"]
        )

        annual["BIML_RMSE"] = np.sqrt(
            annual["BIML_MSE"]
        )

        improvement = (
            annual["ML_RMSE"]
            - annual["BIML_RMSE"]
        )

        rows131.append({
            "Configuration": config,
            "Excluded_Country": (
                "NONE" if excluded is None else excluded
            ),
            "Mean_ML_RMSE": annual["ML_RMSE"].mean(),
            "Mean_BIML_RMSE": annual["BIML_RMSE"].mean(),
            "RMSE_Improvement": improvement.mean(),
            "Years_Improved": int(
                (improvement > 0).sum()
            )
        })

sensitivity131 = pd.DataFrame(rows131)

# ------------------------------------------------------------
# Full-sample reference
# ------------------------------------------------------------

reference131 = sensitivity131.loc[
    sensitivity131["Excluded_Country"] == "NONE"
]

print("=" * 72)
print("CELL 131 — COUNTRY-INFLUENCE SENSITIVITY")
print("=" * 72)

print("\nFULL SAMPLE")
print(
    reference131.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# Excluding Rwanda
# ------------------------------------------------------------

print("\nEXCLUDING RWANDA")
print(
    sensitivity131.loc[
        sensitivity131["Excluded_Country"] == "Rwanda"
    ].round(4).to_string(index=False)
)

# ------------------------------------------------------------
# All single-country exclusions
# ------------------------------------------------------------

excluded131 = sensitivity131.loc[
    sensitivity131["Excluded_Country"] != "NONE"
].copy()

summary131 = excluded131.groupby(
    "Configuration"
).agg(
    Minimum_Improvement=("RMSE_Improvement", "min"),
    Median_Improvement=("RMSE_Improvement", "median"),
    Maximum_Improvement=("RMSE_Improvement", "max"),
    Exclusions_With_Positive_Improvement=(
        "RMSE_Improvement",
        lambda x: int((x > 0).sum())
    ),
    Total_Exclusions=("RMSE_Improvement", "size")
)

print("\nSENSITIVITY SUMMARY")
print(
    summary131.round(4).to_string()
)

print("\nCOUNTRY EXCLUSIONS WITH LARGEST IMPACT")

for config in configs131:

    block = excluded131.loc[
        excluded131["Configuration"] == config
    ].copy()

    baseline_gain = float(
        reference131.loc[
            reference131["Configuration"] == config,
            "RMSE_Improvement"
        ].iloc[0]
    )

    block["Change_From_Full_Sample"] = (
        block["RMSE_Improvement"]
        - baseline_gain
    )

    block["Absolute_Impact"] = (
        block["Change_From_Full_Sample"].abs()
    )

    print(f"\n{config}")

    print(
        block.nlargest(
            5, "Absolute_Impact"
        )[
            [
                "Excluded_Country",
                "RMSE_Improvement",
                "Change_From_Full_Sample"
            ]
        ].round(4).to_string(index=False)
    )

print("\nCELL 131 COMPLETE")
print("No retraining performed")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 132 — GEOGRAPHICAL HETEROGENEITY OF BIML
# BIML | REGIONAL AND ENVIRONMENTAL ASSOCIATIONS
#
# Uses existing Cell 129 predictions
# No model retraining
# Development years 2019–2022 only
# ============================================================

import numpy as np
import pandas as pd

from scipy.stats import spearmanr

assert "data129" in globals()
assert "predictions129" in globals()

years132 = [2019, 2020, 2021, 2022]

configs132 = {
    "BIML_Reduced": "BIML_Competence_Index",
    "BIML_Full": "BIML_All_B2"
}

context132 = [
    "Incidence_t",
    "ERA5_T2M_C_mean",
    "Rainfall_total_mm",
    "RH_pct_mean",
    "NDVI_mean"
]

assert set(context132).issubset(data129.columns)

rows132 = []

# ------------------------------------------------------------
# 1. Reconstruct paired forecasts and context
# ------------------------------------------------------------

for year in years132:

    val = data129.loc[
        data129["Target_Year"] == year
    ].copy()

    actual = val[
        "Target_Incidence_t_plus_1"
    ].to_numpy(dtype=float)

    ml_pred = np.asarray(
        predictions129[(year, "ML_No_Biology")],
        dtype=float
    )

    for label, config in configs132.items():

        biml_pred = np.asarray(
            predictions129[(year, config)],
            dtype=float
        )

        for i in range(len(val)):

            record = {
                "Year": year,
                "Country": val.iloc[i]["Country"],
                "Configuration": label,
                "Actual": actual[i],
                "ML_Prediction": ml_pred[i],
                "BIML_Prediction": biml_pred[i],
                "AE_Improvement": (
                    abs(actual[i] - ml_pred[i])
                    - abs(actual[i] - biml_pred[i])
                ),
                "SE_Improvement": (
                    (actual[i] - ml_pred[i]) ** 2
                    - (actual[i] - biml_pred[i]) ** 2
                )
            }

            for feature in context132:
                record[feature] = val.iloc[i][feature]

            rows132.append(record)

heterogeneity132 = pd.DataFrame(rows132)

assert (
    heterogeneity132.groupby(
        ["Year", "Configuration"]
    )["Country"].nunique() == 29
).all()

# ------------------------------------------------------------
# 2. Country-level performance and context
# ------------------------------------------------------------

country132 = (
    heterogeneity132.groupby(
        ["Configuration", "Country"]
    )
    .agg(
        Mean_AE_Improvement=(
            "AE_Improvement", "mean"
        ),
        Total_SE_Improvement=(
            "SE_Improvement", "sum"
        ),
        Mean_Incidence=(
            "Incidence_t", "mean"
        ),
        Mean_Temperature=(
            "ERA5_T2M_C_mean", "mean"
        ),
        Mean_Rainfall=(
            "Rainfall_total_mm", "mean"
        ),
        Mean_Humidity=(
            "RH_pct_mean", "mean"
        ),
        Mean_NDVI=(
            "NDVI_mean", "mean"
        )
    )
    .reset_index()
)

# ------------------------------------------------------------
# 3. Between-country associations
# ------------------------------------------------------------

association_features132 = {
    "Mean_Incidence": "Historical incidence",
    "Mean_Temperature": "Temperature",
    "Mean_Rainfall": "Rainfall",
    "Mean_Humidity": "Relative humidity",
    "Mean_NDVI": "NDVI"
}

def safe_spearman132(x, y):

    pair = pd.DataFrame({
        "x": x,
        "y": y
    }).replace(
        [np.inf, -np.inf], np.nan
    ).dropna()

    if (
        len(pair) < 5
        or pair["x"].nunique() < 2
        or pair["y"].nunique() < 2
    ):
        return np.nan, np.nan, len(pair)

    rho, p = spearmanr(
        pair["x"],
        pair["y"]
    )

    return float(rho), float(p), len(pair)


between_rows132 = []

for config in configs132:

    subset = country132.loc[
        country132["Configuration"] == config
    ]

    for feature, label in association_features132.items():

        rho, p, n = safe_spearman132(
            subset[feature],
            subset["Mean_AE_Improvement"]
        )

        between_rows132.append({
            "Configuration": config,
            "Context": label,
            "N_Countries": n,
            "Spearman_Rho": rho,
            "Exploratory_P": p
        })

between132 = pd.DataFrame(between_rows132)

# ------------------------------------------------------------
# 4. Within-country associations
# ------------------------------------------------------------

within_rows132 = []

for config in configs132:

    subset = heterogeneity132.loc[
        heterogeneity132["Configuration"] == config
    ].copy()

    # Remove each country's average to study
    # temporal departures within countries.
    subset["Improvement_Within"] = (
        subset["AE_Improvement"]
        - subset.groupby("Country")[
            "AE_Improvement"
        ].transform("mean")
    )

    for feature, label in {
        "Incidence_t": "Historical incidence",
        "ERA5_T2M_C_mean": "Temperature",
        "Rainfall_total_mm": "Rainfall",
        "RH_pct_mean": "Relative humidity",
        "NDVI_mean": "NDVI"
    }.items():

        x_within = (
            subset[feature]
            - subset.groupby("Country")[
                feature
            ].transform("mean")
        )

        rho, p, n = safe_spearman132(
            x_within,
            subset["Improvement_Within"]
        )

        within_rows132.append({
            "Configuration": config,
            "Context": label,
            "N_Country_Years": n,
            "Spearman_Rho": rho,
            "Exploratory_P": p
        })

within132 = pd.DataFrame(within_rows132)

# ------------------------------------------------------------
# 5. Incidence-stratified descriptive analysis
# ------------------------------------------------------------

strata_rows132 = []

for config in configs132:

    subset = heterogeneity132.loc[
        heterogeneity132["Configuration"] == config
    ].copy()

    # Fixed country groups based on their
    # development-period mean historical incidence.
    country_incidence = (
        subset.groupby("Country")["Incidence_t"]
        .mean()
    )

    groups = pd.qcut(
        country_incidence.rank(method="first"),
        q=3,
        labels=[
            "Lower historical incidence",
            "Middle historical incidence",
            "Higher historical incidence"
        ]
    )

    subset["Incidence_Group"] = (
        subset["Country"].map(groups)
    )

    for group, block in subset.groupby(
        "Incidence_Group",
        observed=True
    ):

        strata_rows132.append({
            "Configuration": config,
            "Incidence_Group": str(group),
            "Countries": block["Country"].nunique(),
            "Country_Years": len(block),
            "Mean_AE_Improvement": (
                block["AE_Improvement"].mean()
            ),
            "Median_AE_Improvement": (
                block["AE_Improvement"].median()
            ),
            "Mean_SE_Improvement": (
                block["SE_Improvement"].mean()
            ),
            "Share_Improved": (
                (block["AE_Improvement"] > 0).mean()
            )
        })

strata132 = pd.DataFrame(strata_rows132)

# ------------------------------------------------------------
# 6. Results
# ------------------------------------------------------------

print("=" * 76)
print("CELL 132 — GEOGRAPHICAL HETEROGENEITY OF BIML")
print("=" * 76)

print("\nBETWEEN-COUNTRY ASSOCIATIONS")
print("Spearman rho vs mean AE improvement")

print(
    between132.round(4).to_string(index=False)
)

print("\nWITHIN-COUNTRY ASSOCIATIONS")
print("Country-demeaned, 2019–2022")

print(
    within132.round(4).to_string(index=False)
)

print("\nHISTORICAL INCIDENCE STRATA")

print(
    strata132.round(4).to_string(index=False)
)

print("\nCELL 132 COMPLETE")
print("Exploratory heterogeneity assessment finished")
print("P-values are descriptive, not confirmatory")
print("No models retrained")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 133 — ROLLING-ORIGIN PERFORMANCE AND UNCERTAINTY
# BIML 2.0 | DEVELOPMENT PERIOD: 2019–2022
#
# Models:
#   1. Conventional ML (GB)
#   2. BIML-Reduced (Competence + Index)
#   3. BIML-Full (All B2)
#   4. Independent GAM baseline
#
# Metrics: RMSE, MAE, R2
# Country-clustered bootstrap for paired comparisons
# No retraining / No test-target access
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

assert "data129" in globals()
assert "predictions129" in globals()
assert "results120" in globals()

years133 = [2019, 2020, 2021, 2022]

model_keys133 = {
    "ML": "ML_No_Biology",
    "BIML_Reduced": "BIML_Competence_Index",
    "BIML_Full": "BIML_All_B2"
}

# ------------------------------------------------------------
# 1. Reconstruct existing paired predictions
# ------------------------------------------------------------

records133 = []

for year in years133:

    val = data129.loc[
        data129["Target_Year"] == year
    ].copy()

    actual = val[
        "Target_Incidence_t_plus_1"
    ].to_numpy(dtype=float)

    preds = {}

    for name, key in model_keys133.items():

        preds[name] = np.asarray(
            predictions129[(year, key)],
            dtype=float
        )

        assert len(preds[name]) == len(actual)

    for i in range(len(val)):

        record = {
            "Year": year,
            "Country": val.iloc[i]["Country"],
            "Actual": actual[i]
        }

        for name in model_keys133:
            record[name] = preds[name][i]

        records133.append(record)

evaluation133 = pd.DataFrame(records133)

assert len(evaluation133) == 116
assert evaluation133[["Year", "Country"]].duplicated().sum() == 0

# ------------------------------------------------------------
# 2. Metric function
# ------------------------------------------------------------

def metrics133(y, pred):

    y = np.asarray(y, dtype=float)
    pred = np.asarray(pred, dtype=float)

    return {
        "RMSE": np.sqrt(mean_squared_error(y, pred)),
        "MAE": mean_absolute_error(y, pred),
        "R2": r2_score(y, pred)
    }

# ------------------------------------------------------------
# 3. Annual model performance
# ------------------------------------------------------------

annual_rows133 = []

for year, block in evaluation133.groupby("Year"):

    for model in model_keys133:

        m = metrics133(
            block["Actual"],
            block[model]
        )

        annual_rows133.append({
            "Year": year,
            "Model": model,
            **m
        })

annual133 = pd.DataFrame(annual_rows133)

print("=" * 78)
print("CELL 133 — ROLLING-ORIGIN PERFORMANCE")
print("=" * 78)

print("\nANNUAL RMSE")
print(
    annual133.pivot(
        index="Year",
        columns="Model",
        values="RMSE"
    ).round(4).to_string()
)

print("\nANNUAL MAE")
print(
    annual133.pivot(
        index="Year",
        columns="Model",
        values="MAE"
    ).round(4).to_string()
)

print("\nANNUAL R2")
print(
    annual133.pivot(
        index="Year",
        columns="Model",
        values="R2"
    ).round(4).to_string()
)

# ------------------------------------------------------------
# 4. Mean annual and pooled performance
# ------------------------------------------------------------

mean_annual133 = (
    annual133.groupby("Model")[
        ["RMSE", "MAE", "R2"]
    ].mean()
)

pooled_rows133 = []

for model in model_keys133:

    pooled_rows133.append({
        "Model": model,
        **metrics133(
            evaluation133["Actual"],
            evaluation133[model]
        )
    })

pooled133 = pd.DataFrame(
    pooled_rows133
).set_index("Model")

print("\nMEAN ANNUAL PERFORMANCE")
print(
    mean_annual133.round(4).to_string()
)

print("\nPOOLED PERFORMANCE (116 COUNTRY-YEARS)")
print(
    pooled133.round(4).to_string()
)

# ------------------------------------------------------------
# 5. Country-clustered paired bootstrap
# ------------------------------------------------------------

rng133 = np.random.default_rng(20261022)

countries133 = np.array(
    sorted(evaluation133["Country"].unique())
)

B133 = 3000

comparisons133 = [
    "BIML_Reduced",
    "BIML_Full"
]

bootstrap_rows133 = []

# Pre-split countries to preserve their temporal blocks.
country_blocks133 = {
    country: evaluation133.loc[
        evaluation133["Country"] == country
    ].copy()
    for country in countries133
}

def mean_annual_metrics133(frame, model):

    values = []

    for year, block in frame.groupby("Year"):

        values.append(
            metrics133(
                block["Actual"],
                block[model]
            )
        )

    return pd.DataFrame(values).mean()

# Improvement definitions:
# RMSE: ML - BIML (positive favors BIML)
# MAE:  ML - BIML (positive favors BIML)
# R2:   BIML - ML (positive favors BIML)

for b in range(B133):

    sampled = rng133.choice(
        countries133,
        size=len(countries133),
        replace=True
    )

    boot = pd.concat(
        [country_blocks133[c] for c in sampled],
        ignore_index=True
    )

    ml = mean_annual_metrics133(boot, "ML")

    for model in comparisons133:

        biml = mean_annual_metrics133(
            boot, model
        )

        bootstrap_rows133.append({
            "Iteration": b,
            "Model": model,
            "RMSE_Improvement": (
                ml["RMSE"] - biml["RMSE"]
            ),
            "MAE_Improvement": (
                ml["MAE"] - biml["MAE"]
            ),
            "R2_Improvement": (
                biml["R2"] - ml["R2"]
            )
        })

bootstrap133 = pd.DataFrame(
    bootstrap_rows133
)

# ------------------------------------------------------------
# 6. Confidence intervals and point estimates
# ------------------------------------------------------------

ci_rows133 = []

for model in comparisons133:

    ml_point = mean_annual133.loc["ML"]
    biml_point = mean_annual133.loc[model]

    point = {
        "RMSE_Improvement": (
            ml_point["RMSE"] - biml_point["RMSE"]
        ),
        "MAE_Improvement": (
            ml_point["MAE"] - biml_point["MAE"]
        ),
        "R2_Improvement": (
            biml_point["R2"] - ml_point["R2"]
        )
    }

    boot = bootstrap133.loc[
        bootstrap133["Model"] == model
    ]

    for metric, estimate in point.items():

        low, high = np.percentile(
            boot[metric].dropna(),
            [2.5, 97.5]
        )

        ci_rows133.append({
            "Model": model,
            "Metric": metric,
            "Point_Estimate": estimate,
            "CI_Lower": low,
            "CI_Upper": high
        })

ci133 = pd.DataFrame(ci_rows133)

print("\nCOUNTRY-CLUSTERED BOOTSTRAP INTERVALS")
print("Positive improvement favors BIML")

print(
    ci133.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# 7. GAM baseline: available historical RMSE
# ------------------------------------------------------------

# Cell 120 reported standalone GAM annual RMSE.
# GAM MAE and R2 require its actual paired predictions.
# Do not infer them from RMSE.

print("\nINDEPENDENT GAM BASELINE — REPORTED RMSE")

gam_rmse133 = pd.Series({
    2019: 24.99834,
    2020: 39.79042,
    2021: 22.10918,
    2022: 16.25955
}, name="GAM_RMSE")

print(gam_rmse133.round(4).to_string())
print(
    "Mean annual GAM RMSE:",
    round(gam_rmse133.mean(), 4)
)

print(
    "\nGAM MAE/R2: not computed here because "
    "paired GAM predictions have not been verified."
)

# ------------------------------------------------------------
# 8. Completion
# ------------------------------------------------------------

print("\nCELL 133 COMPLETE")
print("RMSE, MAE and R2 calculated for ML/BIML")
print("Country-clustered bootstrap completed")
print("GAM retained as independent baseline")
print("No models retrained")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 138 — BIML REGRESSION PERFORMANCE FIGURES
# Observed vs predicted | Residuals | Annual metrics
# No retraining | Development years 2019–2022 only
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

assert "predictions129" in globals()
assert "data125" in globals()
assert "results129" in globals()

out138 = Path("outputs/biml_figures")
out138.mkdir(parents=True, exist_ok=True)

model_keys138 = {
    "Conventional GB": "ML_No_Biology",
    "BIML-Reduced": "BIML_Competence_Index",
    "BIML-Full": "BIML_All_B2"
}

colors138 = {
    "Conventional GB": "#64748b",
    "BIML-Reduced": "#0f766e",
    "BIML-Full": "#2563eb"
}

records138 = []

# Recover predictions without assuming an undocumented structure
for year in [2019, 2020, 2021, 2022]:
    truth = (
        data125.loc[
            data125["Target_Year"] == year,
            "Target_Incidence_t_plus_1"
        ].to_numpy(dtype=float)
    )

    for display_name, key in model_keys138.items():
        pred = np.asarray(
            predictions129[(year, key)],
            dtype=float
        ).reshape(-1)

        assert len(truth) == len(pred), (
            f"Length mismatch: {year}, {key}"
        )

        for observed, predicted in zip(truth, pred):
            records138.append({
                "Year": year,
                "Model": display_name,
                "Observed": observed,
                "Predicted": predicted,
                "Residual": observed - predicted
            })

df138 = pd.DataFrame(records138)

assert df138[["Observed", "Predicted"]].notna().all().all()

metrics138 = []

for (year, model), group in df138.groupby(["Year", "Model"]):
    y = group["Observed"].to_numpy()
    p = group["Predicted"].to_numpy()

    metrics138.append({
        "Year": year,
        "Model": model,
        "RMSE": np.sqrt(mean_squared_error(y, p)),
        "MAE": mean_absolute_error(y, p),
        "R2": r2_score(y, p)
    })

metrics138 = pd.DataFrame(metrics138)

def save138(fig, name):
    for ext in ["png", "pdf"]:
        fig.savefig(
            out138 / f"{name}.{ext}",
            dpi=400,
            bbox_inches="tight"
        )

# ------------------------------------------------------------
# FIGURE 8 — OBSERVED VERSUS PREDICTED
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(16, 5.2))

maximum138 = max(
    df138["Observed"].max(),
    df138["Predicted"].max()
)

for ax, (name, group) in zip(
    axes,
    df138.groupby("Model", sort=False)
):
    ax.scatter(
        group["Observed"],
        group["Predicted"],
        s=26,
        alpha=0.65,
        color=colors138[name],
        edgecolors="none"
    )

    ax.plot(
        [0, maximum138],
        [0, maximum138],
        "--",
        color="black",
        linewidth=1.2
    )

    rmse = np.sqrt(
        mean_squared_error(
            group["Observed"],
            group["Predicted"]
        )
    )

    r2 = r2_score(
        group["Observed"],
        group["Predicted"]
    )

    ax.set_title(
        f"{name}\nRMSE = {rmse:.2f} | R² = {r2:.3f}",
        fontweight="bold"
    )

    ax.set_xlabel("Observed malaria incidence")
    ax.set_ylabel("Predicted malaria incidence")
    ax.set_xlim(0, maximum138 * 1.03)
    ax.set_ylim(0, maximum138 * 1.03)
    ax.set_aspect("equal", adjustable="box")
    ax.grid(alpha=0.15)

fig.suptitle(
    "Observed versus predicted malaria incidence",
    fontsize=15,
    fontweight="bold"
)

fig.tight_layout()
save138(fig, "figure8_observed_vs_predicted")
plt.show()
plt.close(fig)

# ------------------------------------------------------------
# FIGURE 9 — RESIDUAL DISTRIBUTIONS
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))

for ax, name in zip(axes, model_keys138):
    residuals = df138.loc[
        df138["Model"] == name,
        "Residual"
    ]

    ax.hist(
        residuals,
        bins=25,
        color=colors138[name],
        alpha=0.85,
        edgecolor="white"
    )

    ax.axvline(0, color="black", linestyle="--")

    ax.set_title(name, fontweight="bold")
    ax.set_xlabel("Observed − predicted")
    ax.set_ylabel("Frequency")
    ax.grid(axis="y", alpha=0.15)

fig.suptitle(
    "Forecast residual distributions",
    fontsize=15,
    fontweight="bold"
)

fig.tight_layout()
save138(fig, "figure9_residual_distributions")
plt.show()
plt.close(fig)

# ------------------------------------------------------------
# FIGURE 10 — ANNUAL PERFORMANCE
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))

for ax, metric in zip(axes, ["RMSE", "MAE", "R2"]):
    for name in model_keys138:
        subset = (
            metrics138.loc[metrics138["Model"] == name]
            .sort_values("Year")
        )

        ax.plot(
            subset["Year"],
            subset[metric],
            marker="o",
            linewidth=2.2,
            color=colors138[name],
            label=name
        )

    ax.set_title(metric, fontweight="bold")
    ax.set_xlabel("Evaluation year")
    ax.set_ylabel(metric)
    ax.set_xticks([2019, 2020, 2021, 2022])
    ax.grid(alpha=0.2)

handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="lower center",
    ncol=3,
    bbox_to_anchor=(0.5, -0.04),
    frameon=False
)

fig.suptitle(
    "Rolling-origin malaria forecasting performance",
    fontsize=15,
    fontweight="bold"
)

fig.tight_layout(rect=[0, 0.05, 1, 0.94])
save138(fig, "figure10_annual_model_performance")
plt.show()
plt.close(fig)

print("=" * 70)
print("CELL 138 — REGRESSION PERFORMANCE")
print("=" * 70)
print(metrics138.round(4).to_string(index=False))
print("\nPooled metrics")

pooled138 = (
    df138.groupby("Model")
    .apply(
        lambda g: pd.Series({
            "RMSE": np.sqrt(mean_squared_error(
                g["Observed"], g["Predicted"]
            )),
            "MAE": mean_absolute_error(
                g["Observed"], g["Predicted"]
            ),
            "R2": r2_score(
                g["Observed"], g["Predicted"]
            )
        }),
        include_groups=False
    )
)

print(pooled138.round(4))
print("\nCELL 138 COMPLETE")
print("Output:", out138.resolve())
print("No models retrained")
```

```python

# ============================================================
# CELL 139 — BIML FEATURE IMPORTANCE (CORRECTED)
# BIML 2.0 | PUBLICATION FIGURES
# No retraining
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.patches import Patch

assert "models129" in globals()
assert "feature_sets125" in globals()
assert "biological_levels125" in globals()
assert "environment125" in globals()
assert "history125" in globals()

out139 = Path("outputs/biml_figures")
out139.mkdir(parents=True, exist_ok=True)

model_name139 = "BIML_Competence_Index"

features139 = (
    list(feature_sets125["ML"])
    + [
        "B2_competence_supported_mean",
        "B2_index_supported_mean"
    ]
)

assert len(features139) == len(set(features139))

records139 = []

for year in [2019, 2020, 2021, 2022]:

    pipeline = models129[(year, model_name139)]

    # Retrieve the final fitted estimator without
    # assuming a particular pipeline step name.
    if hasattr(pipeline, "steps"):
        estimator = pipeline.steps[-1][1]
    else:
        estimator = pipeline

    assert hasattr(estimator, "feature_importances_"), (
        f"No feature_importances_ for {year}"
    )

    importances = estimator.feature_importances_

    assert len(importances) == len(features139), (
        f"Feature mismatch in {year}: "
        f"{len(importances)} vs {len(features139)}"
    )

    for feature, importance in zip(
        features139,
        importances
    ):

        if feature in biological_levels125:
            group = "Biological"

        elif feature in environment125:
            group = "Environmental"

        elif (
            feature in history125
            or feature == "History_Missing"
        ):
            group = "Historical"

        else:
            group = "Other"

        records139.append({
            "Year": year,
            "Feature": feature,
            "Group": group,
            "Importance": float(importance)
        })

importance139 = pd.DataFrame(records139)

summary139 = (
    importance139.groupby(
        ["Feature", "Group"],
        as_index=False
    )["Importance"]
    .mean()
    .sort_values("Importance", ascending=False)
)

print("=" * 72)
print("CELL 139 — BIML FEATURE IMPORTANCE")
print("=" * 72)

print("\nMEAN FEATURE IMPORTANCE")
print(summary139.round(4).to_string(index=False))

colors139 = {
    "Biological": "#0f766e",
    "Environmental": "#d97706",
    "Historical": "#64748b",
    "Other": "#7c3aed"
}

def save139(fig, name):
    for ext in ["png", "pdf"]:
        fig.savefig(
            out139 / f"{name}.{ext}",
            dpi=400,
            bbox_inches="tight"
        )

# ------------------------------------------------------------
# FIGURE 11 — RANKED FEATURE IMPORTANCE
# ------------------------------------------------------------

plot139 = summary139.sort_values(
    "Importance",
    ascending=True
)

fig, ax = plt.subplots(figsize=(10, 7))

ax.barh(
    plot139["Feature"],
    plot139["Importance"],
    color=[
        colors139[group]
        for group in plot139["Group"]
    ]
)

ax.set_xlabel("Mean impurity-based feature importance")
ax.set_title(
    "BIML-Reduced: relative predictor importance",
    fontweight="bold",
    pad=14
)

ax.legend(
    handles=[
        Patch(facecolor=color, label=group)
        for group, color in colors139.items()
        if group in plot139["Group"].values
    ],
    frameon=False
)

ax.grid(axis="x", alpha=0.2)

fig.tight_layout()
save139(fig, "figure11_biml_feature_importance")
plt.show()
plt.close(fig)

# ------------------------------------------------------------
# FIGURE 12 — FEATURE IMPORTANCE BY YEAR
# ------------------------------------------------------------

matrix139 = importance139.pivot_table(
    index="Feature",
    columns="Year",
    values="Importance"
)

matrix139 = matrix139.reindex(
    summary139["Feature"]
)

fig, ax = plt.subplots(figsize=(9, 8))

im = ax.imshow(
    matrix139.to_numpy(),
    cmap="YlGnBu",
    aspect="auto"
)

ax.set_xticks(range(len(matrix139.columns)))
ax.set_xticklabels(
    matrix139.columns.astype(str)
)

ax.set_yticks(range(len(matrix139)))
ax.set_yticklabels(
    matrix139.index,
    fontsize=9
)

ax.set_xlabel("Evaluation year")
ax.set_title(
    "Temporal stability of BIML feature importance",
    fontweight="bold",
    pad=14
)

cbar = fig.colorbar(
    im,
    ax=ax,
    shrink=0.8
)

cbar.set_label("Impurity-based importance")

fig.tight_layout()
save139(fig, "figure12_feature_importance_by_year")
plt.show()
plt.close(fig)

# ------------------------------------------------------------
# SAVE TABLES
# ------------------------------------------------------------

importance139.to_csv(
    out139 / "feature_importance_by_year.csv",
    index=False
)

summary139.to_csv(
    out139 / "feature_importance_summary.csv",
    index=False
)

print("\nCELL 139 COMPLETE")
print("Feature importance figures generated")
print("Output:", out139.resolve())
print("No models retrained")
```

```python

# ================================================================
# CELL 141 — FINAL BIML PERFORMANCE BY INCIDENCE LEVEL
# BIML 2.0 | FINAL DIAGNOSTIC ANALYSIS
# No retraining | No test targets accessed
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)
from pathlib import Path

assert "predictions129" in globals()
assert "data125" in globals()

out141 = Path("outputs/biml_figures")
out141.mkdir(parents=True, exist_ok=True)

model_map141 = {
    "Conventional GB": "ML_No_Biology",
    "BIML-Reduced": "BIML_Competence_Index",
    "BIML-Full": "BIML_All_B2"
}

# ------------------------------------------------------------
# 1. RECOVER EXISTING PREDICTIONS
# ------------------------------------------------------------

records141 = []

for year in [2019, 2020, 2021, 2022]:

    validation = (
        data125.loc[data125["Target_Year"] == year]
        .copy()
    )

    observed = validation[
        "Target_Incidence_t_plus_1"
    ].to_numpy(dtype=float)

    for label, key in model_map141.items():

        pred = np.asarray(
            predictions129[(year, key)],
            dtype=float
        ).reshape(-1)

        assert len(pred) == len(observed), (
            f"Prediction length mismatch: {year}, {label}"
        )

        for obs, prediction in zip(observed, pred):

            records141.append({
                "Year": year,
                "Model": label,
                "Observed": obs,
                "Predicted": prediction
            })

evaluation141 = pd.DataFrame(records141)

# ------------------------------------------------------------
# 2. DEFINE INCIDENCE STRATA
# ------------------------------------------------------------

# One observation per country-year is used to determine
# the shared incidence thresholds.

reference141 = evaluation141.loc[
    evaluation141["Model"] == "Conventional GB",
    "Observed"
].to_numpy()

q1, q2 = np.quantile(reference141, [1/3, 2/3])

assert q1 < q2, "Incidence quantiles are not distinct."

bins141 = [-np.inf, q1, q2, np.inf]

labels141 = [
    "Low incidence",
    "Medium incidence",
    "High incidence"
]

evaluation141["Incidence_Group"] = pd.cut(
    evaluation141["Observed"],
    bins=bins141,
    labels=labels141,
    include_lowest=True
)

# ------------------------------------------------------------
# 3. CALCULATE PERFORMANCE METRICS
# ------------------------------------------------------------

results141 = []

for (group, model), subset in evaluation141.groupby(
    ["Incidence_Group", "Model"],
    observed=True
):

    y_true = subset["Observed"].to_numpy()
    y_pred = subset["Predicted"].to_numpy()

    results141.append({
        "Incidence_Group": str(group),
        "Model": model,
        "N": len(subset),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, y_pred)
        ),
        "MAE": mean_absolute_error(
            y_true, y_pred
        ),
        "R2": r2_score(
            y_true, y_pred
        )
    })

results141 = pd.DataFrame(results141)

print("=" * 76)
print("CELL 141 — PERFORMANCE BY MALARIA INCIDENCE LEVEL")
print("=" * 76)

print(
    f"\nIncidence tertile boundaries: "
    f"{q1:.3f}, {q2:.3f}"
)

print("\nPERFORMANCE BY INCIDENCE GROUP")
print(
    results141.round(4).to_string(index=False)
)

# ------------------------------------------------------------
# 4. PUBLICATION FIGURE
# ------------------------------------------------------------

colors141 = {
    "Conventional GB": "#64748b",
    "BIML-Reduced": "#0f766e",
    "BIML-Full": "#2563eb"
}

fig, axes = plt.subplots(
    1, 3,
    figsize=(17, 5.5)
)

metrics141 = ["RMSE", "MAE", "R2"]

x = np.arange(len(labels141))
width = 0.25

for ax, metric in zip(axes, metrics141):

    for i, model in enumerate(model_map141):

        subset = (
            results141.loc[
                results141["Model"] == model
            ]
            .set_index("Incidence_Group")
            .reindex(labels141)
        )

        ax.bar(
            x + (i - 1) * width,
            subset[metric].to_numpy(),
            width=width,
            label=model,
            color=colors141[model]
        )

    ax.set_xticks(x)
    ax.set_xticklabels(
        ["Low", "Medium", "High"]
    )

    ax.set_xlabel("Observed malaria incidence level")
    ax.set_ylabel(metric)
    ax.set_title(
        metric.replace("R2", r"$R^2$"),
        fontweight="bold"
    )
    ax.grid(axis="y", alpha=0.15)

fig.suptitle(
    "Malaria forecasting performance across incidence levels",
    fontsize=16,
    fontweight="bold"
)

handles, legends = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    legends,
    loc="lower center",
    bbox_to_anchor=(0.5, -0.07),
    ncol=3,
    frameon=False
)

fig.tight_layout(rect=[0, 0.08, 1, 0.94])

for ext in ["png", "pdf"]:
    fig.savefig(
        out141 / f"figure15_incidence_strata_performance.{ext}",
        dpi=400,
        bbox_inches="tight"
    )

plt.show()
plt.close(fig)

# ------------------------------------------------------------
# 5. SAVE FINAL OUTPUTS
# ------------------------------------------------------------

results141.to_csv(
    out141 / "incidence_strata_performance.csv",
    index=False
)

print("\nCELL 141 COMPLETE")
print("Final incidence-level performance analysis completed")
print("Figure 15 and results table saved")
print("Output:", out141.resolve())
print("No models retrained")
print("No test targets accessed")
print("\nEXPLORATORY ANALYSIS PHASE COMPLETE")
```

Biologically Informed Machine Learning for Malaria Incidence Forecasting: Integrating Mechanistic Vector Traits with Environmental and Epidemiological Predictors

```python

# ============================================================
# CELL 120 — HISTORY-AWARE GAM BASELINE
# BIML 2.0 | ROLLING-ORIGIN GAM EVALUATION
# DEVELOPMENT YEARS 2016–2022 ONLY
# NO BIOLOGICAL FEATURES IN GAM
# NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    SplineTransformer,
    StandardScaler
)
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "trajectory118" in globals()
assert "feature_groups" in globals()

data120 = trajectory118.loc[
    trajectory118["Target_Year"] <= 2022
].copy()

target120 = "Target_Incidence_t_plus_1"

# Historical incidence predictors
history120 = [
    "Incidence_t",
    "Incidence_Change_t"
]

# Conventional environmental predictors
environment120 = list(feature_groups["E"])

gam_features120 = (
    history120 + environment120
)

assert len(gam_features120) == len(
    set(gam_features120)
)

assert set(gam_features120).issubset(
    data120.columns
)

# Each predictor receives its own spline basis.
# Low spline complexity + regularization for small N.
preprocessor120 = ColumnTransformer([
    (
        "splines",
        Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "spline",
                SplineTransformer(
                    n_knots=4,
                    degree=2,
                    knots="quantile",
                    extrapolation="constant",
                    include_bias=False
                )
            ),
            (
                "scaler",
                StandardScaler()
            )
        ]),
        gam_features120
    )
])

years120 = [2019, 2020, 2021, 2022]

results120 = []
predictions120 = {}
models120 = {}

for year in years120:

    train = data120.loc[
        data120["Target_Year"] < year
    ].copy()

    val = data120.loc[
        data120["Target_Year"] == year
    ].copy()

    assert len(train) > 0
    assert len(val) > 0
    assert train["Target_Year"].max() < year

    y_train_delta = (
        train[target120].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    y_val = val[
        target120
    ].to_numpy(dtype=float)

    incidence_val = val[
        "Incidence_t"
    ].to_numpy(dtype=float)

    model = Pipeline([
        ("preprocessor", preprocessor120),
        (
            "regressor",
            Ridge(alpha=20.0)
        )
    ])

    model.fit(
        train[gam_features120],
        y_train_delta
    )

    predicted_delta = model.predict(
        val[gam_features120]
    )

    predicted_incidence = np.maximum(
        incidence_val + predicted_delta,
        0
    )

    results120.append({
        "Year": year,
        "Model": "History_Environment_GAM",
        "Train_N": len(train),
        "Val_N": len(val),
        "MAE": mean_absolute_error(
            y_val,
            predicted_incidence
        ),
        "RMSE": np.sqrt(
            mean_squared_error(
                y_val,
                predicted_incidence
            )
        ),
        "R2": r2_score(
            y_val,
            predicted_incidence
        )
    })

    predictions120[year] = predicted_incidence
    models120[year] = model

results120 = pd.DataFrame(results120)

print("=" * 76)
print("CELL 120 — HISTORY-AWARE GAM BASELINE")
print("=" * 76)

print("\nGAM PREDICTORS")
for feature in gam_features120:
    print(" -", feature)

print("\nROLLING-ORIGIN GAM PERFORMANCE")
print(
    results120.round(5).to_string(index=False)
)

print("\nGAM VS HISTORY-AWARE ML")

comparison120 = (
    results120[
        ["Year", "MAE", "RMSE"]
    ]
    .rename(
        columns={
            "MAE": "GAM_MAE",
            "RMSE": "GAM_RMSE"
        }
    )
    .merge(
        results119.loc[
            results119["Model"] == "History_ML",
            ["Year", "MAE", "RMSE"]
        ].rename(
            columns={
                "MAE": "ML_MAE",
                "RMSE": "ML_RMSE"
            }
        ),
        on="Year",
        validate="one_to_one"
    )
)

comparison120["GAM_RMSE_Improvement"] = (
    comparison120["ML_RMSE"]
    - comparison120["GAM_RMSE"]
)

print(
    comparison120.round(5).to_string(index=False)
)

print("\nCELL 120 COMPLETE")
print("GAM baseline established")
print("No biological predictors used")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 121 — GAM VS BIML: MATCHED COMPARISON
# BIML 2.0 | PRIMARY BASELINE = GAM
# EXISTING ROLLING-ORIGIN RESULTS ONLY
# NO MODEL FITTING | NO TEST DATA ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "results115" in globals()
assert "results119" in globals()
assert "results120" in globals()

# ------------------------------------------------------------
# 1. Collect existing model results
# ------------------------------------------------------------

gam121 = results120.copy()

gam121["Model"] = "GAM"

gam121 = gam121[
    ["Year", "Model", "MAE", "RMSE", "R2"]
]

original121 = results115.loc[
    results115["Model"].isin([
        "ML",
        "BIML_Levels",
        "BIML_Changes",
        "BIML_Levels_Changes"
    ]),
    ["Year", "Model", "MAE", "RMSE", "R2"]
].copy()

history121 = results119.loc[
    results119["Model"].isin([
        "History_ML",
        "History_BIML_Levels",
        "History_BIML_Changes"
    ]),
    ["Year", "Model", "MAE", "RMSE", "R2"]
].copy()

comparison121 = pd.concat(
    [gam121, original121, history121],
    ignore_index=True
)

# ------------------------------------------------------------
# 2. Validate matched evaluation years
# ------------------------------------------------------------

expected_years121 = {2019, 2020, 2021, 2022}

for name, group in comparison121.groupby("Model"):

    assert set(group["Year"]) == expected_years121, (
        f"Missing evaluation years for {name}"
    )

    assert not group["Year"].duplicated().any(), (
        f"Duplicate evaluation years for {name}"
    )

# ------------------------------------------------------------
# 3. RMSE comparison
# ------------------------------------------------------------

rmse121 = comparison121.pivot(
    index="Year",
    columns="Model",
    values="RMSE"
)

mae121 = comparison121.pivot(
    index="Year",
    columns="Model",
    values="MAE"
)

r2_121 = comparison121.pivot(
    index="Year",
    columns="Model",
    values="R2"
)

print("=" * 78)
print("CELL 121 — GAM VS BIML")
print("=" * 78)

print("\nROLLING-ORIGIN RMSE — LOWER IS BETTER")
print(
    rmse121.round(4).to_string()
)

print("\nROLLING-ORIGIN MAE — LOWER IS BETTER")
print(
    mae121.round(4).to_string()
)

print("\nROLLING-ORIGIN R2 — HIGHER IS BETTER")
print(
    r2_121.round(4).to_string()
)

# ------------------------------------------------------------
# 4. Improvement relative to GAM
# ------------------------------------------------------------

gam_rmse121 = rmse121["GAM"]

improvement121 = pd.DataFrame(index=rmse121.index)

for model in rmse121.columns:

    if model == "GAM":
        continue

    improvement121[model] = (
        (gam_rmse121 - rmse121[model])
        / gam_rmse121
    ) * 100

print("\nRMSE IMPROVEMENT RELATIVE TO GAM (%)")
print("Positive = model outperforms GAM")

print(
    improvement121.round(2).to_string()
)

# ------------------------------------------------------------
# 5. Summary across evaluation years
# ------------------------------------------------------------

summary121 = comparison121.groupby("Model").agg(
    Mean_MAE=("MAE", "mean"),
    Mean_RMSE=("RMSE", "mean"),
    Mean_R2=("R2", "mean")
)

summary121["Years_Better_Than_GAM"] = [
    (
        int((rmse121[name] < gam_rmse121).sum())
        if name != "GAM"
        else np.nan
    )
    for name in summary121.index
]

summary121 = summary121.sort_values(
    "Mean_RMSE"
)

print("\nFOUR-YEAR SUMMARY")
print(
    summary121.round(4).to_string()
)

# ------------------------------------------------------------
# 6. Identify yearly RMSE winners
# ------------------------------------------------------------

print("\nBEST MODEL BY EVALUATION YEAR")

for year in sorted(expected_years121):

    winner = rmse121.loc[year].idxmin()

    print(
        f"{year}: {winner} "
        f"(RMSE={rmse121.loc[year, winner]:.4f})"
    )

print("\nCELL 121 COMPLETE")
print("Primary baseline: GAM")
print("No models retrained")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 122 — GAM ERROR VS BIOLOGICAL INFORMATION
# BIML 2.0 | BIOLOGICAL RESIDUAL DIAGNOSTIC
# DEVELOPMENT YEARS 2019–2022 ONLY
# NO MODEL FITTING | NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

assert "trajectory118" in globals()
assert "predictions120" in globals()

bio122 = [
    "B2_competence_supported_mean",
    "B2_EIP_supported_mean",
    "B2_biting_supported_mean",
    "B2_index_supported_mean",
    "B2_competence_season_mean_change",
    "B2_EIP_season_mean_change",
    "B2_biting_season_mean_change",
    "B2_index_season_mean_change"
]

rows122 = []

for year in [2019, 2020, 2021, 2022]:

    subset = trajectory118.loc[
        trajectory118["Target_Year"] == year
    ].copy()

    actual = subset[
        "Target_Incidence_t_plus_1"
    ].to_numpy(dtype=float)

    gam_prediction = np.asarray(
        predictions120[year],
        dtype=float
    )

    assert len(actual) == len(gam_prediction)

    gam_error = actual - gam_prediction

    for feature in bio122:

        values = subset[
            feature
        ].to_numpy(dtype=float)

        valid = (
            np.isfinite(values)
            & np.isfinite(gam_error)
        )

        if (
            valid.sum() >= 3
            and np.std(values[valid]) > 0
            and np.std(gam_error[valid]) > 0
        ):
            correlation = np.corrcoef(
                values[valid],
                gam_error[valid]
            )[0, 1]
        else:
            correlation = np.nan

        rows122.append({
            "Year": year,
            "Feature": feature,
            "N": int(valid.sum()),
            "Correlation_With_GAM_Error": correlation
        })

results122 = pd.DataFrame(rows122)

print("=" * 76)
print("CELL 122 — GAM ERRORS VS BIOLOGICAL FEATURES")
print("=" * 76)

print("\nCORRELATIONS BY EVALUATION YEAR")

print(
    results122.pivot(
        index="Feature",
        columns="Year",
        values="Correlation_With_GAM_Error"
    ).round(4).to_string()
)

print("\nCORRELATION SIGN CONSISTENCY")

sign122 = results122.pivot(
    index="Feature",
    columns="Year",
    values="Correlation_With_GAM_Error"
)

summary122 = pd.DataFrame({
    "Mean_Correlation": sign122.mean(axis=1),
    "Mean_Abs_Correlation": sign122.abs().mean(axis=1),
    "Positive_Years": (sign122 > 0).sum(axis=1),
    "Negative_Years": (sign122 < 0).sum(axis=1)
})

print(
    summary122.sort_values(
        "Mean_Abs_Correlation",
        ascending=False
    ).round(4).to_string()
)

print("\nCELL 122 COMPLETE")
print("Descriptive diagnostic only")
print("No new models fitted")
print("No test data accessed")
```

```python

# ============================================================
# CELL 123 — LEAKAGE-SAFE GAM OUT-OF-FOLD RESIDUALS
# BIML 2.0 | GAM-ANCHORED CORRECTION ARCHITECTURE
# DEVELOPMENT DATA ONLY: TARGET YEARS <= 2022
# NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    SplineTransformer,
    StandardScaler
)
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

assert "trajectory118" in globals()
assert "feature_groups" in globals()
assert "results120" in globals()
assert "predictions120" in globals()

data123 = trajectory118.loc[
    trajectory118["Target_Year"] <= 2022
].copy()

# Preserve original row identity for subsequent experiments.
assert data123.index.is_unique
assert data123["Target_Year"].between(2016, 2022).all()

target123 = "Target_Incidence_t_plus_1"

gam_features123 = [
    "Incidence_t",
    "Incidence_Change_t"
] + list(feature_groups["E"])

# Reproduce the Cell 120 GAM specification exactly.
def build_gam123():

    preprocessor = ColumnTransformer([
        (
            "splines",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="median")
                ),
                (
                    "spline",
                    SplineTransformer(
                        n_knots=4,
                        degree=2,
                        knots="quantile",
                        extrapolation="constant",
                        include_bias=False
                    )
                ),
                (
                    "scaler",
                    StandardScaler()
                )
            ]),
            gam_features123
        )
    ])

    return Pipeline([
        ("preprocessor", preprocessor),
        ("regressor", Ridge(alpha=20.0))
    ])


# ------------------------------------------------------------
# 1. Generate expanding-window out-of-fold predictions
# ------------------------------------------------------------

oof_records123 = []
oof_models123 = {}

# Start in 2018: train on 2016–2017.
# Earlier years have insufficient historical training coverage.
for year in range(2018, 2023):

    train = data123.loc[
        data123["Target_Year"] < year
    ].copy()

    holdout = data123.loc[
        data123["Target_Year"] == year
    ].copy()

    assert len(train) > 0
    assert len(holdout) == 29

    assert train["Target_Year"].max() < year

    y_train_delta = (
        train[target123].to_numpy(dtype=float)
        - train["Incidence_t"].to_numpy(dtype=float)
    )

    model = build_gam123()

    model.fit(
        train[gam_features123],
        y_train_delta
    )

    delta_pred = model.predict(
        holdout[gam_features123]
    )

    incidence_pred = np.maximum(
        holdout["Incidence_t"].to_numpy(dtype=float)
        + delta_pred,
        0
    )

    actual = holdout[
        target123
    ].to_numpy(dtype=float)

    residual = actual - incidence_pred

    block = pd.DataFrame({
        "Row_ID": holdout.index.to_numpy(),
        "Country": holdout["Country"].to_numpy(),
        "Target_Year": year,
        "GAM_OOF_Prediction": incidence_pred,
        "GAM_OOF_Residual": residual,
        "Actual_Incidence": actual
    })

    oof_records123.append(block)

    oof_models123[year] = model


oof123 = pd.concat(
    oof_records123,
    ignore_index=True
)

assert not oof123["Row_ID"].duplicated().any()

assert len(oof123) == 5 * 29

# ------------------------------------------------------------
# 2. Verify GAM reproduction for 2019–2022
# ------------------------------------------------------------

reproduction123 = []

for year in [2019, 2020, 2021, 2022]:

    reproduced = oof123.loc[
        oof123["Target_Year"] == year,
        "GAM_OOF_Prediction"
    ].to_numpy(dtype=float)

    original = np.asarray(
        predictions120[year],
        dtype=float
    )

    max_difference = np.max(
        np.abs(reproduced - original)
    )

    reproduction123.append({
        "Year": year,
        "Max_Absolute_Difference": max_difference
    })

reproduction123 = pd.DataFrame(
    reproduction123
)

assert (
    reproduction123["Max_Absolute_Difference"] < 1e-8
).all(), "GAM OOF predictions do not reproduce Cell 120"

# ------------------------------------------------------------
# 3. Evaluate historical GAM out-of-fold errors
# ------------------------------------------------------------

diagnostics123 = []

for year, block in oof123.groupby("Target_Year"):

    y_true = block[
        "Actual_Incidence"
    ].to_numpy(dtype=float)

    y_pred = block[
        "GAM_OOF_Prediction"
    ].to_numpy(dtype=float)

    diagnostics123.append({
        "Year": year,
        "N": len(block),
        "MAE": mean_absolute_error(
            y_true, y_pred
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, y_pred)
        ),
        "Mean_Residual": np.mean(
            y_true - y_pred
        ),
        "Residual_SD": np.std(
            y_true - y_pred,
            ddof=1
        )
    })

diagnostics123 = pd.DataFrame(
    diagnostics123
)

# ------------------------------------------------------------
# 4. Check available correction-training sample sizes
# ------------------------------------------------------------

sample_sizes123 = []

for evaluation_year in [2019, 2020, 2021, 2022]:

    available = oof123.loc[
        oof123["Target_Year"] < evaluation_year
    ]

    sample_sizes123.append({
        "Evaluation_Year": evaluation_year,
        "Correction_Train_N": len(available),
        "Historical_OOF_Years": (
            ", ".join(
                map(
                    str,
                    sorted(
                        available["Target_Year"].unique()
                    )
                )
            )
        )
    })

sample_sizes123 = pd.DataFrame(
    sample_sizes123
)

# ------------------------------------------------------------
# 5. Report
# ------------------------------------------------------------

print("=" * 76)
print("CELL 123 — LEAKAGE-SAFE GAM OOF RESIDUALS")
print("=" * 76)

print("\nOUT-OF-FOLD DATASET")
print("Rows:", len(oof123))
print("Countries:", oof123["Country"].nunique())
print(
    "Years:",
    sorted(oof123["Target_Year"].unique())
)

print("\nGAM REPRODUCTION CHECK")
print(
    reproduction123.round(10).to_string(index=False)
)

print("\nHISTORICAL OUT-OF-FOLD GAM PERFORMANCE")
print(
    diagnostics123.round(5).to_string(index=False)
)

print("\nAVAILABLE CORRECTION-TRAINING DATA")
print(
    sample_sizes123.to_string(index=False)
)

print("\nRESIDUAL DISTRIBUTION")
print(
    oof123["GAM_OOF_Residual"]
    .describe()
    .round(5)
    .to_string()
)

print("\nCELL 123 COMPLETE")
print("Historical GAM OOF residuals generated")
print("No biological correction model fitted")
print("No test targets accessed")
```

```python

# ============================================================
# CELL 124 — ROBUST GAM-ANCHORED BIML
# BIML 2.0 | MATCHED BIOLOGICAL CORRECTION
# DEVELOPMENT YEARS 2019–2022 ONLY
# NO TEST TARGET ACCESS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import HuberRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

assert "data123" in globals()
assert "oof123" in globals()
assert "predictions120" in globals()

# ------------------------------------------------------------
# 1. Correction feature sets
# ------------------------------------------------------------

conventional124 = [
    "ERA5_T2M_C_mean",
    "Rainfall_total_mm"
]

biological124 = [
    "B2_biting_supported_mean",
    "B2_EIP_supported_mean"
]

feature_sets124 = {
    "GAM_ML_Correction": conventional124,
    "GAM_BIML_Correction": (
        conventional124 + biological124
    )
}

assert all(
    set(features).issubset(data123.columns)
    for features in feature_sets124.values()
)

# ------------------------------------------------------------
# 2. Attach OOF residuals to their original rows
# ------------------------------------------------------------

correction_data124 = data123.copy()

oof_lookup124 = oof123.set_index("Row_ID")[
    ["GAM_OOF_Residual", "GAM_OOF_Prediction"]
]

correction_data124 = correction_data124.join(
    oof_lookup124,
    how="left",
    validate="one_to_one"
)

assert correction_data124.loc[
    correction_data124["Target_Year"] >= 2018,
    "GAM_OOF_Residual"
].notna().all()

# ------------------------------------------------------------
# 3. Rolling-origin evaluation
# ------------------------------------------------------------

results124 = []
predictions124 = {}
models124 = {}

MAX_CORRECTION124 = 15.0

for year in [2019, 2020, 2021, 2022]:

    # Only historical OOF residuals can train correction.
    train = correction_data124.loc[
        (correction_data124["Target_Year"] >= 2018)
        & (correction_data124["Target_Year"] < year)
    ].copy()

    val = correction_data124.loc[
        correction_data124["Target_Year"] == year
    ].copy()

    assert len(train) > 0
    assert train["Target_Year"].max() < year

    gam_prediction = np.asarray(
        predictions120[year],
        dtype=float
    )

    actual = val[
        "Target_Incidence_t_plus_1"
    ].to_numpy(dtype=float)

    # Include unchanged GAM baseline.
    candidates = {
        "GAM": gam_prediction.copy()
    }

    for name, features in feature_sets124.items():

        model = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            ),
            (
                "regressor",
                HuberRegressor(
                    epsilon=1.35,
                    alpha=20.0,
                    max_iter=1000
                )
            )
        ])

        model.fit(
            train[features],
            train["GAM_OOF_Residual"]
        )

        raw_correction = model.predict(
            val[features]
        )

        bounded_correction = np.clip(
            raw_correction,
            -MAX_CORRECTION124,
            MAX_CORRECTION124
        )

        corrected_prediction = np.maximum(
            gam_prediction + bounded_correction,
            0
        )

        candidates[name] = corrected_prediction
        models124[(year, name)] = model

    for name, prediction in candidates.items():

        results124.append({
            "Year": year,
            "Model": name,
            "Train_N": (
                0 if name == "GAM" else len(train)
            ),
            "Val_N": len(val),
            "MAE": mean_absolute_error(
                actual,
                prediction
            ),
            "RMSE": np.sqrt(
                mean_squared_error(
                    actual,
                    prediction
                )
            ),
            "R2": r2_score(
                actual,
                prediction
            )
        })

        predictions124[(year, name)] = prediction.copy()

results124 = pd.DataFrame(results124)

# ------------------------------------------------------------
# 4. Compare models
# ------------------------------------------------------------

rmse124 = results124.pivot(
    index="Year",
    columns="Model",
    values="RMSE"
)

mae124 = results124.pivot(
    index="Year",
    columns="Model",
    values="MAE"
)

improvement124 = pd.DataFrame(
    index=rmse124.index
)

for name in feature_sets124:

    improvement124[name] = (
        rmse124["GAM"] - rmse124[name]
    )

# ------------------------------------------------------------
# 5. Verify GAM baseline unchanged
# ------------------------------------------------------------

baseline_difference124 = (
    rmse124["GAM"]
    - results120.set_index("Year")["RMSE"]
).abs().max()

assert baseline_difference124 < 1e-8

# ------------------------------------------------------------
# 6. Output
# ------------------------------------------------------------

print("=" * 76)
print("CELL 124 — GAM-ANCHORED BIML")
print("=" * 76)

print("\nCORRECTION FEATURE SETS")
for name, features in feature_sets124.items():
    print(name, ":", features)

print("\nROLLING-ORIGIN RMSE")
print(
    rmse124.round(5).to_string()
)

print("\nROLLING-ORIGIN MAE")
print(
    mae124.round(5).to_string()
)

print("\nRMSE IMPROVEMENT OVER GAM")
print("Positive values indicate improvement")
print(
    improvement124.round(5).to_string()
)

print("\nYEARS BETTER THAN GAM")
print(
    (improvement124 > 0).sum().to_string()
)

print("\nMEAN ANNUAL RMSE")
print(
    rmse124.mean()
    .sort_values()
    .round(5)
    .to_string()
)

print("\nCELL 124 COMPLETE")
print("GAM-anchored BIML evaluated")
print("Historical OOF residuals used for correction training")
print("No test targets accessed")
```
